# Введение

На прошлых двух семинарах мы уделили все время квадратурным формулам интерполяционного типа, обсудили сами квадратурные формулы, способы их уточнения, ошибку при вычислении по этим квадратурным формулам. На этом семинаре мы будем обсуждать второй класс методов численного интегрирования - квадратурные формулы Гаусса. Пусть необходимо вычислить интеграл

$$
I= \int \limits _a^b f(x) dx.
$$

Считаем, что оба предела интегрирования $a$ и $b$ конечны, а подынтегральная функция не имеет особенностей на отрезке $[a,b]$ (интеграл рассматривается в собственном смысле).

# Квадратурные формулы Гаусса

Для начала предложим способ приближенного вычисления интеграла на отрезке $[-1;1]$. Напомним, что $x_i$ - узлы квадратурной формулы, $c_i$ - веса квадратурной формулы. Формула примет вид


$$
\int \limits _{-1}^{1} f(x) dx \approx \sum \limits _{i=1}^n c_i f(x_i).
$$

Веса $c_i$ и узлы $x_i$ выбираются так, чтобы квадратурная формула была точна для многочленов степени не выше $2n - 1$. Справедлива следующая теорема


**Теорема.** Если в качестве узлов квадратурной формулы взять нули
полиномов Лежандра $q_n(x)$:  
$$ q_n(x_i)=0,\quad i=1,\ldots,n, $$
а в качестве весов — интегралы от базисных функций многочленов Лагранжа
$l_i(x)$ степени $n-1$:
$$
c_i=\int_{-1}^{1}
\frac{(x-x_1)(x-x_2)\cdots(x-x_{i-1})(x-x_{i+1})\cdots(x-x_n)}
{(x_i-x_1)(x_i-x_2)\cdots(x_i-x_{i-1})(x_i-x_{i+1})\cdots(x_i-x_n)}
\, dx,
$$
то квадратурная формула будет точна при подстановке в неё вместо $f(x)$ любого
многочлена степени не выше $2n-1$.


Остается вопрос, как пользоваться предложенной квадратурной формулой, если интеграл требуется вычислить на произвольном конечном отрезке $[a;b]$. В таком случае

$$
\int \limits_{a}^{b} f(x)\,dx
= \frac{b-a}{2} \sum_{i=1}^{n} c_i \,
f\!\left( \frac{a+b}{2} + \frac{b-a}{2}\, t_i \right),
$$

где $c_i$, $t_i$ — веса и узлы на $[-1,1]$.


Для оценки ошибки интегрирования по формулам Гаусса можно воспользоваться формулой

$$
\Delta_n
= \frac{(b-a)^{2n+1}\,(n!)^4}{(2n+1)\,\bigl((2n)!\bigr)^3}\,
f^{(2n)}(\xi), \qquad \xi \in [a,b].
$$


Узлы и веса квадратурной формулы Гаусса нет необходимости вычислять вручную, их можно найти в готовых таблицах. Мы приведем небольшую таблицу, которой будет достаточно в учебных целях

$$
\begin{array}{c|c|c|c}
n & c_i\ - \text{веса на}\ [-1,1] & t_i\ -  \text{узлы на}\ [-1,1] & \Delta_n \\ \hline
2 & c_{1,2}=1 &
t_{1,2}=\pm\frac{\sqrt{3}}{3} &
\dfrac{(b-a)^5}{2^5}\cdot\dfrac{1}{135}\,\max|f^{(4)}| \\[6pt] \hline
3 &
c_{1,3}=\frac{5}{9},\quad c_2=\frac{8}{9} &
t_{1,3}=\pm\sqrt{\frac{3}{5}},\quad t_2=0 &
\dfrac{(b-a)^7}{2^7}\cdot\dfrac{1}{15750}\,\max|f^{(6)}| \\[6pt] \hline
4 &
c_{1,4}=0.347855,\quad c_{2,3}=0.652145 &
t_{1,4}=\pm 0.861136,\quad t_{2,3}=\pm 0.339981 &
\dfrac{(b-a)^9}{2^9}\cdot\dfrac{2}{3472875}\,\max|f^{(8)}|
\end{array}
$$

## Задача 1

Вычислите интеграл по формулам Гаусса с точностью $ \varepsilon = 10^{-3} $

$$
I=\int \limits_{-1}^{1} \ln\!\left(1+\frac{x^{2}}{3}\right)\,dx
$$



**Решение.** В первую очередь заметим, что подынтегральная функция является четной, тогда верно, что
$$
I = 2 \int_{0}^{1} \ln\!\left(1+\frac{x^{2}}{3}\right)\,dx.
$$

Начнем с определения числа узлов, которые необходимо задать для вычисления с заданной точностью. Для начала проверим вариант $n=2$

$$\Delta_{2} = \dfrac{(b-a)^5}{2^5}\cdot\dfrac{1}{135}\,\max|f^{(4)}|$$

$$
\Delta_{2} = \frac{1}{2^{5}}\cdot \frac{1}{135}\cdot \frac{4}{3} \approx 3\cdot 10^{-4} < \frac{\varepsilon}{2}
\;\Rightarrow\; \text{достаточно}
$$

Константа $M_4$ может быть найдена различными способами, в прошлом семинаре мы получили формулу

$$
f(x)=\ln\!\left(1+\frac{x^2}{a^2}\right),\qquad
\bigl|f^{(n)}(x)\bigr|\le \frac{2\,(n-1)!}{a^{\,n}},
$$
откуда $M_4=4/3$

Если бы вариант с двумя узлами не подошел, мы бы последовательно увеличивали бы $n$, пока не нашли бы подходящий вариант.

Теперь осталось записать квадратурную формулу

$$
I \approx 2 \cdot \frac{b-a}{2}\,(c_{1} f(x_{1}) + c_{2} f(x_{2}))
$$

Узлы и веса берем из таблицы
$$
c_{1} = c_{2} = 1,\qquad
t_{1,2} = \pm \frac{\sqrt{3}}{3}
$$

$$
x_{1,2} = \frac12 + \frac12\, t_{1,2}
       = \frac12 \pm \frac{\sqrt{3}}{6}
$$
В итоге получаем
$$
I = f(x_{1}) + f(x_{2})
   = \ln\!\left(1+\frac{x_{1}^{2}}{3}\right)
   + \ln\!\left(1+\frac{x_{2}^{2}}{3}\right)
   = 0.2031928
$$
Точное значение интеграла
$$
I^{*} = 0.2029629
$$
хорошо совпадает с полученным в результате вычислений по квадратурной формуле

# Квадратурные формулы Гаусса-Кристоффеля

Квадратурные формулы Гаусса-Кристоффеля позволяют вычислять интеграл с известной весовой функцией $p(x)$ как на конечном отрезке $[a,b]$, так и в пределах $(-\infty; +\infty)$. Формула мало отличается от предложенной выше, ее узлы и веса выбираются по похожему принципу

$$
\int \limits_{a}^{b} f(x)\,p(x)\,dx \;=\; \sum_{i=1}^{n} c_{i}\, f(x_{i})
$$


**Теорема.** Квадратурная формула точна для многочленов степени не выше $2n-1$, если её узлами $x_i$ являются корни многочлена $Q_n(x)$ из семейства многочленов, ортогональных на промежутке интегрирования $(a,b)$ с весом $p(x)$, а весовыми коэффициентами являются числа

$$
c_i = \int_{a}^{b} l_i(x)\, p(x)\, dx.
$$


## Задача 2

Постройте формулу для вычисления интеграла по квадратурным формулам Гаусса-Кристоффеля с двумя узлами.

$$
I(f) = \int \limits _{-\infty}^{+\infty} f(x)\,\exp(-x^{2})\,dx
$$


**Решение.** Так как в условии сказано использовать формулу с двумя узлами, то ответ ищем в виде

$$
I = c_1\,f(x_1) + c_2\,f(x_2)
$$

Формула должна быть точна для многочленов
$$
P_{0}(x),\, P_{1}(x),\, P_{2}(x),\, P_{3}(x)
$$

Выпишем эти условия. Сначала подставим многочлен нулевой степени

$$
P_{0}(x)=1,\qquad
I=\int\limits_{-\infty}^{+\infty} 1\cdot e^{-x^{2}}\,dx=\sqrt{\pi},
$$
$$
I=c_{1}\cdot 1+c_{2}\cdot 1 \;\Longrightarrow\; c_{1}+c_{2}=\sqrt{\pi}.
$$

Далее многочлен первой степени
$$
P_{1}(x)=x,\qquad
I=\int\limits_{-\infty}^{+\infty} x e^{-x^{2}}\,dx=0,
$$
$$
I=c_{1}x_{1}+c_{2}x_{2} \;\Longrightarrow\; c_{1}x_{1}+c_{2}x_{2}=0.
$$

Далее многочлен второй степени
$$
P_{2}(x)=x^{2},\qquad
I=\int\limits_{-\infty}^{+\infty} x^{2} e^{-x^{2}}\,dx=\frac{\sqrt{\pi}}{2},
$$
$$
I=c_{1}x_{1}^{2}+c_{2}x_{2}^{2} \;\Longrightarrow\;
c_{1}x_{1}^{2}+c_{2}x_{2}^{2}=\frac{\sqrt{\pi}}{2}.
$$
И многочлен третьей степени
$$
P_{3}(x)=x^{3},\qquad
I=\int\limits_{-\infty}^{+\infty} x^{3} e^{-x^{2}}\,dx=0,
$$
$$
I=c_{1}x_{1}^{3}+c_{2}x_{2}^{3} \;\Longrightarrow\;
c_{1}x_{1}^{3}+c_{2}x_{2}^{3}=0.
$$

Записываем получившуюся СЛАУ и решаем

$$
\begin{cases}
c_{1}+c_{2}=\sqrt{\pi}, \\[6pt]
c_{1}x_{1}+c_{2}x_{2}=0, \\[6pt]
c_{1}x_{1}^{2}+c_{2}x_{2}^{2}=\dfrac{\sqrt{\pi}}{2}, \\[6pt]
c_{1}x_{1}^{3}+c_{2}x_{2}^{3}=0
\end{cases}
$$
Воспользуемся соображениями симметрии, тогда положим

$$c_1 = c_2 = c, \quad x_1 = -x_2 $$

Из первого уравнения найдем
$$c = \frac{\sqrt{\pi}}{2}$$
А из третьего
$$2 \cdot \frac{\sqrt{\pi}}{2} \, x_1^2 = \frac{\sqrt{\pi}}{2}
\quad\Rightarrow\quad
x_{1,2} = \pm \frac{1}{\sqrt{2}}$$

В итоге получаем
$$
I(f) = \frac{\sqrt{\pi}}{2}\,\bigl(f(\tfrac{1}{\sqrt{2}}) + f(-\tfrac{1}{\sqrt{2}})\bigr)
$$



# Интеграл от быстро осциллирующей функции

На практике можно столкнуться с интегралами, в которых подынтегральная функция совершает множество "осцилляций" на отрезке интегрирования. Это интегралы вида

$$
I = \int\limits_a^b f(x)\sin(\omega x)\,dx, \qquad \omega(b-a)\gg 1
$$

или

$$
I = \int\limits_a^b f(x)\cos(\omega x)\,dx, \qquad I = \int\limits_a^b f(x)\exp(i\omega x)\,dx,
$$

В таком случае вычисление по предложенным выше квадратурным формулам как интерполяционного типа, так и по квадратурным формулам Гаусса нецелесообразно. Поступим иначе. Построим $\tilde f(x)$ - аппроксимацию $ f(x)$, например сплайном. Тогда

$$\int\limits_a^b \tilde f(x)e^{i\omega x}dx$$

интегрируется аналитически. Ошибка интегрирования

$$
\Delta = \left| \int \limits_a^b (f(x) - \tilde f(x)) e^{i\omega x}\, dx \right|.
$$

Ошибка интегрирования на каждом отрезке

$$
\Delta_n = \left| \int_{x_n}^{x_{n+1}} f(x)e^{i\omega x}\,dx - \int_{x_n}^{x_{n+1}} \tilde S_3(x,n)e^{i\omega x}\,dx \right|
\le
\left| \int_{x_n}^{x_{n+1}} R_n(x)e^{i\omega x}\,dx \right|
\le
\int_{x_n}^{x_{n+1}} |R_n(x)|\,dx,
$$

где $ R_n(x)$ - остаточный член интерполяции.

# Несобственный интеграл

Рассмотрим подходы, позволяюшие вычислить несобственные интегралы с заданной точностью. Конечно, задача имеет смысл, только если интеграл сходится абсолютно. Для начала рассмотрим несколько простых подходов на примере интеграла

$$
I = \int\limits_{0}^{1} \frac{\cos x}{\sqrt{x}}\,dx
$$

1) Замена переменной

$$
x = t^{2}, \qquad I = 2 \int\limits_{0}^{1} \cos(t^{2})\, dt.
$$

Полученный интеграл является собственным.

2) Интегрирование по частям

$$
I = \int\limits_{0}^{1} \frac{\cos x}{\sqrt{x}}\, dx
    = 2 \cos x\, \sqrt{x}\Big|_{0}^{1}
    + 2 \int\limits_{0}^{1} \sqrt{x}\, \sin x\, dx
$$

$$
= 2\cos 1
 + 2\sin x \cdot \frac{2}{3} x^{3/2}\Big|_{0}^{1}
 - 2 \int\limits_{0}^{1} \cos x \cdot \frac{2}{3} x^{3/2}\, dx
$$

После первого интегрирования по частям полученный интеграл уже является собственным, однако вторая производная подынтегральной функции неограничена, поэтому интегрирование по частям выполняется еще раз.

3) Метод явного выделения особенности
$$
I=\int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx
    =\int\limits_{0}^{\delta}\frac{\cos x}{\sqrt{x}}\,dx
    +\int\limits_{\delta}^{1}\frac{\cos x}{\sqrt{x}}\,dx
$$

Первый интеграл $I_{1}$ вычисляется приближённо аналитически, второй интеграл $I_{2}$ — численно.

4) Метод Канторовича

$$
\int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx
=
\int\limits_{0}^{1}\frac{\cos x - 1}{\sqrt{x}}\,dx
    +\int\limits_{0}^{1}\frac{dx}{\sqrt{x}}
$$

Первый интеграл не содержит особенности и находится численно,  
второй интеграл легко вычисляется аналитически.



## Задача 3

Предложить алгоритм, как пользуясь программой, реализующей метод трапеций для численного интегрирования регулярных функций с произвольным шагом, вычислить следующий интеграл с точностью $\varepsilon=10^{-3}$ (использовать метод явного выделения особенности)

$$
I= \int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx$$

**Решение.**

$$
I= \int\limits_{0}^{1}\frac{\cos x}{\sqrt{x}}\,dx
=
\underbrace{\int\limits_{0}^{\delta}\frac{\cos x}{\sqrt{x}}\,dx}_{I_{1}}
+
\underbrace{\int\limits_{\delta}^{1}\frac{\cos x}{\sqrt{x}}\,dx}_{I_{2}}
$$

Интеграл $I_{1}$ вычисляется приближённо аналитически с точностью $\varepsilon/2$. Интеграл $I_{2}$ находится численно с точностью  $\varepsilon/2$.


*Вариант 1.*

1) подсчёт $I_{1}$, $\delta = ?$

Пусть $\cos x \approx 1$, первое отброшенное слагаемое $\frac{x^{2}}{2}$. Интеграл $I_{1}$ приближенно вычисляется следующим образом

$$
I_{1} = \int\limits_{0}^{\delta}\frac{1}{\sqrt{x}}\,dx = 2\sqrt{\delta}
$$

Оценим ошибку
$$
\Delta =
\int\limits_{0}^{\delta} \frac{x^{2}}{2\sqrt{x}}\,dx
= \frac12 \int\limits_{0}^{\delta} x^{3/2}\,dx
= \frac{\delta^{5/2}}{5}
$$

Потребуем, чтобы ошибка не превышала $\frac{\varepsilon}{2}$

$$
\Delta \le \frac{\varepsilon}{2}
$$

и найдем $\delta$

$$
\frac{\delta^{5/2}}{5} \le \frac{\varepsilon}{2}
$$

$$
\frac{\delta^{5/2}}{5} = \frac12\cdot 10^{-3}
$$

$$
\delta \le 9.1\cdot 10^{-2}
$$

2) подсчёт $I_{2}$, $h = ?$

$$
I_{2} = \int\limits_{\delta}^{1} \frac{\cos x}{\sqrt{x}} \, dx,
\qquad
\max |f''| = M_{2} = 299.
$$

Потребуем, чтобы погрешность вычисления интеграла по формуле трапеций не превышала  $\frac{\varepsilon}{2}$:
$$
\frac1{12} \max |f''(x)| (b-a)\, h^{2} \le \frac{\varepsilon}{2},
$$
то есть
$$
\frac1{12} \cdot 299 \cdot (1-\delta)\, h^{2} \le \frac12 \cdot 10^{-3},
$$
откуда
$$
h \le 4.45\cdot 10^{-3}.
$$

Найдем количество интервалов равномерной сетки

$$
\frac{b-a}{h} = \frac{1-\delta}{4.45\cdot10^{-3}} \approx 220.2
\quad\Rightarrow\quad
N = 221,\qquad
h = \frac{1-\delta}{221}.
$$

Итог:

$$
I = 2\sqrt{\delta} + I_{2}, \qquad
\delta = 9.1\cdot10^{-2},
$$

в программу передаём

$$
f(x) = \frac{\cos x}{\sqrt{x}},\qquad
a = 9.1\cdot10^{-2},\quad
b = 1,\quad
N = 221.
$$

Приближение $\cos x \approx 1$ является довольно грубым, поэтому $\delta = 9.1\cdot10^{-2}$ довольно мало, а также из-за малости $\delta$ оказывается довольно большой константа $M_2$. Это в свою очередь приводит к сравнительному большому количеству точек $N+1 = 222$. Рассмотрим другой вариант вычисления интеграла $I_{1}$.


*Вариант 2.*

1) подсчёт $ I_{1},\ \delta - ?$

Теперь положим $\cos x \approx 1 - \frac{x^{2}}{2}$, первый отброшенный член $\frac{x^{4}}{4!}$

Приближенно вычислим интеграл

$$
I_{1} \approx \int\limits_{0}^{\delta} \frac{1 - \frac{x^{2}}{2}}{\sqrt{x}}\,dx
= 2\sqrt{\delta} - \frac{\delta^{5/2}}{5}.
$$

Оценим ошибку и найдем $\delta$

$$
\Delta = \int\limits_{0}^{\delta} \frac{x^{4}}{4!\sqrt{x}}\,dx
= \frac{\delta^{9/2}}{108}, \qquad \Delta \le \frac{\varepsilon}{2}.
$$

$$
\frac{\delta^{9/2}}{108} \le \frac{1}{2}\cdot10^{-3}
\quad\Rightarrow\quad
\delta \le 0.48.
$$

2) подсчет $I_{2},\ h-?$

Оценим погрешность вычисления интеграла $I_{2}$ по формуле трапеций и найдем ограничение на шаг сетки, $M_{2}=4.28,$

$$
\frac{1}{12}\max|f''|\,(b-a)\,h^{2}\le\frac{\varepsilon}{2}
$$

$$
\frac{1}{12}\cdot4.28\cdot(1-0.48)\,h^{2}\le\frac{1}{2}\cdot10^{-3}
$$

$$
h\le5.19\cdot10^{-2}
$$

Найдем количество интервалов равномерной сетки
$$
\frac{b-a}{h}=\frac{1-\delta}{h}=\frac{0.52}{5.19\cdot10^{-2}}\approx10.02
\Rightarrow h=\frac{0.52}{11},\quad N=11.
$$

Итог:
$$
I=2\sqrt{\delta}-\frac{\delta^{5/2}}{5}+I_{2},\qquad \delta=0.48.
$$

Передаём в программу:
$$
f(x)=\frac{\cos x}{\sqrt{x}},\qquad a=0.48,\quad b=1,\quad N=11.
$$


## Задача 4

Предложить алгоритм, как пользуясь программой, реализующей метод трапеций для численного интегрирования регулярных функций с произвольным шагом, вычислить следующий интеграл с точностью $\varepsilon=0.5\cdot 10^{-3}$

$$
I=\int\limits_{0}^{+\infty}\frac{dx}{1+x^{4}}$$

**Решение.**
$$
I=\underbrace{\int\limits_{0}^{M}\frac{dx}{1+x^{4}}}_{I_{1}}
+\underbrace{\int\limits_{M}^{+\infty}\frac{dx}{1+x^{4}}}_{I_{2}}
$$

Интеграл $I_{1}$ находится численно с точностью $\varepsilon/2$,  
Интеграл $I_{2}$ вычисляется приближенно аналитически с точностью $\varepsilon/2$.

*Вариант 1.*

1) подсчёт $I_2$, $M -?$

Выберем $M$ таким образом, чтобы интеграл $I_2$ не превышал по величине точности $\frac{\varepsilon}{2}$
$$
\left|\int\limits_{M}^{+\infty}\frac{dx}{1+x^{4}}\right|\le \frac{\varepsilon}{2}
$$

$$
I_2=\int\limits_{M}^{+\infty}\frac{1}{1+x^{4}}\,dx
    <\int\limits_{M}^{+\infty}\frac{dx}{x^{4}}
    =-\frac{1}{3x^{3}}\Big|_{M}^{+\infty}
    =\frac{1}{3M^{3}}
    <\frac{\varepsilon}{2}
$$

$$
\frac{1}{3M^{3}}\le \frac12\cdot 10^{-3}
\quad\Rightarrow\quad
M\ge 11
$$

2) подсчет $I_1$, $h-?$

Найдем ограничение на шаг сетки, воспользовавшись формулой для оценки ошибки метода по формуле трапеций, $M_2 = 2{,}5$

$$
\frac{1}{12} M_2 (b-a) h^2 \le \frac{\varepsilon}{2}
$$

$$
\frac{1}{12} \cdot 2{,}5 \cdot (11-0) \, h^2 \le \frac{1}{4}\cdot 10^{-3}
$$

## Задача 5

Предложить алгоритм, как пользуясь программой, реализующей метод трапеций для численного интегрирования регулярных функций с произвольным шагом, вычислить следующий интеграл с точностью $\varepsilon=10^{-3}$

$$I = \int \limits _{0}^{\pi/2} \ln(\sin x)\,dx$$

**Решение.**

Проведем следующие преобразования интеграла (воспользуемся методом Канторовича)

$$
I = \int \limits _{0}^{\pi/2} \ln(\sin x)\,dx
= \int\limits_{0}^{\pi/2} \ln(\sin x)\,dx
- \int\limits_{0}^{\pi/2} \ln x\,dx
+ \int\limits_{0}^{\pi/2} \ln x\,dx
$$

$$
= \int\limits_{0}^{\pi/2} \ln\!\left(\frac{\sin x}{x}\right)\!dx
+ \frac{\pi}{2}\left(\ln\frac{\pi}{2} - 1\right)
$$

Первый интеграл вычислим по методу трапеций (численно). Шаг найдем, воспользовавшись оценкой ошибки метода
$$
M_2 = 0.59
$$

$$
\frac{1}{12} M_2 (b-a) h^2 \le \varepsilon
$$

$$
\frac{1}{12}\cdot 0.59 \cdot \left(\frac{\pi}{2}-0\right) h^2
\le 10^{-3}
$$

$$
h \le 0.0114
$$

Найдем число интервалов равномерной сетки

$$
\frac{b-a}{h}=\frac{\pi/2}{0.0114} = 13.8 \;\Rightarrow\; N = 14
$$

Итог:

$$
f(x) = \ln\!\left(\frac{\sin x}{x}\right),
\qquad
a = 0,\quad b = \frac{\pi}{2},\quad N = 14.
$$
