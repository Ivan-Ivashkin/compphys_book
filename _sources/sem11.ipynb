# Введение

В прошлом семинаре мы начали изучать квадратурные формулы Ньютона-Котеса и уже научились прмиенять их для вычисления интегралов. В этом семинаре мы уделим время обсуждению ошибок численного дифференцирования, а также способов уточнения квадратурных формул. Мы по-прежнему будем говорить о вычислении интеграла
$$
I = \int \limits_a^b f(x)\,dx,
$$
где $a$, $b$ - конечные пределы интегрирования, а у функции $f(x)$ нет особенностей на заданном интервале.

# Ошибка интегрирования по квадратурным формулам Ньютона-Котеса

Напомним способ построения квадратурных формул Ньютона-Котеса. На каждом элементарном интервале функция заменается на интерполянт и вычисляется интеграл от интерполянта. Из этого размышления понятно, что ошибка численного интегрирования по таким формулам определяется ошибкой интерополяции. Напомним, что при замене $f(x)$ на интерполянт $P_n(x)$ ошибка метода определяется остаточным членом интерполяции $R_n(x)$

$$
R_n(x) = f(x) - P_n(x) = \frac{f^{(n+1)}(\xi)}{(n+1)!} \omega_{n+1}(x), \quad a < \xi < b
$$

Тогда на каждом элементарном интервале $[x_i; x_{i+1}]$
$$
f(x) = P_n(x) + R_n(x)
$$

$$
\int \limits_{x_i}^{x_{i+1}} f(x)\,dx = \int\limits_{x_i}^{x_{i+1}}  P_n(x)\,dx + \int\limits _{x_i}^{x_{i+1}}  R_n(x)\,dx.
$$

Слева в формуле точное значение интеграла, первое слагаемое справа - приближенное значение инетеграла, а оставшееся слагаемое определеет ошибку метода при чсиленном инетгрировании. Окончательно

$$
\Delta_{\text{мет}} = \sum_{i=0}^{n-1} \int_{x_i}^{x_{i+1}} R_n(x)\,dx
$$

Применяя эту формулу, мы можем получить ошибку метода для формул, которые мы обсуждали на прошлом семинаре. БУдем полагать, что шаг сетки постоянен и равен $h$

1. Формула прямоугольников
$$
\frac{1}{2} \max |f'(x)| (b - a) h
$$

Формула прямоугольников с центральной точкой
$$
\frac{1}{24} \max |f''(x)| (b - a) h^2
$$

2) Формула трапеций
$$
\frac{1}{12} \max |f''(x)| (b - a) h^2
$$

3) Формула Симпсона
$$
\frac{1}{2880} \max |f^{(4)}(x)| (b - a) h^4
$$

Формула Симпсона с целыми узлами
$$
\frac{1}{180} \max |f^{(4)}(x)| (b - a) h^4
$$

4) Правило 3/8
$$
\frac{1}{6480} \max |f^{(4)}(x)| (b - a) h^4
$$


## Связь квадратурных формул

Приведем простой пример, показывающий, что вычисление интеграла по формуле более низкого порядка ппроксимации на сетках с шагами $h$ и $2h$ может давать тот же результат, что и однократное вычисление интеграла по формуле более высокого порядка аппроксимации.

Рассмотрим сетку, состоящую из всего трех узлов ${x_{i-1}, \space x_{i}, \space x_{i+1}}$. Вычислим интеграл по формуле трапеций с шагами $h$ и $2h$

$$
I_h^{\text{тр}} = \frac{h}{2}(f_i + f_{i+1}) + \frac{h}{2}(f_{i-1} + f_i)
= \frac{h}{2}(f_{i+1} + 2f_i + f_{i-1}),
$$

$$
I_{2h}^{\text{тр}} = \frac{2h}{2}(f_{i-1} + f_{i+1}) = h(f_{i-1} + f_{i+1}),
$$

а также интеграл по формуле Симпсона с целыми узлами
$$
I^c = \frac{h}{3}(f_{i-1} + 4f_i + f_{i+1}).
$$
Заметим, что
$$
I^c = I_h^{\text{пр}} + \frac{I_h^{\text{пр}} - I_{2h}^{\text{пр}}}{3}
$$
Последнее слагаемое в формуле называется поправка Ричардсона. В этом примере мы увидели частный случай, в следующем разделе обсудим общие подходы

## Повышение порядка точности квадратурных формул

### 1) Экстраполяция Ричардсона

Пусть вычисление интеграла $I$ производится по некоторой квадратурной формуле с известным порядком аппроксимации $p$. При вычислении с шагом $h$ получится

$$
I = I_h + C h^p.
$$

При вычислении с шагом $h/2$ получится
$$
I = I_{h/2} + C_1 ( \frac{h}{2} )^p.
$$

В предположении, что $C=C_1$ можно получить, что
$$
I^R = I_{h/2} + \frac{I_{h/2} - I_h}{2^p - 1}.
$$

где $I^R$ — уточнённое значение интеграла в соответствии с экстраполяцией Ричардсона. Такой подход обеспечивает повышение порядка точности на единицу, если $p$ — нечётное, а если $p$ — чётное, то повышение порядка на две единицы. Этим и объясняется полученный выше результат, когда вычисление интеграла по формуле трапеций второго порядка аппроксимации на разных сетках позволило в итоге получить результат, совпадающий с вычислением по формуле Симпсона четвертого порядка аппроксимации.

Слагаемое $\Delta = \dfrac{I_{h/2} - I_h}{2^p - 1}$ можно использовать для оценки погрешности вычисления по квадратурной формуле на практике. В реальных задачах редко известны константы $M_1$, $M_2$ или $M_4$, входящие в оценки погрешностей, выписанные выше. Поэтому для оценки точности результата проделывают вычислени на двух разных сетках  и вычисляют величну $\Delta$. Такой подход называют правилом Рунге



### 2) Процесс Эйткена

Пусть вычисление интеграла $I$ производится по некоторой квадратурной формуле, однако порядок сходимости $p$ неизвестен. В таком случае также есть возможность найти уточненное значение интеграла, а также можно определить порядок сходимости. Пусть интеграл теперь вычислен на сетках с шагами $h$, $h/2$, $h/4$. Пренебрежем отличием констант, входящих в ошибку метода, получим


$$
I = I_h + C h^p \quad \text{(1)}
$$

$$
I = I_{h/2} + C ( \frac{h}{2} )^p \quad \text{(2)}
$$

$$
I = I_{h/4} + C ( \frac{h}{4} )^p \quad \text{(3)}
$$
Вычтем друг из друга уравнения
$$
(2) - (1): \quad I_h - I_{h/2} = C h^p (2^{-p} - 1)
$$

$$
(3) - (2): \quad I_{h/2} - I_{h/4} = 2^{-p} C h^p (2^{-p} - 1)
$$

и получим
$$
2^p = \frac{I_h - I_{h/2}}{I_{h/2} - I_{h/4}}.
$$

Из этой формулы можно получить порядок сходимости. Найдем далее слагаемое
$$
C h^p = \frac{(I_{h/2} - I_h)^2}{2 I_{h/2} - I_h - I_{h/4}}.
$$
Тогда уточненное значение инетграла в соответствии с процессом Эйткена может быть найдено в виде
$$
I^{\text{Эйт}} = I_h + \frac{(I_{h/2} - I_h)^2}{2 I_{h/2} - I_h - I_{h/4}}.
$$


### 3) Формула Эйлера–Маклорена

Уточнить значение интеграла можно не только за счет проведения вычислений на разных сетках, но и за счет информации о производных. Например, значение интеграла, полученное по формуле трапеций можно уточнить в соответствии с формулой Эйлера-Маклорена
$$
I_{\text{ЭМ}} = I_h^{\text{пр}} - \frac{h^2}{12} \left( f'(b) - f'(a) \right).
$$

Формула обладает четвертым порядком аппроксимации, то есть ошибка составит $O(h^4)$.

\int_0^1 \frac{1}{1 + x^2} \, dx = \frac{\pi}{4} = 0{,}785398,

## Задача 3

Оцените число узлов для расчета интеграла $\int \limits_{-1}^{1} \ln(1 + x^2)\,dx$
с точностью $\varepsilon=6\cdot10^{-3}$ по:

1.   формуле прямоугольников с центральной точкой
2.   формуле трапеций
3.   формуле Симпсона



**Решение.** В данной задаче нам известна подынтегральная функция, поэтому можем использовать имеющиеся теоретические оценки. Чтобы сделать расчет потребуются значения $M_2$ и $M_4$. Можно найти их вычислив соответствующие производные и построив график. Но в этой задаче получится их найти аналитически. Пусть
$$
f(x)=\ln(1+x^2).
$$
Вычислим первую производную и представим ее в виде, удобном для нахождения следующих производных
$$
f'(x)=\frac{2x}{1+x^2}=\frac{1}{x+i}+\frac{1}{x-i}.
$$
Найдем еще несколько производных и определим закономерность
$$
f''(x)=-(x+i)^{-2}-(x-i)^{-2},
$$

$$
f^{(3)}(x)=2(x+i)^{-3}+2(x-i)^{-3},
$$

$$
f^{(4)}(x)=-6(x+i)^{-4}-6(x-i)^{-4}.
$$
Очевидно, что
$$
\bigl|f^{(n)}(x)\bigr|\le 2\,(n-1)!
$$
Также можно сделать и более общий вывод
$$
f(x)=\ln\!\left(1+\frac{x^2}{a^2}\right),\qquad
\bigl|f^{(n)}(x)\bigr|\le \frac{2\,(n-1)!}{a^{\,n}}
$$

1) *Формула прямоугольников с центральной точкой*

Оценка имеет вид

$$
\Delta \le \frac{1}{24}\,\max|f''(x)|\,(b-a)\,h^2 \le \varepsilon.
$$

Подставим в нее все известное
$$
\frac{1}{24}\cdot 2 \cdot 2 \cdot h^2 \le 6\cdot 10^{-3}.
$$
Получим
$$
h^2 \le 3.6\cdot 10^{-2}, \qquad h \le 0.190
$$

Определим шаг равномерной сетки на выбранном отрезке, для этого вычислим
$$
\frac{b-a}{h} = \frac{2}{0.190} \approx 10.5
$$
Округляем вверх до целого
$$
N = 11
$$
В итоге
$$
h = \frac{2}{11}
$$
Для вычисления потребуется 12 значений функции.



2) *Формула трапеций*

Выпишем аналогичную оценку
$$
\Delta \le \frac{1}{12} \max |f''(x)| \,(b-a)\, h^2 \le \varepsilon
$$

Подставим все известное
$$
\frac{1}{12} \cdot 2 \cdot 2 \cdot h^2 \le 6 \cdot 10^{-3}.
$$
Решая уравнение, получим
$$
h^2 \le 1.8 \cdot 10^{-2}, \qquad h \le 0.134.
$$
Найдем количество точек на отрезке
$$
\frac{b-a}{h} = \frac{2}{0.134} = 14.9, \qquad N = 15.
$$
Окончательно
$$
h = \frac{2}{15}
$$
Для вычисления интеграла потребуется 16 значений функции на отрезке.

3) *Формула Симпсона с целыми узлами*

Снова выпишем оценку

$$
\Delta \le \frac{1}{180} \max |f^{(4)}(x)| (b - a) h^4 \le \varepsilon
$$
и подставим в нее все известное
$$
\frac{1}{180} \cdot 12 \cdot 2 \cdot h^4 \le 6 \cdot 10^{-3}.
$$
Решим уравнение
$$
h^4 \le 4.5 \cdot 10^{-2}, \quad h \le 0.461.
$$
Оценим количество точек на отрезке
$$
\frac{b - a}{h} = \frac{2}{0.461} = 4.3, \quad N = 5
$$
Так как планируется использовать формулу Симпсона с целыми узлами, то необходимо разделить отрезок на четное количество интервалов, тогда окончательно
$$
N = 6, \quad h = \frac{2}{6} = \frac{1}{3}.
$$
Для вычисления интеграла потребуется 7 значений функции на отрезке.
