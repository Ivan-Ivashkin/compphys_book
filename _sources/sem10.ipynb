# Введение

На прошлых двух семинарах мы говорили о глобальной интерполяции. Один из выводов, к которому мы пришли, состоит в том, что увеличение числа узлов интерполяции на отрезке далеко не всегда приводит к уменьшению ошибки метода, более того, чем более подробная сетка, тем больше могут оказаться ошибки округления. Поэтому строить глобальный интерполянт на отрезке с использованием большого числа точек бессмысленно. Решением задачи в таком случае является использование локальной интерполяции, которую еще называют интерполяцией сплайнами. В таком случае на каждом отрезке между узлами сплайна строится своя функция, а в узлах производится по возможности гладкая сшивка.


**Определение.**
Сплайном $S_m(x) \in C^{l}[a, b]$ называется функция
$$
S_m(x) = P_m(x) \quad \text{на } [x_k, x_{k+1}],
$$
где $m-l$ — дефект сплайна.

$$
\{x_k\} \text{ — узлы сплайна}, \qquad \{x_i\} \text{ — узлы интерполяции}
$$

Если $S_m(x_i) = f(x_i)$, то сплайн называют интерполяционным.

Есть разные виды сплайнов, мы рассмотрим один из них - кубический сплайн дефекта 1.

# Построение кубического сплайна дефекта 1

Пусть узлы сплайна и узлы интерполяции совпадают
$$
\{x_k\} = \{x_i\}, \quad a \equiv x_0 < x_1 < \ldots < x_n \equiv b
$$

На каждом отрезке $[x_k, x_{k+1}]$ зададим функцию:

$$
\varphi_k(x) = S_3(x_{k},k) = a_k + b_k (x - x_k) + c_k (x - x_k)^2 + d_k (x - x_k)^3
$$

$$
h_k = x_{k+1} - x_k
$$

Всего введено $4n$ неизвестных.

Потребуем в первую очередь, чтобы сплайн был интерполяционным. В узлах интерполяции:

$$
S_3(x_k, k) = f_k, \quad k = 0, \ldots, n-1,
$$
$$
S_3(x_{k+1}, k) = f_{k+1}, \quad k = 0, \ldots, n-1,
$$

Всего $2n$ уравнений. Есть возможность задать еще $2n$ условий. Сделаем гладкую сшивку в узлах интерполяции. В точке  $x_k$ :

$$
{S_3}'(x_k, k) = {S_3}'(x_k, k-1), \quad k = 1, \ldots, n-1,
$$
$$
{S_3}''(x_k, k) = {S_3}''(x_k, k-1), \quad k = 1, \ldots, n-1,
$$

Мы задали еще  $2n - 2$ уравнения. Запишем полученные уравнения в виде системы

\begin{cases}
a_k = f_k, & k = 0, \ldots, n-1, \\[6pt]
a_k + b_k h_k + c_k h_k^2 + d_k h_k^3 = f_{k+1}, & k = 0, \ldots, n-1, \\[6pt]
b_k = b_{k-1} + 2 c_{k-1} h_{k-1} + 3 d_{k-1} h_{k-1}^2, & k = 1, \ldots, n-1, \\[6pt]
2 c_k = 2 c_{k-1} + 6 d_{k-1} h_{k-1}, & k = 1, \ldots, n-1.
\end{cases}


Всего записано $4n - 2$  уравнения, нужно ещё 2. Есть различные варианты, как их задать



*   свободный сплайн
$$ S_3''(x_0, 0) = S_3''(x_n, n-1) = 0,$$  
$$ c_0 = 0, \quad c_n = 0.$$

При таком задании дополнительных условий полученный сплайн имеет физическую интерпретацию. Если взять упругую рейку, поставить на ребро и закрепить так, чтобы она проходила через заданные точки, получится кривая, задаваемая именно свободным сплайном.
*   сплайн Шонберга
*   эмпирический сплайн
*   естественный сплайн

Информацию о других способах задания двух недостающих условий можно найти в литературе.

Закончим построение свободного кубического сплайна. Укажем эффективный способ решения выписанной системы, дополненной условиями $ c_0 = 0, \quad c_n = 0$. Коэффициенты $a_k = f_k$ уже известны. Исключим  $b_k$ и $d_k$, получим систему уравнений относительно $c_k$, которая решается методом скалярной трехточечной прогонки, имеющим линейную вычислительню сложность

$$
d_{k} = \frac{c_{k+1} - c_{k}}{3h_{k}}
$$

$$
b_k = \frac{f_{k+1} - f_k}{h_k} - \frac{2}{3} c_k h_k - \frac{1}{3} c_{k+1} h_k
$$

$$
\frac{h_{k-1}}{h_{k-1} + h_k} c_{k-1} + 2c_k + \frac{h_k}{h_{k-1} + h_k} c_{k+1} = 3f(x_{k-1}, x_k, x_{k+1})
$$

где $f(x_{k-1}, x_k, x_{k+1})$ — разделённая разность второго порядка:

$$
f(x_{k-1}, x_k, x_{k+1}) =
\frac{\dfrac{f_{k+1} - f_k}{h_k} - \dfrac{f_k - f_{k-1}}{h_{k-1}}}{h_k + h_{k-1}}
$$

Если $h_k = h_{k-1} = h$, то:

$$
f(x_{k-1}, x_k, x_{k+1}) = \frac{f_{k+1} - 2f_k + f_{k-1}}{2h^2}
$$

c_2 + 4c_3 + c_4 = 3(-0.5 - 2 \cdot 0 + 0.5), \\

## Задача 2

# Численное интегрирование

Во многих прикладных задачах возникает необходимость вычислять определённые интегралы вида

$$I=\int\limits_a^b\mathop{dx}f(x).$$

Здесь подынтегральная функция $f=f(x)$, являющаяся скалярной функцией скалярного аргумента, предполагается достаточно гладкой (по крайней мере непрерывной) на промежутке интегрирования $(a,b)$, так что интеграл $I$ существует и в смысле определения Лебега, и в смысле определения Римана. Пределы интегрирования $a$ и $b$, вообще говоря, могут быть как конечными, так и бесконечными (если интеграл понимается в несобственном смысле, то предполагается, что он сходится абсолютно).

В математическом анализе основной вычислительной формулой, применяемой для нахождения интеграла $I$ (в случае конечных $a$ и $b$), служит **формула Ньютона—Лейбница**, согласно которой

$$I=F(b)-F(a),$$

где $F=F(x)$ — первообразная функции $f$ на отрезке $[a,b]$. Однако воспользоваться этой формулой можно лишь тогда, когда первообразную $F$ удаётся выразить в элементарных функциях. К сожалению, на практике подобные случаи почти не встречаются: хорошо известно, что в современных научных задачах подынтегральные функции обычно оказываются настолько "сложными", что соответствующие интегралы "не берутся".

Типичные примеры подобных задач легко найти в курсах обыкновенных дифференциальных уравнений и уравнений в частных производных. Решения многих из этих уравнений (за исключением простейших) **выражаются**, как принято говорить, **в квадратурах**, т.е. имеют вид интегралов от некоторых комбинаций конечного числа элементарных функций. После получения квадратуры обычно считается, что рассматриваемое уравнение решено (конечно, более правильно говорить о решении начальной, краевой или начально-краевой задачи, сформулированной для данного уравнения). Однако сама по себе квадратура, являясь решением уравнения с теоретической точки зрения, ещё не даёт решения в практическом смысле. Чтобы "что-то посчитать" с помощью найденной квадратуры в рамках решения поставленной предметной задачи, надо тем или иным способом вычислить интеграл, входящий в квадратуру. Чем "сложнее" дифференциальное уравнение, тем "сложнее" интеграл, и на практике для его нахождения часто приходится прибегать к численным методам.

Наконец, некоторые дифференциальные уравнения не удаётся решить даже в квадратурах. В таких случаях решения находят приближённо с помощью численных методов, основанных, например, на конечно-разностной аппроксимации рассматриваемых уравнений (мы будем изучать соответствующие методы в следующем семестре). По сути, такой подход тоже представляет собой численное интегрирование: можно сказать, что предложенный способ решения дискретного аналога "настоящего" уравнения и есть способ численного интегрирования исходного уравнения, которое не удаётся проинтегрировать аналитически.

Для численного интегрирования применяют различные подходы. Наиболее широкое распространение получили методы, основанные на замене интеграла $I$ конечной суммой

$$\widetilde{I}=\sum\limits_{i=0}^N c_i f(x_i),$$

в которой $c_i$ — некоторые числовые коэффициенты, $x_i$ — точки, принадлежащие отрезку интегрирования $[a,b]$, а $N$ — целое число. Сумма $\widetilde{I}$ называется **квадратурной суммой** (этим как бы подчёркивается, что она заменяет собой квадратуру, т.е. интеграл). Эта сумма строится так, чтобы выполнялось приближённое равенство $I\simeq\widetilde{I}$, называемое **квадратурной формулой**. Коэффициенты $c_i$ называют **весами квадратурной формулы**, а точки $x_i$ — её **узлами**.

Квадратурные формулы можно строить различными способами, из которых мы рассмотрим два наиболее распространённых.

Первый способ основан на следующей идее. Подынтегральная функция $f$  приближается на отрезке $[a,b]$ алгебраическим полиномом $P_n$ :

$$I=\int\limits_a^b\mathop{dx}f(x)\simeq\int\limits_a^b\mathop{dx}P_n(x)\equiv \widetilde{I}.$$

Поскольку

$$P_n(x)=\sum\limits_{i=0}^n f(x_i)\cdot\ell_i(x),$$

можно написать:

$$\widetilde{I}=\sum\limits_{i=0}^n c_if(x_i),\quad c_i=\int\limits_a^b\mathop{dx}\ell_i(x).$$

Как видим, при замене функции $f$ на полином $P_n$ действительно получилась квадратурная формула. Ниже мы увидим, что в процедуру построения этой формулы следует внести одно небольшое усложнение, заключающееся в предварительном разбиении отрезка интегрирования на несколько подотрезков. Такие формулы называются **квадратурными формулами интерполяционного типа**, или **квадратурными формулами Ньютона—Котеса** (строго говоря, последнее название применимо в специальном случае равномерной сетки — см. ниже).

Второй способ построения формул численного интегрирования заключается в специальном подборе весов $c_i$ и узлов $x_i$ с таким расчётом, чтобы приближённое равенство

$$I=\int\limits_a^b\mathop{dx}f(x)\simeq \sum\limits_{i=0}^N c_i f(x_i)=\widetilde{I}$$

становилось точным на полиномах как можно более высокой степени. Формулы, получающиеся на основе этого требования, называются **квадратурными формулами Гаусса** (или Гаусса—Кристоффеля).

В настоящем семинаре мы сосредоточимся на рассмотрении квадратурных формул Ньютона—Котеса. Далее будем предполагать, что оба предела интегрирования $a$ и $b$ конечны, а подынтегральная функция не имеет особенностей на отрезке $[a,b]$ (интеграл рассматривается в собственном смысле).

# Квадратурные формулы Ньютона—Котеса

Как уже отмечалось, квадратурные формулы Ньютона—Котеса получаются при замене подыинтегральной функции интерполяционным полиномом. Выполним эту замену более подробно, чем это было сделано выше.

Из определения остаточного члена интерполяции следует, что

$$f(x)=P_n(x)+R_n(x),\quad x\in[a,b].$$

Запишем интерполяционный полином в форме Лагранжа и выражение для остаточного члена, полученное на прошлом семинаре:

$$f(x)=\sum\limits_{i=0}^nf(x_i)\cdot\ell_i(x)+\frac{f^{(n+1)}(\xi)}{(n+1)!}\cdot\omega_n(x),\quad \xi\in[a,b].$$

Интегрирование этого равенства по отрезку $[a,b]$ даёт:

$$I=\widetilde{I}+\frac{f^{(n+1)}(\xi)}{(n+1)!}\cdot\int\limits_a^b\mathop{dx}\omega_n(x),\quad \widetilde{I}=\sum\limits_{i=0}^nc_if(x_i),\quad c_i=\int\limits_a^b\mathop{dx}\ell_i(x).$$

Отсюда следует оценка на ошибку квадратурной формулы (берём точные верхние грани по $x$ и $\xi$):

$$|I-\widetilde{I}|\leq \frac{M_{n+1}\cdot \big|\big|\omega_n(x)\big|\big|_C}{(n+1)!}\cdot\int\limits_a^b\mathop{dx}=\frac{M_{n+1}\cdot \big|\big|\omega_n(x)\big|\big|_C}{(n+1)!}\cdot (b-a).$$

Здесь использованы те же обозначения, что и на прошлом семинаре:

$$M_{n+1}=\sup\limits_{\xi\in[a,b]}\big|f^{(n+1)}(\xi)\big|,\quad \big|\big|\omega_n(x)\big|\big|_C=\sup_\limits{x\in[a,b]}\big|\omega_n(x)\big|.$$

Обратим внимание на два обстоятельства. Во-первых, **оценка получилась пропорциональной длине отрезка интегрирования** $(b-a)$. Во-вторых, в неё вошли величины $M_{n+1}$ и $\big|\big|\omega_n(x)\big|\big|_C$, вычисленные как точные верхние грани, взятые **по всему** отрезку $[a,b]$. Как известно,

$$\sup\limits_{y\in[\alpha,\beta]}\big|g(y)\big|\geq \sup\limits_{y\in[\gamma,\rho]}\big|g(y)\big|,\quad\text{если}\quad[\gamma,\rho]\subset [\alpha,\beta],$$

т.е. чем шире отрезок, тем больше точная верхняя грань. Эти два обстоятельства наводят на мысль разбить исходный отрезок интегрирования $[a,b]$ на подотрезки с целью уменьшения оценки погрешности. Выполним это разбиение.

Введём на отрезке $[a,b]$ сетку (вообще говоря, неравномерную; эту сетку в дальнейшем будем называть **основной**):

$$x_0=a,\quad x_1,\quad \dots, \quad x_N=b;\quad h_i=x_{i+1}-x_i,\quad\sum\limits_{i=0}^{N-1} h_i=b-a.$$

Узлы сетки делят отрезок интегрирования $[a,b]$ на подотрезки (**частичные отрезки интегрирования**) $[x_i,x_{i+1}]$ ($0\leq i\leq N-1$). Пользуясь аддитивностью интеграла, можем написать:

$$I=\int\limits_a^b\mathop{dx}f(x)=\sum\limits_{i=0}^{N-1}\int\limits_{x_i}^{x_{i+1}}\mathop{dx}f(x)\equiv \sum\limits_{i=0}^{N-1} I_i.$$

Интеграл $I_i$ приближаем квадратурной суммой

$$\widetilde{I}_i=\sum\limits_{j=0}^nc_j^{(i)}f\big(x_j^{(i)}\big),\quad c_j^{(i)}=\int\limits_{x_i}^{x_{i+1}}\mathop{dx}\ell_j(x),$$

построенной на **подсетке** (вообще говоря, тоже неравномерной)

$$x_0^{(i)}=x_i,\quad x_1^{(i)},\quad\dots,\quad x_n^{(i)}=x_{i+1};\quad x_j^{(0)}\in[x_i,x_{i+1}],\quad 0\leq j\leq n.$$

Понятно, что оценка погрешности этого приближения аналогична оценке, полученной выше на отрезке $[a,b]$ (при отсутствии разбиения):

$$|I_i-\widetilde{I}_i|\leq \frac{M_{n+1}^{(i)}\cdot \big|\big|\omega_n(x)\big|\big|_C^{(i)}}{(n+1)!}\cdot h_i.$$

Принципиальное отличие состоит в том, что **точные верхние грани берутся по частичным отрезкам**, а не по всему отрезку интегрирования:

$$M_{n+1}^{(i)}=\sup\limits_{\xi\in[x_i,\,x_{i+1}]}\big|f^{(n+1)}(\xi)\big|,\quad \big|\big|\omega_n(x)\big|\big|_C^{(i)}=\sup_\limits{\xi\in[x_i,\,x_{i+1}]}\big|\omega_n(x)\big|.$$

Погрешность вычисления всего интеграла $I$:

$$|I-\widetilde{I}|=\Bigg|\sum\limits_{i=0}^{N-1}(I_i-\widetilde{I}_i)\Bigg|\leq\sum\limits_{i=0}^{N-1}|I_i-\widetilde{I}_i|\leq \frac{1}{(n+1)!}\sum\limits_{i=0}^{N-1}M_{n+1}^{(i)}\cdot \big|\big|\omega_n(x)\big|\big|_C^{(i)}\cdot h_i.$$

Ясно, что эта оценка не хуже полученной выше оценки. Это следует из неравенств

$$M^{(i)}_{n+1}\leq M_{n+1},\quad \big|\big|\omega_n(x)\big|\big|_C^{(i)}\leq \big|\big|\omega_n(x)\big|\big|_C,$$

справедливых при любом $i$ в силу упоминавшегося выше свойства точной верхней грани.

Квадратурные формулы интерполяционного типа (формулы Ньютона—Котеса), использующие разбиение отрезка интегрирования на частичные отрезки, называют **составными формулами**. Отметим, что термин "квадратурная формула Ньютона—Котеса" используется, вообще говоря, в том случае, когда **сетка, на которой выполняется интерполяция** (т.е. основная сетка в случае "обычной" формулы и подсетка в случае составной формулы), **является равномерной**. Впрочем, тот же термин иногда применяют как синоним более общего термина "квадратурная формула интерполяционного типа".

На практике чаще используют именно составные формулы, поскольку при прочих равных условиях они дают меньшую погрешность. Мы убедились в этом выше, выполнив грубую оценку погрешности: при переходе от составной формулы к "обычной" сумма

$$\sum\limits_{i=0}^{N-1}M_{n+1}^{(i)}\cdot \big|\big|\omega_n(x)\big|\big|_C^{(i)}\cdot h_i$$

заменяется величиной

$$M_{n+1}\cdot \big|\big|\omega_n(x)\big|\big|_C\cdot (b-a),$$

что, очевидно, только ухудшает оценку. Можно также показать, что **составная формула менее чувствительна к ошибкам округления и входных данных**, т.е. обнаруживает лучшую численную устойчивость. То есть не имеет смысла строить интерполяционный многолчен двадцатой степени на сетке из 21 узла, например, более надежно применить составную формулу, в построении которой задействован интерполянт-многочлен невысокой степени.

Дальнейший анализ погрешностей квадратурных формул интерполяционного типа мы отложим до следующего семинара. В настоящем семинаре ограничимся только построением конкретных формул. Для простоты будем считать сетку, на которой осуществляется интерполяция, равномерной (т.е. рассматривать формулы Ньютона—Котеса "в узком смысле").

## Формула прямоугольников

Положим $n=0$. Иначе говоря, приблизим функцию $f$ на каждом частичном отрезке полиномом нулевой степени $P_0$, т.е. константой. В этом случае сетка для интерполяции (подсетка) состоит из единственного узла (вырождается). Напомним, что при единственном узле в сетке вспомогательный полином Лагранжа $\ell_0(x)$ по определению полагается равным единице (см. восьмой семинар). Поэтому

$$\widetilde{I}_i=c_0^{(i)}f\big(x_0^{(i)}\big),\quad c_0^{(i)}=\int\limits_{x_i}^{x_{i+1}}\mathop{dx}=x_{i+1}-x_i=h_i.$$

Квадратурная сумма принимает следующий вид:

$$\widetilde{I}=\sum\limits_{i=0}^{N-1}f\big(x_0^{(i)}\big)\cdot h_i.$$

Соответствующая квадратурная формула называется **формулой прямоугольников**. Это название объясняется её геометрической интерпретацией. Если понимать вычисляемый интеграл как площадь под графиком кривой $y=f(x)$, то приближённое вычисление этой площади по найденной формуле сводится к разделению ("разрезанию") её на прямоугольники и суммированию площадей этих прямоугольников.

На следующем семинаре мы увидим, что **формула прямоугольников имеет первый порядок аппроксимации**. Это значит, что погрешность формулы прямоугольников определяется слагаемым, пропорциональным $h$ (ср. с понятием порядка аппроксимации в случае формул численного дифференцирования). Последнее верно при любом $x_0^{(i)}\in[x_i,x_{i+1}]$ (т.е. при любом выборе единственного узла подсетки на частичном отрезке интегрирования), за исключением одного специального случая. Оказывается, что если узел подсетки совпадает с серединой отрезка $[x_i,x_{i+1}]$, то порядок аппроксимации формулы прямоугольников повышается до второго.

**Явление повышения порядка аппроксимации на единицу** (по сравнению с ожидаемым порядком $n+1$) **является общим для всех симметричных формул Ньютона—Котеса** (т.е. таких формул, в которых середина частичного отрезка интегрирования является узлом, а другие узлы выбраны симметрично относительно него).

## Формула трапеций

При $n=1$ функция $f$ приближается на каждом частичном отрезке интегрирования полиномом первой степени $P_1$, а каждая подсетка состоит из двух узлов (совпадающих с граничными точками отрезка $[x_i,x_{i+1}]$). Поэтому

$$\widetilde{I}_i=c_0^{(i)}f\big(x_0^{(i)}\big)+c_1^{(i)}f\big(x_1^{(i)}\big).$$

Вычисляем коэффициенты:

$$c_0^{(i)}=\int\limits_{x_i}^{x_{i+1}}\mathop{dx}\ell_0(x)=\int\limits_{x_i}^{x_{i+1}}\mathop{dx}\frac{x-x_1^{(i)}}{x_0^{(i)}-x_1^{(i)}}=\frac{\Big[x_{i+1}-x_1^{(i)}\Big]^2-\Big[x_{i}-x_1^{(i)}\Big]^2}{(-2)\cdot h_i}=\frac{h_i}{2};$$

$$c_1^{(i)}=\int\limits_{x_i}^{x_{i+1}}\mathop{dx}\ell_1(x)=\int\limits_{x_i}^{x_{i+1}}\mathop{dx}\frac{x-x_0^{(i)}}{x_1^{(i)}-x_0^{(i)}}=\frac{\Big[x_{i+1}-x_0^{(i)}\Big]^2-\Big[x_{i}-x_0^{(i)}\Big]^2}{2h_i}=\frac{h_i}{2}.$$

Квадратурная сумма:

$$\widetilde{I}=\sum\limits_{i=0}^{N-1}\frac{f\big(x_0^{(i)}\big)+f\big(x_1^{(i)}\big)}{2}\,h_i=\sum\limits_{i=0}^{N-1}\frac{f(x_i)+f(x_{i+1})}{2}\,h_i.$$

Полученная формула численного интегрирования называется **формулой трапеций**. Она имеет **второй порядок аппроксимации**.

## Формула Симпсона

На каждом элементарном отрезке функция заменяется на многочлен $P_2(x)$ и получается формула
$$
I =\sum_{i=0}^{n-1} \frac{h_i}{6} (f(x_i) + 4f(x_i+h_i/2) + f(x_{i+1}))\equiv \sum_{i=0}^{n-1} \frac{h_i}{6} (f_i + 4f_{i+\frac{1}{2}} + f_{i+1})
$$

Можно преобразовать формулу так, чтобы в нее входили только целые узлы (в таком случае в сетке должно быть нечётное число точек):

$$
I = \sum_{i=0}^{n/2-1} \frac{h}{3} (f_{2i} + 4f_{2i+1} + f_{2i+2}).
$$

Формула Симпсона обладает четвертым порядком аппроксимации.

## Правило 3/8

На каждом элементарном отрезке функция заменяется на многочлен $P_3(x)$ и получается формула

$$
I = \sum_{i=0}^{n-1} h_i \left( \frac{1}{8}f_i + \frac{3}{8}f_{i+\frac{1}{3}} + \frac{3}{8}f_{i+\frac{2}{3}} + \frac{1}{8}f_{i+1} \right).
$$

Эта формула также обладает четвертым порядком аппроксимации.

## Задача 3

Вычислите интеграл от таблично заданной функции при помощи метода трапеций. Чему равно значение интеграла от $f(x)$, вычисленного по формуле Симпсона?

$$\begin{array}{|c|c|c|c|c|c|c|c|c|c|}
    \hline
    x & 0.00 & 0.125 & 0.250 & 0.375 & 0.500 & 0.625 & 0.750 & 0.875 & 1.000 \\ \hline
    f(x) & 1.000000 & 0.984615 & 0.941176 & 0.876712 & 0.800000 & 0.719101 & 0.640000 & 0.566372 & 0.500000 \\ \hline
  \end{array}$$



**Решение.**

Для удобства придадим таблице вид

$$
\begin{array}{c|ccccccccc}
x_i & 0 & \tfrac{1}{8} & \tfrac{1}{4} & \tfrac{3}{8} & \tfrac{1}{2} & \tfrac{5}{8} & \tfrac{3}{4} & \tfrac{7}{8} & 1 \\
\hline
f_i & f_0 & f_1 & f_2 & f_3 & f_4 & f_5 & f_6 & f_7 & f_8
\end{array}
$$

Шаг сетки постоянный, $h=1/8$. Вычислим интеграл по формуле трапеций с шагом $h$
$$
I_h^{\text{тр}} = \frac{h}{2} (f_0 + f_8 + 2(f_1 + f_2 + f_3 + f_4 + f_5 + f_6 + f_7))
$$
$$
I_h^{\text{тр}} = 0.784747
$$

Вычислим интеграл по формуле трапеций с шагом $2h$
$$
I_{2h}^{\text{тр}} = \frac{2h}{2} (f_0 + f_8 + 2(f_2 + f_4 + f_6)) = 0.782794
$$

Вычислим интеграл по формуле Симпсона с целыми узлами
$$
I^c = \frac{h}{3} (f_0 + 4f_1 + f_2) + \frac{h}{3} (f_2 + 4f_3 + f_4) + \frac{h}{3} (f_4 + 4f_5 + f_6) + \frac{h}{3} (f_6 + 4f_7 + f_8)
$$
$$
I^c = 0.785398
$$

В таблице были даны значения функции $f(x)=\dfrac{1}{1+x^2}$, интеграл может быть вычислен аналитически
$$
\int_0^1 \frac{1}{1+x^2}\,dx = \frac{\pi}{4} = 0.785398
$$

Мы получили, что формула Симпсона в применении к вычислению данного интеграла дала ответ с точностью до 6 знаков после запятой (в таблице как раз даны значения функции с этой же точностью).
