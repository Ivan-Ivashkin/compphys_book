# Введение

На прошлых двух семинарах мы говорили о глобальной интерполяции. Один из выводов, к которому мы пришли, состоит в том, что увеличение числа узлов интерполяции на отрезке далеко не всегда приводит к уменьшению ошибки метода, более того, чем более подробная сетка, тем больше могут оказаться ошибки округления. Поэтому строить глобальный интерполянт на отрезке с использованием большого числа точек бессмысленно. Решением задачи в таком случае является использование локальной интерполяции, которую еще называют интерполяцией сплайнами. В таком случае на каждом отрезке между узлами сплайна строится своя функция, а в узлах производится по возможности гладкая сшивка.


**Определение.**
Сплайном $S_m(x) \in C^{l}[a, b]$ называется функция
$$
S_m(x) = P_m(x) \quad \text{на } [x_k, x_{k+1}],
$$
где $m-l$ — дефект сплайна.

$$
\{x_k\} \text{ — узлы сплайна}, \qquad \{x_i\} \text{ — узлы интерполяции}
$$

Если $S_m(x_i) = f(x_i)$, то сплайн называют интерполяционным.

Есть разные виды сплайнов, мы рассмотрим один из них - кубический сплайн дефекта 1.

# Построение кубического сплайна дефекта 1

Пусть узлы сплайна и узлы интерполяции совпадают
$$
\{x_k\} = \{x_i\}, \quad a \equiv x_0 < x_1 < \ldots < x_n \equiv b
$$

На каждом отрезке $[x_k, x_{k+1}]$ зададим функцию:

$$
\varphi_k(x) = S_3(x_{k},k) = a_k + b_k (x - x_k) + c_k (x - x_k)^2 + d_k (x - x_k)^3
$$

$$
h_k = x_{k+1} - x_k
$$

Всего введено $4n$ неизвестных.

Потребуем в первую очередь, чтобы сплайн был интерполяционным. В узлах интерполяции:

$$
S_3(x_k, k) = f_k, \quad k = 0, \ldots, n-1,
$$
$$
S_3(x_{k+1}, k) = f_{k+1}, \quad k = 0, \ldots, n-1,
$$

Всего $2n$ уравнений. Есть возможность задать еще $2n$ условий. Сделаем гладкую сшивку в узлах интерполяции. В точке  $x_k$ :

$$
{S_3}'(x_k, k) = {S_3}'(x_k, k-1), \quad k = 1, \ldots, n-1,
$$
$$
{S_3}''(x_k, k) = {S_3}''(x_k, k-1), \quad k = 1, \ldots, n-1,
$$

Мы задали еще  $2n - 2$ уравнения. Запишем полученные уравнения в виде системы

\begin{cases}
a_k = f_k, & k = 0, \ldots, n-1, \\[6pt]
a_k + b_k h_k + c_k h_k^2 + d_k h_k^3 = f_{k+1}, & k = 0, \ldots, n-1, \\[6pt]
b_k = b_{k-1} + 2 c_{k-1} h_{k-1} + 3 d_{k-1} h_{k-1}^2, & k = 1, \ldots, n-1, \\[6pt]
2 c_k = 2 c_{k-1} + 6 d_{k-1} h_{k-1}, & k = 1, \ldots, n-1.
\end{cases}


Всего записано $4n - 2$  уравнения, нужно ещё 2. Есть различные варианты, как их задать



*   свободный сплайн
$$ S_3''(x_0, 0) = S_3''(x_n, n-1) = 0,$$  
$$ c_0 = 0, \quad c_n = 0.$$

При таком задании дополнительных условий полученный сплайн имеет физическую интерпретацию. Если взять упругую рейку, поставить на ребро и закрепить так, чтобы она проходила через заданные точки, получится кривая, задаваемая именно свободным сплайном.
*   сплайн Шонберга
*   эмпирический сплайн
*   естественный сплайн

Закончим построение свободного кубического сплайна. Укажем эффективный способ решения выписанной системы, дополненной условиями $ c_0 = 0, \quad c_n = 0$. Коэффициенты $a_k = f_k$ уже известны. Исключим  $b_k$ и $d_k$, получим систему уравнений относительно $c_k$, которая решается методом скалярной трехточечной прогонки, имеющим линейную вычислительню сложность

$$
d_{k} = \frac{c_{k+1} - c_{k}}{3h_{k}}
$$

$$
b_k = \frac{f_{k+1} - f_k}{h_k} - \frac{2}{3} c_k h_k - \frac{1}{3} c_{k+1} h_k
$$

$$
\frac{h_{k-1}}{h_{k-1} + h_k} c_{k-1} + 2c_k + \frac{h_k}{h_{k-1} + h_k} c_{k+1} = 3f(x_{k-1}, x_k, x_{k+1})
$$

где $f(x_{k-1}, x_k, x_{k+1})$ — разделённая разность второго порядка:

$$
f(x_{k-1}, x_k, x_{k+1}) =
\frac{\dfrac{f_{k+1} - f_k}{h_k} - \dfrac{f_k - f_{k-1}}{h_{k-1}}}{h_k + h_{k-1}}
$$

Если $h_k = h_{k-1} = h$, то:

$$
f(x_{k-1}, x_k, x_{k+1}) = \frac{f_{k+1} - 2f_k + f_{k-1}}{2h^2}
$$


## Задача 1

Построить кубический сплайн для функции, заданной в узлах интерполяции, предполагая, что сплайн имеет нулевую кривизну при $x= x_0$, $x= x_n$. Вычислить приближенное значение функции в точке $x=1.5$. Сравнить с точным значением.

\begin{array}{|c|c|c|c|c|c|}
\hline
x & 0 & 1 & 2 & 3 & 4 \\
\hline
\cos\left(\frac{\pi x}{6}\right) & 1.000 & 0.866 & 0.500 & 0.000 & -0.500 \\
\hline
\end{array}


**Решение.** По таблице определим $h = 1$, по условию $c_0 = c_4 = 0$. Запишем систему уравнений относиельно коэффициентов $c_k$

$$
\frac{1}{2}c_{k-1} + 2c_k + \frac{1}{2}c_{k+1} = \frac{3}{2}(f_{k+1} - 2f_k + f_{k-1})
$$

$$
c_{k-1} + 4c_k + c_{k+1} = 3(f_{k+1} - 2f_k + f_{k-1})
$$

$$
\begin{cases}
c_0 = 0, \\
c_0 + 4c_1 + c_2 = 3(0.5 - 2 \cdot 0.866 + 1), \\
c_1 + 4c_2 + c_3 = 3(0 - 2 \cdot 0.5 + 0.866), \\
c_2 + 4c_3 + c_4 = 3(-0.5 - 2 \cdot 0 + 0.5), \\
c_4 = 0.
\end{cases}
$$

Её решение
$$
c_0 = 0, \quad c_4 = 0
$$

$$
c_1 = -0.158, \quad c_2 = -0.065, \quad c_3 = 0.016
$$

Так как в условии требуется найти значение при $x = 1.5$, потребуется найти значения по $S_3(x,1)$


$$
S_3(x,1) = a_1 + b_1(x - 1) + c_1(x - 1)^2 + d_1(x - 1)^3
$$

Найдем недостающие коэффициенты
$$
a_1 = f_1 = 0.866
$$

$$
b_1 = \frac{f_2 - f_1}{1} - \frac{2}{3}c_1 - \frac{1}{3}c_2 = 0.5 - 0.866 + \frac{2}{3}\cdot(-0.158) + \frac{1}{3}\cdot(-0.065) = -0.239
$$

$$
d_1 = \frac{c_2 - c_1}{3 \cdot 1} = \frac{-0.065 + 0.158}{3} = 0.031
$$

Запишем $S_3(x,1)$:
$$
S_3(x,1) = 0.866 - 0.239(x - 1) - 0.158(x - 1)^2 + 0.031(x - 1)^3
$$
Найдем требуемое значение
$$
S_3(1.5,1) = 0.866 - 0.239 \cdot \frac{1}{2} - 0.158 \cdot \frac{1}{4} + 0.031 \cdot \frac{1}{8} = 0.711
$$
Сравним с точным значением
$$
\cos\left(\frac{\pi \cdot 1.5}{6}\right) = \cos\left(\frac{\pi}{4}\right) = 0.707
$$
Ошибка $0.6\%$.



## Задача 2

1) Найти производную $f'(1.5)$, построив интерпояционный многочлен, проходящий через все точки таблицы.

2) Найти производную $f'(1.5)$, построив свободный кубический сплайн дефекта 1.
\begin{array}{c|cccc}
x_i & 0 & 1 & 2 & 3 \\
\hline
f_i & 0 & 1 & 0 & -1
\end{array}

**Решение.**

1) Построим таблицу разделённых разностей:

$$
\begin{array}{c|c|c|c|c}
x & y & \Delta^1 & \Delta^2 & \Delta^3 \\ \hline
0 & 0 & 1 &   &   \\
1 & 1 & -1 & -1 & \tfrac{1}{3} \\
2 & 0 & -1 & 0 &   \\
3 & -1 &   &   &
\end{array}
$$

Запишем интерполяционный многочлен в форме Ньютона
$$
P(x) = 0 + 1(x - 0) - 1(x - 0)(x - 1) + \tfrac{1}{3}(x - 0)(x - 1)(x - 2)
$$

Раскроем скобки и приведем подобные
$$
P(x) = \tfrac{1}{3}x^3 - 2x^2 + \tfrac{8}{3}x
$$
Найдем производную интерполянта
$$
P'(x) = x^2 - 4x + \tfrac{8}{3}
$$

$$
P'(1.5) = 2.25 - 6 + \tfrac{8}{3} = -1.083
$$
В таблице приведены значения функции $y=\sin \frac{\pi x}{2}$. Вычислим точное знаечние производной в указанной точке
$$
\left(\sin \frac{\pi x}{2}\right)'_{x=1.5} = \frac{\pi}{2}\cos\frac{\pi x}{2}\bigg|_{x=1.5}
= \frac{\pi}{2}\cos\frac{3\pi}{4} = -1.11072
$$

Погрешность нахождения производной составляет $\approx 3\%$.

2) Построим свободный кубический сплайн дефекта 1, тогда $c_0 = c_3 = 0$

$$
\frac{1}{2}c_{k-1} + 2c_k + \frac{1}{2}c_{k+1}
= 3\,f(x_{k-1}, x_k, x_{k+1})
$$
Система относительно коэффициентов $c_k$ примет вид

$$
\begin{pmatrix}
1 & 0 & 0 & 0\\
\frac12 & 2 & \frac12 & 0\\
0 & \frac12 & 2 & \frac12\\
0 & 0 & 0 & 1
\end{pmatrix}
\begin{pmatrix}
c_0\\ c_1\\ c_2\\ c_3
\end{pmatrix}
=
\begin{pmatrix}
0\\ -3\\ 0\\ 0
\end{pmatrix}
$$

Найдем ее решение

$$
\begin{cases}
2c_1 + \tfrac12 c_2 = -3,\\[2pt]
\tfrac12 c_1 + 2c_2 = 0
\end{cases}
\qquad\Rightarrow\qquad
c_1 = -\tfrac{8}{5},\;\; c_2 = \tfrac{2}{5}.
$$

Найдем недостающие коэффициенты $a_1$, $b_1$, $d_1$
$$
a_1 = f_1 = 1
$$

$$
b_1 = \frac{f_2 - f_1}{1} - \frac{2}{3}c_1 - \frac{1}{3}c_2
= -1 - \frac{1}{3}\left(\frac{2}{5} - \frac{8}{5}\right) = -\frac{7}{15}
$$

$$
d_1 = \frac{c_2 - c_1}{3 \cdot 1}
= \frac{\frac{2}{5} - \left(-\frac{8}{5}\right)}{3} = \frac{2}{3}
$$

Многочлен третьей степени, задающий сплайн на отрезке $[1,2]$ имеет вид
$$
S_3(x,1) = a_1 + b_1(x - 1) + c_1(x - 1)^2 + d_1(x - 1)^3
$$

$$
S_3(x,1) = 1 - \frac{7}{15}(x - 1) - \frac{8}{5}(x - 1)^2 + \frac{2}{3}(x - 1)^3
$$

Найдем производную многочлена

$$
S_3'(x,1) = -\frac{7}{15} - \frac{16}{5}(x - 1) + 2(x - 1)^2
$$
И вычислм значение производной в интересующей нас точке
$$
S_3'(1.5,1) = -\frac{7}{6} = -1.17
$$

Погрешность составляет $\approx 5\%$.
