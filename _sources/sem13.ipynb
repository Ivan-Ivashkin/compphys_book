# Сходимость, аппроксимация, устойчивость

Вернемся еще раз к задаче Коши, с помощью которой описывается колебание математического маятника

$$\ddot\varphi+k\sin\varphi=0,\quad \varphi(0)=\phi_0, \quad \dot\varphi(0)=\omega_0.$$

Здесь $\varphi=\varphi(t)$  — угол отклонения маятника от вертикали, $t$ — время, $k=g/l$, где $l$ — длина маятника, а $g$ — ускорение свободного падения.

Перейдем от уравнения второго порядка к системе уравнений первого порядка, введем $v(t)=\dot\varphi(t)$, тогда система примет вид

$$\dot v=-k\sin \varphi,$$
$$\dot\varphi=v.$$

Выпишем также начальные условия

$$\varphi(0)=\phi_0,\quad v(0)=\omega_0.$$

На втором семинаре мы рассматривали различные конечно-разностные аппроксимации производных.  Введем одномерную сетку $\{t_i\}_{i=0}^N$ с шагами $\tau_i=t_{i+1}-t_i$, причём $t_0=0$, а $t_N=T$. Для простоты будем считать, что $\tau_0=...=\tau_{N-1}\equiv\tau$ (равномерная сетка). Воспользуемся производной вперед и аппроксимируем уравнения системы

$$\dfrac{v_{i+1}-v_{i}}{\tau}=-k\sin \varphi_i,$$
$$\dfrac{\varphi_{i+1}-\varphi_{i}}{\tau}=v_i.$$

Правые части взяты в точке $x_i$, мы также могли бы записать их в точке $x_{i+1}$. Теперь мы можем выполнить расчет по формулам при $i=1, ... , N-1$

$$\varphi_0=\phi_0,\quad v_0=\omega_0.$$
$$v_{i+1}=v_{i}-\tau k\sin \varphi_i,$$
$$\varphi_{i+1}=\varphi_{i}+\tau v_i.$$

Тем самым мы сформулировали алгоритм, позволяющий за конечное количество количество операций найти приближенное решение задачи.

Однако пока совершенно неясно, с какой точностью получено решение, да и выполняемые операции совсем не похожи на методы, применяемые для решения обыкновенных дифференциальных уравнений. Исходная задача является **дифференциальной**, а ее решение - достаточно **гладкая функция** $\varphi(t)$. Та задача, на которую мы ее заменили, является **разностной**, а ее решение - **сеточная функция**. Для того, чтобы сравнить решение этих двух задач, вводится определение сходимости, но чтобы это сделать, сначала введем некоторые обозначения. Пусть $u$ - решение дифференциальной задачи

$$
\mathcal{L}u =\mathcal{f},
$$

а $y$ - решение дифференциальной задачи

$$
L_s y = f_{s}.
$$

Введем также $u_h$ -сеточную проекцию функции $u$.

**Опр 1. Сходимость** Будем говорить, что решение по разностной схеме $L_s y = f_{s}$ сходится к решению дифференциальной задачи $\mathcal{L}u =\mathcal{f}$, если

$$
\lVert u_h - y \rVert \to 0,\quad h \to 0
$$

Если кроме того

$$\lVert u_h - y \rVert \le Ch^{p},$$

то имеет место сходимость порядка $p$.


**Опр 2. Невязка** Невязкой уравнения $L_s y = f_{s}$ называется результат подстановки в разностное уравнение сеточной проекции точного решения дифференциальной задачи $\mathcal{L}u =\mathcal{f}$

$$
r_h = L_s u_h - f_s
$$

**Опр 3. Аппроксимация** Если $\lVert r_h \rVert \to 0,\; h \to 0$, то разностная схема аппроксимирует дифференциальную задачу.

Если кроме того $\lVert r_h \rVert \le C_1 h^{k}$,
то имеет место аппроксимация порядка $k$.


**Опр 4. Устойчивость** Говорят, что разностная схема устойчива, если для двух возмущенных задач справедливо

$$
\begin{aligned}
L_s y^{(1)} &= f_s + \varepsilon^{(1)},\\
L_s y^{(2)} &= f_s + \varepsilon^{(2)},\\
\|y^{(1)} - y^{(2)}\| &\le C_2\bigl(\|\varepsilon^{(1)}\| + \|\varepsilon^{(2)}\|\bigr)
\end{aligned}
$$


Обычно исследовать сходимость по определению возможно только для узкого класса задач, поэтому обычно для исследования сходимости пользуются теоремой, которая ввиду своей особой значимости называется основной теоремой вычислительной математики.

**Теорема.** Основная теорема вычислительной математики (теорема Лакса – Рябенького – Филиппова)

Пусть семейство разностных схем $L_s y = f_{s}$ аппроксимирует дифференциальную задачу $\mathcal{L}u =\mathcal{f}$ и устойчиво. Тогда решение разностной задачи сходится к решению дифференциальной, причем если аппроксимация имела порядок $p$, то и сходимость будет иметь порядок $p$.

# Простейшие разностные схемы для решения задачи Коши

Мы отдельно будем рассматривать рассматривать методы решения задачи Коши, краевой задачи, задачи Шутрма-Лиувилля, но в рамках этого семинара сосредоточимся на задаче Коши. Приведем примеры простейших разностных схем для задачи Коши для уравнения первого порядка

$$
\begin{cases}
u' = f(x, u),\\
u(0) = u_0.
\end{cases}
$$


**Явная схема Эйлера**

$$
\begin{cases}
\dfrac{y_{m+1} - y_m}{h} = f_m,\\[6pt]
y_0 = u_0
\end{cases}
$$

Первое уравнение системы представляет собой разностное уравнение первого порядка. Схема называется явной, так как решение можно выписать явно, именну эту схему мы использовали в примере выше.

**Неявная схема Эйлера**

$$
\begin{cases}
\dfrac{y_{m+1} - y_m}{h} = f_{m+1},\\[6pt]
y_0 = u_0
\end{cases}
$$

Первое уравнение системы также представляет собой разностное уравнение первого порядка. Схема называется неявной, так как решение в новой точке не может быть выписано явно, для его нахождения требуется решить уравнение, которое может быть линейным или нелинейным в зависимости от вида функции правой части.

**Схема с центральной разностью**

$$
\begin{cases}
\dfrac{y_{m+1} - y_{m-1}}{2h} = f_m,\\[6pt]
y_0 = u_0,\\
y_1 = v \;\text{ — надо корректно доопределить}
\end{cases}
$$

В первое уравнения входят уже три величины $y_{m-1}, \ y_{m}, \ y_{m+1}$, поэтому это разностное уравнение второго порядка. Для начала расчета требуется задать два начальных условия $y_0$ и $y_1$.


# Общее решение разностного уравнения

Проведем аналогию между способом решения линейных однородных дифференциальных уравнений с постоянными коэффициентами и способом решения линейных однородных разностных уравнений с постоянными коэффициентами. Рассмотрим простейший пример задачи Коши для дифференциального уравнения

$$
\begin{cases}
u' - a u = 0,\\
u(0) = 1,
\end{cases}
\qquad a \in \mathbb{R}
$$

Можно выписать характеристическое уравнение и решить его

$$
\lambda - a = 0 \;\Rightarrow\; \lambda = a
$$

Решение будет записываться в виде

$$
u = C e^{\lambda x}
$$

Если характеристическое уравнение имеет $l$ различных действительных корней, то решение запишется в виде

$$
u = C_1 e^{\lambda_1 x} + C_2 e^{\lambda_2 x}+ ... + C_l e^{\lambda_l x}.
$$



Теперь рассмотрим простейшую задачу для разностного уравнения

$$
\begin{cases}
y_{m+1} - k\,y_m = 0,\\
y_0 = 1,
\end{cases}
\qquad k \in \mathbb{R}
$$

Решение легко выписать вручную

$$
y_1 = k y_0 = k,\qquad
y_2 = k y_1 = k^2,\qquad
y_3 = k y_2 = k^3,\;\dots
$$

и обнаружить закономерность. Оказывается, аналогично можно выписать и решить характеристическое уравнение

$$
\lambda - k = 0 \;\Rightarrow\; \lambda = k
$$

И получить решение в виде

$$
y_m = C\,\lambda^m.
$$

В более общем случае наличия $l$ различных действительных корней решение записывается в виде

$$
y_m = C_1 \lambda_1^{\,m} + C_2 \lambda_2^{\,m} + ... + C_l \lambda_l^{\,m}.
$$

Конечно, сейчас мы рассмотрели только один пример, но аналогию можно распространить и на другие варианты (корни также могут быть кратными и не только действительными, но и комплексными). Об этом мы поговорим подробнее в следующем семестре.

# Задача 1

Выписать общее решение разностного уравнения – явной схемы Эйлера

$$
\begin{cases}
u' = a u,\\
u(0) = u_0
\end{cases}
\qquad
\begin{cases}
\dfrac{y_{m+1} - y_m}{h} = a y_m,\\[6pt]
y_0 = u_0
\end{cases}
$$


**Решение.** Выпишем характеристическое уравнение и найдем его решение

$$
y_{m+1} - y_m = a h\, y_m
$$

$$
y_{m+1} - (1 + a h)\, y_m = 0
$$

$$
\lambda - (1 + a h) = 0 \;\Rightarrow\; \lambda = 1 + a h
$$

Запишем решение в виде

$$
y_m = C(1 + a h)^{m}
$$

Используем начальное условие $ y_0 = u_0$ для нахождение константы $C$

$$
y_0 = C(1 + a h)^0 = C = u_0
$$

В итоге мы получили ответ
$$
y_m = u_0(1 + a h)^{m}
$$


# Задача 2

Найти решение разностного уравнения, удовлетворяющее начальным условиям

$$
\begin{cases}
\dfrac{y_{m+1} - y_m}{h} = y_m - y_{m-1},\\[6pt]
y_0 = 0,\quad y_1 = 1 - h
\end{cases}
$$


**Решение.** Преобразуем уравнение

$$
y_{m+1} - y_m = h y_m - h y_{m-1}
$$

$$
y_{m+1} - (1 + h)\,y_m + h\,y_{m-1} = 0
$$

Выпишем характеристическое уравнение и найдем его решение

$$
\lambda^{2} - (1 + h)\lambda + h = 0,\quad \lambda_1 = 1,\ \lambda_2 = h
$$

Общее решение однородного уравнения имеет вид
$$
y_m = C_1 + C_2 h^{m}
$$

Найдем константы $C_1$ и $C_2$ из начальных условий
$$
y_0 = 0,\quad y_1 = 1 - h
$$

Запишем систему уравнений
$$
C_1 + C_2 = 0
$$

$$
C_1 + C_2 h = 1 - h
$$

и решим ее
$$
C_1 = 1,\qquad C_2 = -1
$$

Окончательно
$$
y_m = 1 - h^{m}
$$


## Исследование аппроксимации

Исследование аппроксимации проводится по слеующему плану в соответствии с определением
1. выписать невязку;
2. разложить все величины в ряд Тейлора вблизи одной точки;
3. подставить всё в невязку и получить оценку

    $$\lVert r_h \rVert \le C h^{p},$$

    где $p$ — порядок аппроксимации.


# Задача 3

Исследовать явную схему Эйлера на аппроксимацию

$$
\begin{cases}
u' = f(x,u),\\
u(0) = u_0
\end{cases}
\qquad
\begin{cases}
\dfrac{y_{m+1} - y_m}{h} = f_m,\\[6pt]
y_0 = u_0
\end{cases}
$$


**Решение.** Действуем по плану

1) Выписываем невязку

$$r_h = \dfrac{u_{m+1} - u_m}{h} - f_m$$

2) Раскладываем $u_{m+1}$ в ряд Тейлора вблизи $x_m$

$$u_{m+1} = u_m + h\,u'_m + \dfrac{h^2}{2}\,u''_m + \dots$$

3) Подставляем разложение в невязку

$$\quad
r_h = \dfrac{u_m + h u'_m + \dfrac{h^2}{2}u''_m - u_m}{h} - f_m
     = u'_m + \dfrac{h}{2}u''_m - f_m
     = \dfrac{h}{2}u''_m
$$

и получаем треубуемую оценку

$$
\lVert r_h \rVert \le \dfrac{M_2}{2}\,h,\qquad M_2=\max |u''|
$$

$$
p = 1
$$

Схема имеет первый порядок **аппроксимации**

# Задача 4

Исследовать на аппроксимацию разностную схему. Определить, какую дифференциальную задачу она аппроксимирует.

$$
\frac{y_m + y_{m+1} - 2y_{m+3}}{5h} = y_m + y_{m+1}
$$

**Решение.** Выполняем те же три этапа исследования
1) Выписываем невязку

$$
r_h = \frac{u_m + u_{m+1} - 2u_{m+3}}{5h} - u_m - u_{m+1}$$

2) Выписываем разложения в ряд Тейлора

$$
u_{m+1} = u_m + h\,u'_m + \frac{h^2}{2}\,u''_m + \dots$$

$$
u_{m+3} = u_m + 3h\,u'_m + \frac{9h^2}{2}\,u''_m + \dots$$

3) Анализируем невязку

$$
r_h = \frac{1}{5h}\Bigl(u_m + u_m + h u'_m + \frac{h^2}{2}u''_m
      - 2\bigl(u_m + 3h u'_m + \frac{9h^2}{2}u''_m\bigr)\Bigr)
      - u_m - (u_m + h u'_m + \tfrac{h^2}{2}u''_m)$$


$$
r_h = -u'_m - 1.7\,h\,u''_m - 2u_m + u'_m h
$$

Исходную дифференциальную задачу выпишем из условия равенства нулю суммы всех слагаемых невязки, в которые не входит шаг сетки $h$

$$
-u' - 2u = 0
$$

В итоге получаем

$$
\|r_h\| \le C h,\qquad p = 1, \qquad C = 1.7 M_2 + M_1
$$

Схема обладает первым порядком **аппроксимации**.


## Исследование сходимости по определению

Выпишем план исследования сходимости по определению
1. выписать решение дифференциального уравнения
2. выписать решение разностного уравнения
3. получить оценку $\lVert u_m - y_m \rVert \le C h^{p}$, где $p$ — порядок сходимости.

# Задача 5

Доказать сходимость явной схемы Эйлера

$$
\begin{cases}
u' = a\,u,\\
u(0) = u_0
\end{cases}
\qquad
\begin{cases}
\dfrac{y_{m+1} - y_m}{h} = a\,y_m,\\[6pt]
y_0 = u_0
\end{cases}
$$


**Решение.** Выполним исследование сходимости в соответствии с планом
1) Решение дифференциального уравнения мы уже получали выше

$$ u = u_0 e^{a x}$$

2) Решение разностного уравнения мы получали в задаче 1
$$y = u_0(1 + a h)^{m}$$

3) Предварительно преобразуем выражение

$$(1 + a h)^{m}
= e^{\ln (1+ah)^{m}}
= e^{m\ln(1+ah)}
$$

Воспользуемся разложением в ряд Маклорена

$$
\ln(1+ah) \approx a h - \frac{a^{2}h^{2}}{2}
$$

И далее преобразуем выражение

$$
(1+ah)^{m}
= e^{m a h}\, e^{-\,m a^{2} h^{2}/2}
= e^{a m h}\, e^{-\,a^{2} h (m h)/2}
$$

Без ограничения общности положим, что сетка выбрана так, что

$$
x_m = m h
$$

Тогда

$$(1+ah)^{m} = e^{a x_m}\!\left(1 - \frac{a^{2}h\,x_m}{2}\right)$$

Оценим норму разности сеточной проекции решения дифференциальной задачи и решения разностной задачи

$$
\lVert u_m - y_m \rVert
= \left\lVert u_0 e^{a x_m} - u_0 e^{a x_m}\!\left(1 - \frac{a^{2}h\,x_m}{2}\right)\right\rVert
$$

$$
= \left\lVert u_0 e^{a x_m}\,\frac{a^{2}h\,x_m}{2} \right\rVert
\le C h
$$

$$
C = \left\lvert u_0 \frac{a^{2}X}{2} \right\rvert \max\{e^{aX},1\},
\qquad p = 1
$$

Схема первого порядка **сходимости**.

# Задача 6

Исследовать сходимость разностной схемы для решения задачи Коши

$$
\begin{cases}
u' = 3u,\\
u(0) = 1
\end{cases}
\qquad
\begin{cases}
\dfrac{2y_{m+1} - 3y_m + y_{m-1}}{h} = 6y_m - 3y_{m-1},\\[6pt]
y_0 = 1 + 3h,\quad y_1 = 1 + \dfrac{9h}{2}
\end{cases}
$$

**Решение.** Действуем по тому же алгоритму.
1) Найдем решение дифференциальной задачи

$$u = e^{3x}$$

2) Найдем решение разностной задачи. Преобразуем исходное уравнение

$$
2y_{m+1} - 3y_m + y_{m-1} = 6h\,y_m - 3h\,y_{m-1},
$$
$$
2y_{m+1} - (3+6h)y_m + (1+3h)y_{m-1} = 0.
$$

Выпишем характеристическое уравнение

$$
2\lambda^{2} - (3+6h)\lambda + (1+3h) = 0,
$$

и найдем его решения

$$
\lambda_1 = \frac12,\qquad \lambda_2 = 1 + 3h.
$$

Общее решение однороного разностного уравнения запишется в виде

$$
y_m = C_1\left(\frac12\right)^m + C_2(1+3h)^m
$$

Воспользуемся начальными условиями и найдем недостающие константы

$$
\begin{cases}
y_0 = 1 + 3h,\\
y_1 = 1 + \dfrac{9h}{2}
\end{cases}
\qquad
\begin{cases}
C_1 + C_2 = 1 + 3h,\\[4pt]
\dfrac{C_1}{2} + C_2(1+3h) = 1 + \dfrac{9h}{2}
\end{cases}
$$

$$
C_1 = 3h,\qquad C_2 = 1
$$

Окончательно решение разностного уранвения принимает вид

$$
y_m = 3h\left(\frac12\right)^m + (1+3h)^m
$$

3) Оценим норму разности

$$
\lVert u_m - y_m \rVert
$$

Воспользуемся формулой, полученной в прошлой задаче

$$
(1+3h)^m \approx e^{3x_m}\left(1 - \frac{9h}{2}x_m\right)
$$

И получим требуемую оценку

$$
\|u_m - y_m\|
 = \left\| e^{3x_m} - e^{3x_m}\!\left(1 - \frac{9h}{2}x_m\right)
      - 3h\left(\frac12\right)^m \right\|
$$

$$
= \left\| \frac{9h}{2}\,e^{3x_m} x_m - 3h\left(\frac12\right)^m \right\|
  \le C\,h
$$

$$
C = 3 + \frac{9}{2} e^{3X}, \qquad p = 1
$$

Первый порядок **сходимости**.
