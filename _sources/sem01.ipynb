# Введение

Как известно, в современной науке и технике огромную роль играют вычисления, производимые с помощью компьютера. С одной стороны, это связано с тем, что **в хорошо изученных к настоящему времени областях естествознания практически не осталось открытых точно решаемых задач**. Проблемы современной физики (для определённости будем говорить именно о физике) приводят, как правило, к математическим моделям столь высокой сложности, что аналитическое решение соответствующих уравнений оказывается невозможным. С другой стороны, **применение компьютерных вычислений значительно сокращает время и ресурсы, необходимые для решения возникающих на практике задач**. Например, известный из линейной алгебры алгоритм Гаусса позволяет, в принципе, решить "с помощью бумаги и ручки" систему линейных алгебраических уравнений с матрицей любого порядка $n$. Но подобные вычисления  отнимают много времени и сил уже при $n=10$. На практике же необходимо решать системы размерности $n\sim 10^6$ (представьте себе, что в кубе выбрана сетка из 100 точек по $x$, по $y$ и по $z$), а также системы большей размерности. Ясно, что такие задачи могут быть решены лишь на компьютере.

## Что изучает вычислительная математика?

**Вычислительная математика изучает методы решения задач непрерывной математики** посредством систем, выполняющих конечное число операций с конечными величинами (к таким системам относится и компьютер), а также **занимается проблемами обоснования корректности и оценки точности этих методов**.

Численное решение любой задачи сопряжено с целым рядом специфических трудностей, появляющихся из-за особенностей вычислений на компьютере.

1. **Ограниченность памяти компьютера**\
    Решение любой задачи на компьютере возможно найти с использованием лишь **конечного** набора значений переменных. Например, при вычислении интеграла возможно использовать лишь конечный набор значений подынтегральной функции. Решение задачи Коши получится найти только в конечном наборе точек на заданном отрезке. Подобный переход от непрерывной функции к функции, заданной на конечном наборе точек, называется **дискретизацией задачи**.


2. **Представление чисел в памяти компьютера**\
    Для хранения числа в памяти компьютера отводится конечное число бит, поэтому  вычисления производятся с числами, имеющими конечное количество знаков после запятой. При решении задач это приводит к дополнительным сложностям, которых не было в непрерывной математике.


3. **Зависимость результата от способа вычисления (выбора дискретизации)**\
    Многие понятия математики, такие как производная, интеграл и др., должны быть "переведены" с языка анализа бесконечно малых на язык **малых (но конечных) величин**, с которыми проводится **большое (но конечное) число операций**. Как показывает практика, способ "перевода" влияет на результат решения задачи. Отсюда и возникает проблема **обоснования** корректности метода и **выбора для данной задачи наилучшего** из методов (если их существует несколько). Интуитивно понятно, что при переходе от строгих определений математического анализа и работы с бесконечными десятичными дробями к алгоритмам вычислительной математики, состоящим из действий с числами, которые представимы в виде конечных десятичных дробей, неизбежны погрешности, сказывающиеся на результате, и поэтому искомое решение задачи может быть найдено только приближённо. Одна из целей вычислительной математики — выяснить, насколько точным является полученное приближение к решению и какими методами его можно улучшить.


Подчеркнём, что в круг вопросов, рассматриваемых вычислительной математикой, **не входит вопрос о реализации тех или иных приближённых методов на конкретных устройствах**. Как запрограммировать тот или иной алгоритм на выбранном языке, как ускорить выполнение полученной программы, какие готовые структуры языка (библиотеки, модули и пр.) можно использовать — всё это вопросы **научного программирования**, но не вычислительной математики. Тем не менее в рамках курса будут предлагаться как теоретические задачи, например, на оценку точности какого-либо метода, так и практические задачи, заключающиеся в программной реализации некоторого метода.

## Модель — алгоритм — программа

Роль вычислительной математики в процессе описания окружащего мира можно проиллюстрировать с помощью триады **"модель — алгоритм — программа"**, предложенной известным отечественным математиком А.А. Самарским ([ссылка](https://lib.mipt.ru/book/14774/?q=Самарский)). Поясним, о чём идёт речь, с помощью простейшего примера — задачи о колебании математического маятника.

![image](pic/sem01_fig.png)

*Триада "модель — алгоритм — программа" А.А. Самарского*

**1. Модель.**

Отправной точкой для всех теоретических построений является само явление колебания, наблюдаемое в природе. **Создание модели** явления **не является задачей вычислительной математики**, это задача предметной области, в данном случае — механики. Из механики известно, что колебание материальной точки на невесомой нерастяжимой нити (без трения) может быть описано задачей Коши:

$$\ddot\varphi+k\sin\varphi=0,\quad \varphi(0)=\phi_0, \quad \dot\varphi(0)=\omega_0.$$

Предметная область (механика) даёт объяснение смысла входящих в эту задачу величин: $\varphi=\varphi(t)$  — угол отклонения маятника от вертикали, $t$ — время, $k=g/l$, где $l$ — длина маятника, а $g$ — ускорение свободного падения. Этот смысл для вычислительной математики мало интересен (существенны, пожалуй, лишь области определения и значения входящих в уравнения параметров и функций, которые задаются как раз их физическим смыслом). Более важно то, что сформулирована модель — дано конкретное уравнение, которое надо решать численно (на компьютере). Существенно, что **для одного и того же явления можно предлагать различные модели**. Например, для описания **малых** колебаний маятника вместо выписанной выше задачи Коши можно использовать задачу

$$\ddot\varphi+k\varphi=0,\quad \varphi(0)=\phi_0, \quad \dot\varphi(0)=\omega_0.$$

Какую модель применить в данном конкретном случае — это вопрос, решаемый предметной областью. Вычислительная математика "вступает в игру" только после того, как рассматриваемая задача сформулирована в математических терминах (обычно в виде некоторых уравнений).

**2. Алгоритм.**

Чтобы дать **алгоритм** решения задачи, вычислительная математика проводит дискретизацию входящих в задачу объектов и предлагает некий **метод** получения искомого ответа. Например, непрерывное множество значений независимой переменной $t\in[0,+\infty)$ заменяется конечным **набором узлов** (одномерной сеткой) $\{t_i\}_{i=0}^N$ с шагами $\tau_i=t_{i+1}-t_i$, причём $t_0=0$, а $t_N=T$. Непрерывная функция $\varphi=\varphi(t)$ заменяется набором значений $\{\varphi_i\}_{i=0}^N$ в узлах сетки (**сеточной функцией**). Ниже мы для простоты будем считать, что $\tau_0=...=\tau_{N-1}\equiv\tau$ (**равномерная сетка**), хотя это и не обязательно.

Далее встаёт вопрос об **аппроксимации** ("переводе" на дискретный язык) производной. Известное из анализа определение

$$\dot\varphi(t)=\lim_{\tau\to 0}\frac{\varphi(t+\tau)-\varphi(t)}{\tau}$$

адаптируется для дискретного случая, например, следующим образом (так называемая **производная вперёд**):

$$\dot\varphi_i=\frac{\varphi_{i+1}-\varphi_i}{\tau}.$$

Тогда вторую производную можно задать как

$$\ddot\varphi_i=\frac{1}{\tau}\bigg[\frac{\varphi_{i+1}-\varphi_i}{\tau}-\frac{\varphi_{i}-\varphi_{i-1}}{\tau}\bigg]=\frac{\varphi_{i+1}-2\varphi_i+\varphi_{i-1}}{\tau^2}.$$

В итоге получается **разностное уравнение**, связывающее значения сеточной функции $\varphi_i$ в соседних узлах сетки:

$$\frac{\varphi_{i+1}-2\varphi_i+\varphi_{i-1}}{\tau^2}+k\sin\varphi_i=0.$$

 Это уравнение записывается для всех внутренних точек расчетной области и дополняется уравнениями, аппроксимирующими начальные условия. Полученная система уравнений решается на компьютере вместо дифференциального уравнения.



Дальнейшие шаги по аппроксимации начальных условий, решению системы уравнений и т.п. мы сейчас рассматривать не будем. Мы не будем приводить здесь и оценок точности получаемого решения. Нам важно было указать саму идею и подчеркнуть, что целью вычислительной математики является создание **математически обоснованного конечного алгоритма действий с некоторым конечным набором чисел**. После того, как алгоритм создан и дана оценка точности результата, задачу вычислительной математики можно считать выполненной.

## Численное решение задачи Коши

Приведём пример численного решения первой из обсуждавшихся выше задач Коши. Для этого перепишем входящее в неё уравнение второго порядка в виде эквивалентной системы двух зацепленных уравнений первого порядка (этот приём известен из курса дифференциальных уравнений):

$$\begin{cases}
    \dot\varphi = f,\\
    \dot f = -k\sin\varphi.
  \end{cases}$$

Тогда в векторных обозначениях задача Коши выглядит так:

$$
  \dot{\textbf{x}}=\textbf{F},\quad \textbf{x}(0)=(\phi_0, \omega_0)^\text{T},\quad \textbf{F}=(x_2, -k\sin{x_1})^\text{T}.
$$

Системы такого вида можно решать численно с помощью функции ```odeint``` модуля ```scipy.integrate``` библиотеки ```scipy``` ([документация](https://docs.scipy.org/doc/scipy-1.16.0/reference/generated/scipy.integrate.odeint.html)):

````python
import numpy as np
from scipy.integrate import odeint

# функция правой части
def F(x: np.ndarray, t: np.ndarray, k: float):
    return np.array([x[1], -k*np.sin(x[0])])

# функция, решающая задачу Коши
def solve_Cauchy(phi_0: float, omega_0: float, k: float, t: np.ndarray):
    x_0 = np.array([phi_0, omega_0]) # начальные условия

    return odeint(F, x_0, t, args=(k,))
````

В качестве результата мы получаем массив из векторов $\textbf{x}_i=\textbf{x}(t_i)$, где $t_i$ пробегает значения из выбранной сетки по времени. Результат можно визуализировать, построив графики зависимостей $\varphi=\varphi(t)$ и $\dot\varphi=\dot\varphi(t)$, а также фазовый портрет системы в плоскости $(\varphi(t),\dot\varphi(t))$. Запрограммируем функцию для построения графиков:

````python
import matplotlib.pyplot as plt

# функция для построения графиков
def graph_Cauchy(solution: np.ndarray, t: np.ndarray):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    plt.subplots_adjust(wspace=0.3)

    ax1.plot(t, solution[:,0])
    ax2.plot(t, solution[:,1])

    ax1.set_title("Закон движения математического маятника")
    ax2.set_title("Угловая скорость математического маятника")

    ax1.set_xlabel("$t$")
    ax2.set_xlabel("$t$")

    ax1.set_ylabel("$\\phi(t)$")
    ax2.set_ylabel("$\\dot\\phi(t)$")

    ax1.grid()
    ax2.grid()

    plt.show()

    fig, ax = plt.subplots(figsize=(5, 5))

    ax.plot(solution[:,0], solution[:,1])

    ax.set_title("Фазовый портрет математического маятника")
    ax.set_xlabel("$\\phi(t)$")
    ax.set_ylabel("$\\dot\\phi(t)$")

    ax.grid()

    plt.show()
````

Создадим интерактивные графики, пользуясь средствами библиотеки ```ipywidgets``` ([документация](https://ipywidgets.readthedocs.io/en/latest/)).

````python
from ipywidgets import interact, FloatSlider

# ползунки для интерактивного графика (виджета)
phi0_w = FloatSlider(min=0, max=2*np.pi, step=0.05, value=np.pi/6)
omega0_w = FloatSlider(min=-4.0, max=4.0, step=0.05, value=0)
k_w = FloatSlider(min=0, max=10.0, step=0.05, value=1)

def pendulum(phi_0: float, omega_0: float, k: float):
    grid_t = np.linspace(0, 50, 1000) # сетка по времени
    sol = solve_Cauchy(phi_0=phi_0, omega_0=omega_0, k=k, t=grid_t)
    graph_Cauchy(sol, grid_t)

interact(pendulum, phi_0=phi0_w, omega_0=omega0_w, k=k_w);
````

Можно анимировать движение маятника. Для этого учтём, что уравнение траектории маятника в параметрической форме записывается так (начало координат помещено в точке подвеса):

$$
  \begin{cases}
    \displaystyle x=l\sin\varphi(t)=\frac{g}{k}\sin\varphi(t),\\[0.1cm]
    \displaystyle y=-l\cos\varphi(t)=-\frac{g}{k}\cos\varphi(t).
  \end{cases}
$$

Полезно знать, что основные физические константы можно импортировать из модуля ```scipy.constants``` библиотеки ```scipy``` ([документация со списком констант](https://docs.scipy.org/doc/scipy-1.16.0/reference/constants.html)).

````python
import matplotlib.animation as animation
from IPython.display import HTML

from scipy.constants import g

def create_animation(phi_0: float, omega_0: float, k: float):
    grid_t = np.linspace(0, 50, 1000) # сетка по времени
    sol = solve_Cauchy(phi_0=phi_0, omega_0=omega_0, k=k, t=grid_t)

    l = g / k # длина маятника
    lim = 1.25 * l # параметр для установки пределов по осям (чтобы маятник помещался)

    fig, ax = plt.subplots(figsize=(5, 5))
    ax.set_xlim( (-lim, lim) )
    ax.set_ylim( (-lim, lim) )

    plt.xticks([]) # убираем метки на осях
    plt.yticks([])

    x_p0 = g / k * np.sin(sol[0, 0]) # начальная точка расположения маятника
    y_p0 = - g / k * np.cos(sol[0, 0])

    pend = plt.Circle((x_p0, y_p0), l/20, color='k') # создаём объект маятника (круг)
    ax.add_artist(pend) # добавляем маятник на холст

    line, = ax.plot([0, x_p0], [0, y_p0], color='k') # рисуем нить с подвесом в начале координат

    def animate(i: int):
        x_p = g / k * np.sin(sol[i, 0]) # вычисляем координаты маятника для текущего кадра
        y_p = - g / k * np.cos(sol[i, 0])

        pend.center = (x_p, y_p) # отрисовываем
        line.set_data([0, x_p], [0, y_p])

        return line,

    ani = animation.FuncAnimation(fig, animate, interval=20, blit=True, frames=300)
    plt.close()

    return ani

HTML(create_animation(phi_0=np.pi/6, omega_0=0, k=1).to_jshtml()) # конвертируем в JavaScript
````

## Для чего нужно знать вычислительную математику?

Выше мы привели пример численного решения задачи Коши, описывающей движение математического маятника. При составлении программы мы воспользовались **готовой реализацией** алгоритма интегрирования системы ОДУ первого порядка. Нам не пришлось вникать в сущность этого алгоритма или знакомиться с теорией и принципами, на которых он построен. Мы воспользовались им как "чёрным ящиком", выдающим ответ по исходным данным.

Точно так же можно (и нужно!) поступать при решении любой другой прикладной задачи (если, конечно, разработка методов численного решения задачи не является основной целью работы). Подавляющее большинство численных методов, разработанных вычислительной математикой, уже имеет готовую реализацию или даже несколько реализаций в виде специализированных программных пакетов (по крайней мере, про методы, изучающиеся в нашем курсе, это можно утверждать совершенно точно). Современному физику-исследователю нет никакой необходимости самостоятельно писать программу, которая решала бы встретившуюся ему  математическую задачу.

В связи с этим возникает вопрос: **для чего нужно знать вычислительную математику**? Зачем знакомиться с принципами построения давно реализованных методов, если изучение этих методов не являются самоцелью?

Оказывается, что для грамотного применения готовых вычислительных программ (функций, модулей, библиотек, написанных на каком-либо языке) необходимо иметь хотя бы **общее представление** о принципах их работы. В противном случае затруднительно сделать осознанный выбор в пользу той или иной программной реализации из ряда внешне похожих. Кроме того, невозможно правильно истолковать отрицательный результат работы всей программы. Действительно, предположим, что некий готовый алгоритм является составной частью длинного кода, выдающего ошибку при попытке запуска (на практике подобные ситуации встречаются на каждом шагу!). Как понять, связана ли эта ошибка с неверно написанной программой или же дело в том, что **при правильно составленной программе** некорректно "работает" сам численный метод? Ситуация осложняется, если последовательно используются несколько разных вычислительных алгоритмов, так что результаты работы одного яляются входными данными для другого.

Ещё более запутанным является иной случай: программа запускается и выдаёт результат, отличный от ожидаемого. Как определить, является ли он ошибочным или правильным? Может быть, модель явления демонстрирует какое-то новое, ранее неизвестное свойство, т.е. никакой ошибки нет (совершено открытие)? Если же допущена ошибка, то в чём она состоит: в выборе неприменимого метода или в неверно написанной программе?

Очевидно, что анализировать описанные выше **нестандартные случаи** без понимания сути происходящего чрезвычайно трудно. А ведь смысл научной работы состоит именно в таком анализе! Знание же основных теоретических положений, связанных с конкретным численным методом, позволяет при необходимости разобраться в любой готовой программе, предлагающей реализацию этого метода. **Понимание** идей и принципов, стоящих за строками кода (с которыми и предлагается ознакомиться в рамках курса вычислительной математики во время учёбы), сэкономит время и силы при всей дальнейшей научной работе.

В подтверждение сказанного приведём два простых примера, в которых **использование готовых численных алгоритмов приводит к неверным результатам при правильно написанных программах**. Идеи примеров заимствованы из лекции А.В. Чикиткина: [cсылка](https://youtu.be/p2g3dY3nQx4?si=SaOzE7h2488VKKjM&t=659) (YouTube).

## Пример 1

Пусть требуется численно решить систему линейных алгебраических уравнений (СЛАУ)

$$\textbf{A}\textbf{x}=\textbf{f},$$

где $\textbf{A}$ — невырожденная матрица порядка $n$. Нас будет интересовать матрица специального вида — так называемая **матрица Гильберта** (на доказательстве того, что она является невырожденной, мы не останавливаемся):

$$\textbf{A}=(a_{ij}),\quad a_{ij}=\frac{1}{i+j-1},\quad 1\leq i, j\leq n.$$

Чтобы получить СЛАУ с выбранной матрицей и известным точным решением, зададим это решение вручную и вычислим соответствующую правую часть. Для простоты в качестве точного решения возьмём единичный вектор.

Напишем функцию, создающую матрицу Гильберта $\textbf{A}$ заданного порядка $n$ и определяющую вектор правой части $\textbf{f}$ так, чтобы точным решением был единичный вектор. В коде ниже при создании матрицы знак "плюс" в знаменателе дроби поставлен потому, что в языке Python индексирование элементов массива начинается с нуля, а не с единицы.

````python
import numpy as np

def create_SLE(n: int):
    A = np.fromfunction(lambda i, j : 1 / (i + j + 1), (n, n))
    f = A @ np.ones(n) # матричное умножение

    return A, f
````

Теперь мы можем легко задать СЛАУ из произвольного числа уравнений $n$ с матрицей выбранного нами вида и единичным вектором в качестве точного решения. Напишем аналогичную функцию для задания СЛАУ такой же размерности, но со случайной матрицей, элементы которой принадлежат отрезку $[0,1]$:

````python
def create_SLE_rand(n: int):
    A = np.random.rand(n, n) # матрица случайных чисел из отрезка [0, 1]
    f = A @ np.ones(n)

    return A, f
````

Будем численно решать обе системы при различных $n$ с помощью функции ```solve``` из модуля ```linalg``` библиотеки ```numpy``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html)). Так как нам известно точное решение (единичный вектор), мы можем контролировать норму ошибки (норму вектора разности численного и точного решений). Для расчета выберем кубическую норму, которая для вектора $\textbf{u}$ определяется как $||\textbf{u}||=\displaystyle\max_i |u_i|$ и принадлежит к классу $p$-норм векторов, $p=\infty$ (см. третий семинар).

````python
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
n_w = IntSlider(min=1, max=100, step=1, value=3)

def compare_solutions(n: int):
    A, f = create_SLE(n)
    x = np.linalg.solve(A, f)
    eps1 = np.linalg.norm(x-np.ones(n), ord=np.inf) # кубическая норма разности решений

    A, f = create_SLE_rand(n)
    x = np.linalg.solve(A, f)
    eps2 = np.linalg.norm(x-np.ones(n), ord=np.inf)

    print("Специальная матрица. Ошибка: {:.3e}".format(eps1))
    print("Случайная матрица. Ошибка: {:.3e}".format(eps2))

interact(compare_solutions, n=n_w);
````

Как видим, уже при $n=3$ погрешность решения СЛАУ с матрицей Гильберта превосходит погрешность решения СЛАУ со случайной матрицей почти на два порядка. При дальнейшем увеличении $n$ первая ошибка стремительно нарастает и достигает величины $\sim 10^2$ при $n=100$ (т.е. превосходит норму точного решения на два порядка!). Величина же второй ошибки остаётся на уровне $\sim 10^{-14}$ (уровень ошибок округления, см. ниже) во всей области изменения $n$. Наблюдаемый эффект объясняется **плохой обусловленностью** матрицы Гильберта (см. третий семинар). Таким образом, один и тот же готовый алгоритм (одна и та же функция ```np.linalg.solve```) даёт разные результаты в разных случаях, причём причину различий следует искать не на уровне программной реализации, а глубже — в свойствах системы и метода её решения (в рассматриваемой функции реализованы **метод** последовательных исключений **Гаусса** и LU-разложение).

## Пример 2

Пусть дана таблица значений некоторой функции $y=f(x)$ в **конечном упорядоченном** наборе точек:

$$\begin{align}
  x_0,\quad  x_1,\quad ...,\quad  x_{n};\\
  y_0,\quad  y_1,\quad ...,\quad  y_{n}.\\
\end{align}$$

Требуется восстановить гладкую функцию $f(x)$ на отрезке $[x_0,x_{n}]$ так, чтобы её график прошёл через все заданные точки (т.е. решить задачу **точной интерполяции**).

Существует несколько способов интерполирования функций. Все они сводятся к построению **обобщённого многочлена**

$$P_n(x)=\sum\limits_{k=0}^nc_k\varphi_k(x)\quad \text{такого, что}\quad P_n(x_k)=y_k,\quad 0\leq k\leq n.$$

Здесь $\varphi_k(x)$ — некоторые фиксированные функции. Если $\varphi_k(x)=x^k$, то интерполяция называется **алгебраической**.

Оказывается, что если таблица из $n+1$ значения функции не содержит повторяющихся узлов, то существует один и только один полином степени не выше $n$, проходящий через все заданные точки.

Построим таблицу значений функции из $n+1$ узла. Узлы $x_k$ выберем некоторым специальным образом, а значения $y_k$ возьмём случайными из отрезка $[0,1]$.

````python
import numpy as np

def create_table(n: int):
    x = np.pi * np.linspace(-1.0, 0.0, n+1, endpoint=False) # задаём узлы
    x = np.cos(x)
    x = np.sort(x) # упорядочиваем узлы по возрастанию

    y = np.random.rand(n+1)

    return x, y
````

Построим интерполяционный полином двумя способами: с помощью функции ```polyfit``` библиотеки ```numpy``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.polyfit.html)) и с помощью класса ```BarycentricInterpolator``` модуля ```scipy.interpolate``` библиотеки ```scipy``` ([документация](https://docs.scipy.org/doc/scipy-1.16.0/reference/generated/scipy.interpolate.BarycentricInterpolator.html#scipy.interpolate.BarycentricInterpolator)).

````python
import matplotlib.pyplot as plt
from scipy.interpolate import BarycentricInterpolator

def poly_with_numpy(x: np.ndarray, y: np.ndarray):
    poly = np.polyfit(x, y, x.size)

    grid = np.linspace(np.min(x), np.max(x), 1000) # сетка по x для построения графика
    y_graph = np.polyval(poly, grid)

    plt.plot(x, y, 'ro', ms=5)
    plt.plot(grid, y_graph, 'k-')
    plt.title("Интерполяция с помощью numpy")
    plt.show()

def poly_with_scipy(x: np.ndarray, y: np.ndarray):
    poly = BarycentricInterpolator(x, y)

    grid = np.linspace(np.min(x), np.max(x), 1000)
    y_graph = poly(grid)

    plt.plot(x, y, 'ro', ms=5)
    plt.plot(grid, y_graph, 'b-')
    plt.title("Интерполяция с помощью scipy")
    plt.show()
````

Будем строить графики полученных полиномов (**интерполянтов**) при разных $n$:

````python
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
n_w = IntSlider(min=1, max=50, step=1, value=10)

def graph_poly(n: int):
    x, y = create_table(n)
    poly_with_numpy(x, y)
    poly_with_scipy(x, y)

interact(graph_poly, n=n_w);
````

Уже при сравнительно небольших $n$ (порядка 10) на графиках заметны локальные различия. При $n\sim 40$ результаты работы функции ```np.polyfit``` оказываются неудовлетворительными: появляются точки, через которые не проходит график интерполяционного многочлена, причём с увеличением $n$ количество таких точек растёт. В то же время полином, получаемый при помощи класса ```BarycentricInterpolator```, правильно решает поставленную задачу при всех рассматриваемых $n$.

Дело в том, что использованные библиотеки реализуют разные методы интерполяции. Функция ```np.polyfit``` строит интерполяционный многочлен по **методу наименьших квадратов**, а класс ```BarycentricInterpolator``` строит интерполяционный **полином Лагранжа** в модифицированом (барицентрическом) виде. Последний метод обеспечивает свойство интерполяционности по построению, а первый даёт лишь наилучшее приближение к заданным точкам (в смысле минимизации суммы квадратов отклонений от них).

# Погрешности вычислений

Пусть в качестве результата измерения или численного расчёта получена некоторая величина $u^*$. Обозначим **точное** (и заведомо неизвестное) значение этой же величины через $u$.

**Определение 1.** *Абсолютной погрешностью* величины $u^*$ называют *наименьшую* величину $\Delta(u^*)$ такую, что выполняется неравенство

$$|u-u^*|\leq \Delta(u^*).$$

**Определение 2.** *Относительной погрешностью* величины $u^*$ называют *наименьшую* величину $\delta(u^*)$ такую, что выполняется неравенство

$$\frac{|u-u^*|}{|u^*|}\leq \delta(u^*).$$

Иначе говоря, $\delta(u^*)=\Delta(u^*)/|u^*|$.

На практике воспользоваться этими определениями для вычисления $\Delta(u^*)$ или $\delta(u^*)$ невозможно, поскольку точное значение величины $u$ никогда не бывает известным. В связи с этим приходится прибегать к **оценкам погрешностей**.

## Оценки погрешностей

Пусть известно, что величина $u$ есть функция некоторого параметра $t$: $u=u(t)$  причём $u^*=u(t^*)$. Погрешность $\Delta(t^*)$ предполагается известной (либо оценённой ранее). Требуется дать оценку на $\Delta(u^*)$.

Обычно используют три оценки погрешностей:
* Предельная абсолютная погрешность:

  $$D(u^*)=\sup_{t\in\Omega} |u(t)-u^*|,\quad \Omega=\{t\in \mathbb{R}: |t-t^*|\leq \Delta (t^*)\}.$$

* Погрешность по точной верхней грани производной:

  $$D_1(u^*)=\sup_{\theta\in\Omega} |u'(\theta)|\Delta(t^*).$$

* Линейная оценка:

  $$D_2(u^*)= |u'(t^*)|\Delta(t^*).$$

Происхождение последних двух оценок легко понять, если вспомнить **теорему Лагранжа** для дифференцируемой функции $u(t)$, обычно изучаемую в курсе математического анализа 1 семестра.

**Теорема Лагранжа.** Пусть функция $u(t)$ дифференцируема на отрезке $[t^*, t]$. Тогда существует внутренняя точка $\theta$ этого отрезка, для которой

$$u'(\theta)=\frac{u(t^*)-u(t)}{t^*-t}.$$

Иначе говоря, $u(t)-u(t^*)=u'(\theta)(t-t^*)$. В таком виде теорему Лагранжа называют **формулой конечных приращений** или **теоремой о среднем значении**.

Отметим, что эту теорему можно рассматривать как частный случай формулы Тейлора с остаточным членом в форме Лагранжа:

$$u(t)=\sum\limits_{k=0}^{n}\frac{u^{(k)}(t^*)}{k!}(t-t^*)^k+\frac{u^{(n+1)}(\theta)}{(n+1)!}(t-t^*)^{n+1},\quad \theta\in[t^*, t].$$

Применим теорему Лагранжа дважды:

$$u(t^*)-u\big(t^*+\Delta (t^*)\big)=u'(\theta_1)\Delta (t^*),\quad \theta_1 \in[t^*, t^*+\Delta (t^*)],$$

$$u\big(t^*-\Delta (t^*)\big)-u(t^*)=u'(\theta_2)\Delta (t^*),\quad \theta_2 \in[t^*-\Delta (t^*), t^*].$$

Введём обозначения:

$$\Delta_1\equiv |u'(\theta_1)|\Delta (t^*),\quad\Delta_2\equiv |u'(\theta_2)|\Delta (t^*).$$


Тогда очевидно, что

$$D_1(u^*)=\max(\Delta_1,\Delta_1)=\sup_{\theta\in\Omega} |u'(\theta)|\Delta(t^*).$$

Оценку $D_2(u^*)$ можно рассматривать как упрощённую версию оценки $D_1(u^*)$, в которой вместо точной верхней грани производной берётся её значение при $t=t^*$.

## Задача 1

Найти предельную абсолютную погрешность, погрешность по производной и линейную оценку погрешности для функции $u=t^{10}$, если заданы точка приближения $t^*=1$, значение функции $u^*$ в этой точке и погрешность $\Delta (t^*)=0,1$.

**Решение** состоит в использовании приведённых выше формул для оценок погрешностей. Так как $u^*=u(t^*)=1$, а $u'(t)=10t^9$, последовательно находим:

$$D(u^*)=\sup_{|t-1|\leq 0,1}|t^{10}-1|=1,1^{10}-1=1,59,$$

$$D_1(u^*)=\sup_{|t-1|< 0,1}|10t^9|\cdot 0,1=10\cdot 1,1^9\cdot 0,1=2,36,$$

$$D_2(u^*)=|10t^9|_{t=1}\cdot 0,1=10\cdot 0,1=1.$$

## Погрешность неявной функции

Пусть функция $u=u(\textbf{t})$, $\textbf{t}\in \mathbb{R}^n$ задана неявно уравнением $F(u,\textbf{t})=0$. Требуется оценить погрешность $\Delta(u^*)$, если заданы точка приближения $\textbf{t}^*$ и погрешности $\Delta(t_i^*)$.

Воспользуемся обобщением линейной оценки на случай функции многих переменных:

$$D_2(u^*)=\sum\limits_{i=1}^n|u'_{t_i}(t^*_i)|\Delta(t_i^*).$$

Частные производные функции $u$ найдём, продифференцировав уравнение по $t_i$:

$$F'_u\cdot u'_{t_i}+F'_{t_i}=0.$$

Отсюда

$$u'_{t_i}(t^*_i)=-\frac{F'_{t_i}(u^*,\textbf{t}^*)}{F'_u(u^*,\textbf{t}^*)}.$$

Искомая оценка:

$$D_2(u^*)=\big|F'_u(u^*,\textbf{t}^*)\big|^{-1}\cdot\sum\limits_{i=1}^n\big|F'_{t_i}(u^*,\textbf{t}^*)\big|\Delta(t_i^*).$$

Величина $u^*$ может быть найдена как корень уравнения $F(u,\textbf{t}^*)=0$ в предположении выполнения условий теоремы о неявной функции.

## Задача 2

Оценить погрешность в определении действительного корня уравнения $ax^3+bx^2+cx+d=0$, коэффициенты которого заданы приближённо: $a^*=c^*=1,\quad b^*=d^*=-2,\quad \Delta(a^*)=\Delta(b^*)=\Delta(c^*)=\Delta(d^*)=10^{-4}\equiv \Delta$.

**Решение.** Найдём приближённое значение корня. Уравнение имеет вид:

$$x^3-2x^2+x-2=0\quad \text{или}\quad (x^2+1)(x-2)=0.$$

Действительный корень есть $x^*=2$.

Пусть $F=F(x,a,b,c,d)=ax^3+bx^2+cx+d$. Согласно полученной выше формуле,

$$D_2(x^*)=\big|F'_x(*)\big|^{-1}\Big[|F'_a(*)|\Delta(a^*)+|F'_b(*)|\Delta(b^*)+|F'_c(*)|\Delta(c^*)+|F'_d(*)|\Delta(d^*)\Big].$$

Здесь для краткости записей мы положили $F'_x(*)=F'_x(x^*,a^*,b^*,c^*,d^*)$ и т.д. В нашем случае абсолютные погрешности коэффициентов одинаковы и равны $\Delta$, поэтому

$$D_2(x^*)=\big|F'_x(*)\big|^{-1}\cdot\Delta\cdot\Big[|F'_a(*)|+|F'_b(*)|+|F'_c(*)|+|F'_d(*)|\Big].$$

Остаётся найти частные производные функции $F$ при значениях параметров "со звёздочкой".

$$F'_x=3ax^2+2bx+c,\quad F'_x(*)=3\cdot 1\cdot 2^2+2\cdot (-2)\cdot 2 + 1=5;$$

$$\begin{align}
  F'_a=x^3,&\quad F'_a(*)=8;\qquad F'_b=x^2,\quad F'_b(*)=4;\\
  F'_c=x,&\quad F'_c(*)=2;\qquad F'_d=1,\quad F'_d(*)=1.
\end{align}$$

Искомая оценка:

$$D_2(x^*)=5^{-1}\cdot 10^{-4}\cdot\Big[8+4+2+1\Big]=3\cdot 10^{-4}.$$

## Классификация погрешностей

Во введении была упомянута триада А.А. Самарского "модель — алгоритм — программа", отражающая суть численных методов решения практических задач. Понятно, что погрешности (ошибки) могут возникнуть на каждом из трёх этапов решения. Ответ, выдаваемый компьютером, может оказаться далёк от точного (заведомо неизвестного) ответа из-за неудачного выбора исходной модели, использования неверного (не применимого в рассматриваемом случае) алгоритма (метода решения) или ошибок в программе (опечаток, "багов" и пр.). Однако такой взгляд на источники погрешностей вычислений является слишком общим.

В вычислительной математике принята следующая классификация погрешностей вычислений. Погрешности делятся на
* **неустранимые**, к которым относятся ошибки входных данных и ошибки модели,
* **погрешности метода**,
* **погрешности округления**.

Пример **ошибки модели** мы, по сути, привели выше, когда обсуждали описание колебаний математического маятника с помощью двух задач Коши. Вторая модель

$$\ddot\varphi+k\varphi=0,\quad \varphi(0)=\phi_0, \quad \dot\varphi(0)=\omega_0$$

применима в случае малых колебаний. Ошибка, связанная с заменой $\sin\varphi$ на $\varphi$, будет тем больше, чем больше угол $\varphi$. С точки зрения вычислительной математики, это — **неустранимая** ошибка: её принципиально невозможно исключить, поскольку нельзя изменить модель. Модель строит и обосновывает не вычислительная математика, а предметная область. Вычислительная математика работает "с тем, что есть".

Приведём примеры других ошибок.

## Ошибка входных данных

Пусть решается система линейных алгебраических уравнений (СЛАУ)

$$
  \begin{cases}
    0,99x_1 + x_2 =1,\\
    x_1 + 0,99x_2=1.
  \end{cases}
$$

Решение можно найти аналитически:

$$
  x_1=x_2=\frac{100}{199}\simeq 0,5025.
$$

Для численного решения можно воспользоваться функцией ```solve``` из модуля ```linalg``` библиотеки ```numpy``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html)):



````python
import numpy as np

A = np.array( [ [0.99, 1.00], [1.00, 0.99] ] ) # матрица системы
b = np.array( [1.00, 1.00] ) # вектор правых частей

x = np.linalg.solve(A, b) # решение системы

np.set_printoptions(precision=4) # количество десятичных знаков при выводе
print(x)
````

Изменим правые части уравнений системы на малую величину $\varepsilon=10^{-2}$ (будем считать, что эта ошибка определяется, например, точностью некоторого эксперимента):

$$
  \begin{cases}
    0,99x_1 + x_2 =1,01,\\
    x_1 + 0,99x_2=0,99.
  \end{cases}
$$

Снова находим решение (для примера — аналитически и численно):

$$
  x_1^*=-\frac{99}{199}\simeq -0,4975,\quad x_2^*=\frac{299}{199}\simeq 1,5025.
$$

````python
eps = 10**(-2)
b = np.array( [1.00 + eps, 1.00 - eps] )

x_eps = np.linalg.solve(A, b)

print(x_eps)
````

Изменение вектора решения составило $\Delta\textbf{x}=(-1, 1)^\text{T}$:

````python
delta_x = x_eps - x
delta_x
print(delta_x)
````

Заметим, что знак первого корня изменился на противоположный, а модуль второго корня увеличился в три раза. Такие изменения по "физическим меркам" очень значительны.

Как видим, погрешность входных данных (в данном случае — вектора правых частей уравнений системы) привела к погрешности решения. Ошибка $\varepsilon$ является **неустранимой** в том смысле, что её влияние никак нельзя уменьшить или исключить: для этого потребовалось бы точнее задавать правые части, т.е. вмешиваться в эксперимент и предлагать способ произвести более точные измерения. Но подобное вмешательство выходит за рамки задач, решаемых вычислительной математикой. Занимаясь численными методами, можно только оценить погрешность результата, связанную с ошибкой входных данных. Как выполняются подобные оценки, мы рассматривали выше.

Итак, в данном случае малое по относительной величине изменение правых частей СЛАУ привело к существенному изменению решения. Другими словами, данная СЛАУ демонстрирует высокую чувствительность к погрешности входных данных. В подобных случаях говорят, что решаемая задача **плохо обусловлена**. К понятию обусловленности мы ещё вернёмся на третьем семинаре.

Отметим, что правые части (как и коэффициенты матрицы СЛАУ) всегда заданы с точностью до ошибок округления. Анализ чувствительности решения к погрешностям округления и входных данных необходимо проводить при решении любых задач.

## Ошибка метода

Пусть требуется вычислить значение функции $y(x)=\sin x$ в некоторой точке $a$. Для этого рассмотрим её разложение в ряд Маклорена:

$$y(x)=\sum\limits_{k=0}^\infty\frac{(-1)^k}{(2k+1)!}x^{2k+1}.$$

Из курса анализа известно, что этот **знакопеременный** степенной ряд имеет бесконечный радиус сходимости. Это значит, что при любом $x$ он сходится абсолютно и, кроме того, для всякого отрезка числовой оси сходимость является равномерной. Напомним, что равномерная сходимость степенного ряда на некотором отрезке влечёт бесконечную дифференцируемость суммы ряда на этом отрезке. Однако при этом сумма ряда Маклорена может не совпасть с той функцией, по которой он построен. То, что в случае функции $y(x)=\sin{x}$ такое совпадение имеет место при любом $x$ (т.е. что функция $y(x)=\sin{x}$ является **регулярной** на $\mathbb{R}$), также доказывается в курсе анализа.

Практическое следствие приведённых математических фактов состоит в том, что точное значение величины $\sin a$ можно получить, если найти сумму бесконечного количества слагаемых ряда, вычисленных в точке $a$. Просуммировать бесконечное число слагаемых за конечное время невозможно. Однако можно предложить приближённый метод вычисления суммы ряда.

Оборвём ряд на $n$-м члене и будем считать, что $y(a)\simeq S_n(a)$, где $S_n(x)$ — сумма первых $n$ членов ряда:

$$S_n(x)=\sum\limits_{k=0}^{n-1}\frac{(-1)^k}{(2k+1)!}x^{2k+1}.$$

В силу знакопеременности ряда, для его остатка

$$r_n(x)=\sum\limits_{k=n}^{\infty}\frac{(-1)^k}{(2k+1)!}x^{2k+1}$$

справедлива оценка:

$$|r_n(x)|\leq\frac{|x|^{2n+1}}{(2n+1)!}\equiv\varepsilon_n.$$

![image](pic/sem01_sin.gif)

*График функции $y=\sin x$ (чёрная кривая) и графики частичных сумм $S_n(x)$ ряда Маклорена (красные кривые)*

Фиксируя натуральное $n$, мы получаем метод вычисления величины $\sin a$ путём подсчёта конечной суммы. Ошибка метода определяется оценкой на остаток ряда, т.е. оценивается сверху величиной $\varepsilon_n$. Поскольку $\varepsilon_n\to 0$ при $n\to \infty$, ошибку метода можно сделать сколь угодно малой. В этом смысле она является **устранимой**.

Запрограммируем функцию ```sum_sin```, вычисляющую сумму первых $n$ членов ряда в точке $x$:

````python
import numpy as np
from math import factorial

def sum_sin(x: float, n: int):
    # для ускорения суммирования вместо цикла используем list comprehension и векторизацию
    s = np.array( [ (-1)**k * x**(2*k+1) / factorial(2*k+1) for k in range(n) ] )

    return s.sum()
````

Построим график зависимости $S_n(x)$ от числа слагаемых $1\leq n\leq N$. График будем строить при произвольном $x=a$. Для сравнения будем выводить на печать значения частичной суммы $S_N(a)$ и $\sin a$. Последнее получим с помощью функции ```np.sin```.

````python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

from ipywidgets import interact, FloatSlider, IntSlider

# ползунки для интерактивного графика (виджета)
a_w = FloatSlider(min=-20.0, max=20.0, step=0.05, value=np.pi/6)
N_w = IntSlider(min=1, max=50, step=1, value=10) # верхняя граница диапазона для n

# функция для построения графика
def graph_sin(a: float, N: int):
    grid_n = [ i+1 for i in range(N) ] # сетка значений n
    sin_n = [ sum_sin(a, i) for i in range(N) ] # сетка значений S_n(x)

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.scatter(grid_n, sin_n) # точки
    ax.plot(grid_n, sin_n) # линия

    plt.title("Частичные суммы $S_n(x)$ в зависимости от $n$ при $x=a$")
    plt.xlabel("$n$")
    plt.ylabel("$S_n(a)$")

    plt.grid()
    ax.xaxis.set_major_locator(MaxNLocator(integer=True)) # подписи n - только целые

    plt.show()

    print("S_N(a) =", sin_n[N-1])
    print("sin(a) =", np.sin(a))

interact(graph_sin, a=a_w, N=N_w);
````

Мы видели, что в теории ошибка метода $\varepsilon_n$ может быть сделана сколь угодно малой выбором достаточно большого $n$. Однако при вычислении $S_n(a)$ на реальном компьютере из-за быстрого роста **ошибок округления** (см. ниже) достичь некоторой наперёд заданной точности $\varepsilon$ для $a\gg 1$ не удаётся.

Дело в том, что ошибки округления оказываются тем больше, чем больше $a$. Это связано с
различным поведением модулей членов ряда при $a< 1$ и $a>1$. При $a<1$ абсолютные величины членов ряда монотонно убывают, начиная с $n=0$. При $a>1$ модули членов ряда сначала растут (тем сильнее, чем больше $a$) и только потом, достигнув при некотором $k=m$ максимума, начинают убывать. Чтобы обеспечить при вычислении, например, максимального по модулю члена ряда $s_m\sim a^m$ абсолютную погрешность, не
превосходящую $\varepsilon$, необходимо вычислить его с относительной погрешностью, не превосходящей $\varepsilon/|s_m|\sim \varepsilon/|a^m|$. Иначе говоря, требуемая относительная точность вычисления членов конечной суммы $S_n(a)$ возрастает с увеличением $a$. Обеспечить же сколь угодно высокую точность можно только при неограниченной длине мантиссы (см. ниже).

В реальных программных пакетах значения элементарных функций **никогда не вычисляются** через частичные суммы степенных рядов.

## Ошибка округления

Рассмотрим квадратное уравнение

$$x^2-10^8x+1=0.$$

Его решения могут быть найдены по хорошо известной формуле:

$$x_{1,2}=\frac{10^8\pm\sqrt{10^{16}-4}}{2}=10^8\cdot\frac{1\pm\sqrt{1-4\cdot 10^{-16}}}{2}.$$

Обратим внимание на подкоренное выражение $q=1-4\cdot 10^{-16}$. В арифметике **бесконечной разрядности** (когда числа могут иметь сколь угодно много значащих разрядов) значение $q$ вычисляется точно. Однако в реальном компьютере число разрядов всегда ограничено, из-за чего неизбежны **ошибки округления**.

Ошибки округления можно охарактеризовать величиной $\varepsilon_\text{маш}$ (так называемое **машинное эпсилон** — подробнее о нём см. ниже), которая определяется как наибольшее число, для которого в машинной арифметике выполняется равенство $1+\varepsilon_\text{маш}=1$. Иначе говоря, числа, относительная разность которых не превосходит $\varepsilon_\text{маш}$, компьютером не различаются, т.е. $\varepsilon_\text{маш}$ — это максимальная относительная погрешность округления в представлении чисел с плавающей запятой.

Если предположить, что компьютер выполняет расчёты с $\varepsilon_\text{маш}=10^{-8}$, то число $q$ будет округлено до 1 (поскольку $4\cdot 10^{-16}<\varepsilon_\text{маш}=10^{-8})$. В результате получится, что $x_{1}=10^8$, $x_{2}=0$. Меньший корень уравнения будет найден неверно.

Один из возможных способов правильного вычисления меньшего корня сводится к использованию теоремы Виета, согласно которой

$$x_1+x_2=10^8,\quad x_1\cdot x_2 =1.$$

Из второго равенства следует, что $x_2=10^{-8}=\varepsilon_\text{маш}$. При этом первое равенство тоже окажется выполненным в силу определения $\varepsilon_\text{маш}$.

Можно также решить проблему, перейдя к расчётам с $\varepsilon_\text{маш}=10^{-16}$ (т.е. к представлению чисел с **двойной точностью**). Более подробная справка о вычислениях с одинарной и двойной точностью приведена ниже.


Программа для вычисления корней тремя указанными способами дает следующие результаты:

````python
import numpy as np

def quad_roots1(b: float):
    D = np.float32(b * b - 4) # дискриминант с одинарной точностью
    sqrt_D = np.sqrt(D)

    x_1 = ( b + sqrt_D ) / 2
    x_2 = ( b - sqrt_D ) / 2 # меньший корень найден через дискриминант

    return x_1, x_2

def quad_roots2(b: float):
    D = np.float32(b * b - 4)
    sqrt_D = np.sqrt(D)

    x_1 = ( b + sqrt_D ) / 2
    x_2 = 1 / x_1 # меньший корень найден по теореме Виета

    return x_1, x_2

def quad_roots3(b: float):
    D = np.float64(b * b - 4) # дискриминант с двойной точностью
    sqrt_D = np.sqrt(D)

    x_1 = ( b + sqrt_D ) / 2
    x_2 = ( b - sqrt_D ) / 2 # меньший корень найден через (более точный) дискриминант

    return x_1, x_2

x_11, x_12 = quad_roots1(10**8)
x_21, x_22 = quad_roots2(10**8)
x_31, x_32 = quad_roots3(10**8)

print( "Вычисление через дискриминант float 32: ", x_11, ", ", x_12, sep="")
print( "Поправка по теореме Виета: ", x_21, ", ", x_22, sep="")
print( "Вычисление через дискриминант float 64: ", x_31, ", ", x_32, sep="")
````

Точность $10^{-8}$, заданная в условии, хорошо соответствует $\varepsilon_\text{маш}= 6 \cdot 10^{-8}$ при вычислениях с одинарной точностью (см. ниже). В Python используется динамическая типизация, поэтому в программе выше пришлось использовать типы данных из библиотеки ```numpy``` ([документация по типам данных](https://numpy.org/doc/stable/user/basics.types.html)).

## Задача 3

Пусть $z=f(x,y)$, причём величины $x$ и $y$ измеряются приближенно. Известно, что (неустранимая) погрешность измеренного значения $x^*$ составляет $\Delta(x^*) = 10^{–3}$. Будем интересоваться **абсолютной** погрешностью $\Delta(z^*)$. С какой разумной точностью следует измерять $y^*$?

Рассмотреть случаи $f(x,y)=x+10y$ и $f(x,y)=xy+xy^2$.

**Решение в первом случае.** По свойству (вообще говоря, предельных) погрешностей

$$\Delta(z^*)=\Delta(x^*)+10\Delta(y^*).$$

Если какой-то из членов суммы отличается от другого хотя бы на порядок, то меньший член можно выбросить без большого ущерба для оценки величины $\Delta(z^*)$. Поэтому естественно требовать, чтобы вклад в $\Delta(z^*)$ от каждого из слагаемых был примерно одинаков:

$$\Delta(x^*)\sim 10\Delta(y^*).$$

Отсюда находим $\Delta(y^*)=10^{-1}\cdot\Delta(x^*)=10^{-4}$. Гнаться за большей точностью при измерении $y^*$ нет смысла: маленькую погрешность $y^*$ "перекроет" большая погрешность $x^*$.

**Решение во втором случае.** Будем ориентироваться на линейную оценку погрешности (см. выше). Тогда с величинами $\Delta(x^*)$, $\Delta(y^*)$, $\Delta(z^*)$ можно обращаться как с дифференциалами. Поэтому

$$\Delta(z^*)=y^*\Delta(x^*)+(y^*)^2\Delta(x^*)+x^*\Delta(y^*)+2x^*y^*\Delta(y^*).$$

Иначе:

$$\Delta(z^*)=\big(y^*+(y^*)^2\big)\Delta(x^*)+(x^*+2x^*y^*)\Delta(y^*).$$

Руководствуясь той же логикой, что и в первом случае, приравняем вклады от $\Delta(x^*)$ и $\Delta(y^*)$:

$$\big(y^*+(y^*)^2\big)\Delta(x^*)=(x^*+2x^*y^*)\Delta(y^*).$$

Как видим, выразить $\Delta(y^*)$ только через $\Delta(x^*)$ не удаётся: в равенство входят $x^*$ и $y^*$. Это значит, что дать оценку на абсолютную погрешность без дополнительной информации нельзя: надо знать измеренные значения $x^*$ и $y^*$. Однако можно оценить относительную погрешность (знаки модулей не пишем, считая $x^*$ и $y^*$ положительными):

$$\delta(y^*)\equiv\frac{\Delta(y^*)}{y^*}=\frac{1+y^*}{1+2y^*}\frac{\Delta(x^*)}{x^*}\equiv g(y^*)\cdot\delta(x^*).$$

Построим график функции $g(y^*)$:

````python
import matplotlib.pyplot as plt
import numpy as np

from ipywidgets import interact, FloatSlider

# ползунки для интерактивного графика (виджета)
l_w = FloatSlider(min=0.5, max=50.0, step=0.05, value=4)

def graph_g(l: float):
    # выкалываем точку y=-0.5, чтобы не было деления на 0 и ветки гиперболы не соединялись
    eps = 10**(-10)
    y_left = np.linspace(-l, -0.5-eps, 1000)
    y_right = np.linspace(-0.5+eps, l, 1000)
    g_left  = (1 + y_left) / (1 + 2*y_left)
    g_right = (1 + y_right) / (1 + 2*y_right)

    fig, ax = plt.subplots(figsize=(5, 5))
    ax.plot(y_left, g_left, color='k')
    ax.plot(y_right, g_right, color='k')

    # выделяем область y>0
    highlighting = plt.Rectangle( (0, -l-1),  2*l+2, 2*l+2, color='yellow', alpha=0.5 )
    ax.add_artist(highlighting)

    ax.set_title("График функции $g=g(y^*)$")

    ax.set_xlabel("$y^*$")
    ax.set_ylabel("$g(y)$")

    ax.set_ylim( (-l, l) )
    ax.grid()

    plt.show()

interact(graph_g, l=l_w);
````

Из графика видно, что в интересующей нас области $y^*>0$ выполняется условие $1/2\leq g(y^*)\leq 1$. Поэтому окончательно

$$\frac{1}{2}\delta(x^*)\leq\delta(y^*)\leq \delta(x^*).$$

Уточнение измерения $y^*$ до такой степени, что $\delta(y^*)<\delta(x^*)/2$, не имеет смысла.

## *Представление чисел в IEEE-арифметике

IEEE-арифметика, в частности стандарт IEEE754, описывает способ представления чисел с **плавающей точкой** в компьютерах. Все числа в такой арифметике заданы с одинаковой **относительной** точностью. Наоборот, в представлении чисел с **фиксированной точкой** все числа заданы с одинаковой **абсолютной** погрешностью.

В системе счисления с основанием $k$ любое число $a$ можно представить в **экспоненциальном виде**:

$$a_k=(-1)^s\cdot M_k\cdot k^p,$$

где $s\in\{0,1\}$ — знак, $M_k\in\mathbb{R}_+$ — **мантисса**, $p\in \mathbb{Z}$ — порядок.

**Определение 3.** Говорят, что число $a_k$ представлено в *нормализованном виде*, если его мантисса $M_k$ удовлетворяет условию $1\leq M_k<k$.

Очевидно, что при вычислениях с мантиссой и порядком **бесконечно возможной разрядности** к нормализованному виду можно привести **любое** число.

Как известно, все компьютерные вычисления проводятся в двоичной арифметике, поскольку с технической точки зрения она наиболее удобна для реализации вычислительного устройства. Поэтому нас будет интересовать вид нормализованного числа в двоичной системе счисления. По определению имеем:

$$
  a_2^\text{norm}=(-1)^s\cdot M_2\cdot 2^p,\quad 1\leq M_2<2.
$$

Так как $1\leq M_2<2$, старший разряд мантиссы всегда равен 1: $M_2=1,..._2$ (**неявная единица** — implicit one), поэтому в памяти компьютера достаточно хранить только дробную часть мантиссы $f_2$: $M_2=1+f_2$, $0\leq f_2<1$.

В силу ограниченности памяти мантисса не может быть сколь угодно длинной. Предположим, что на хранение $f_2$ выделено $m$ двоичных разрядов (бит). Тогда $f_2=f\cdot 2^{-m}$, где $f$ — $m$-битное целое неотрицательное число.

Таким образом, $a_2^\text{norm}=(-1)^s\cdot (1+f\cdot 2^{-m})\cdot 2^p$, причём $f\in[0,2^{m}-1]$.

Пусть на хранение порядка $p$ выделено $n$ бит. С помощью $n$ бит можно задать $2^n$ целых неотрицательных чисел из отрезка $[0,2^n-1]$. Однако $p$ может быть и отрицательным. Чтобы не хранить отдельно знак порядка, сдвинем $p$ на $x$ единиц: $e=p+x$. В памяти будем хранить смещённый порядок — $n$-битное целое число $e\in[0,2^n-y]$, $y\in\mathbb{N}$. Тогда $p\in [-x, 2^n-y-x]$.

Величину смещения $x<2^n-y$ выбираем из условия равенства модулей границ диапазона для $p$: $x=2^n-y-x$. Тогда $x=2^{n-1}-y/2$. При этом $p=e-2^{n-1}+y/2$, $p\in[-2^{n-1}+y/2,2^{n-1}-y/2]$. Параметр $y$ выбираем наименьшим из возможных (он должен быть чётным): $y=2$.

Окончательно: $p=e-2^{n-1}+1$, $e\in[0, 2^n-2]$, т.е. $p$ может принимать $2^n-1$ различных целых значений. При отдельном хранении знака их было бы $2^n$ за счёт двух нулей: $+0$ и $-0$.

Итак, нормализованное двоичное число $a_2^\text{norm}$ в IEEE-арифметике задаётся тремя величинами:
* знаковым битом $s\in\{0,1\}$,
* $m$-битным целым числом (дробной частью мантиссы) $f\in[0,2^m-1]$,
* $n$-битным целым числом (смещённым порядком): $e\in[0,2^n-2]$.

Отметим, что в действительности в качестве нижней границы диапазона для $e$ выбирается не $0$, а $1$. Это делается для того, чтобы отделить множество нормализованных чисел от нуля "щелью" из так называемых **субнормальных чисел** (как раз у них $e=0$). Субнормальные числа необходимы для реализации  округления. Кроме того, они предотвращают возникновение ряда ошибок при арифметических операциях на границах диапазона нормализованных чисел. За подробностями отсылаем читателя к курсам информатики и вычислительной техники.

По трём параметрам $s$, $f$ и $e$ число восстанавливается в соответствии с формулой:

$$a_2^\text{norm}=(-1)^s\cdot (1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}.$$

Заметим, что нуль нельзя представить в нормализованном виде из-за использования неявной единицы. Поэтому его вводят в IEEE-арифметику искусственно: если $e=0$ (все $n$ бит нулевые) и $f=0$ (все $m$ бит нулевые), то число полагается равным $\pm 0$ (**машинный нуль**), в зависимости от значения знакового бита.

Если $e=2^n-1$ (все $n$ бит единичные), а $f=0$ (все $m$ бит нулевые), то число полагается равным $\pm\infty$ (**машинная бесконечность**), в зависимости от значения знакового бита.

Если же $e=2^n-1$ и $f\neq 0$ (среди $m$ бит есть ненулевые), то число заменяется объектом "нечисло" (NaN — Not-a-Number). Иначе говоря, из арифметики исключаются числа, большие $+\infty$ или меньшие $-\infty$.

С представлением чисел в IEEE-арифметике связаны **три машинные константы**:

* Over Flow Limit (OFL) — **порог переполнения**, т.е. максимальное число, представимое в машинной арифметике.

* Under Flow Limit (UFL) — **порог машинного нуля**, т.е. наименьшее (по модулю) **нормализованное** число, представимое в машинной арифметике.

* Относительная машинная точность ("**машинное эпсилон**") — максимальное число, обеспечивающее справедливость равенства $1+\varepsilon_\text{маш}=1$ в машинной арифметике.

Пользуясь выведенной формулой для $a_2^\text{norm}$, находим:

$$
  \text{OFL}=\max (a_2^\text{norm})=[1+(2^m-1)\cdot 2^{-m}]\cdot 2^{2^n-2-2^{n-1}+1}=[1-2^{-(m+1)}]\cdot 2^{2^{n-1}}\underset{\strut{m\,\gtrsim\, 10}}{\simeq} 2^{2^{n-1}},
$$

$$
  \text{UFL}=\min (a_2^\text{norm})=[1+0\cdot 2^{-m}]\cdot 2^{1-2^{n-1}+1}=2^{2-2^{n-1}}.
$$

Из сказанного следует, что множество нормализованных чисел конечно. Его можно представлять себе как некоторую одномерную сетку с очень большим, но конечным набором узлов. Эта сетка заменяет собой непрерывную числовую прямую. Если число, задаваемое извне или полученное как результат расчёта, не попадает в узел сетки, то его приходится округлять до ближайшего числа, попадающего в узел. Возникает **ошибка округления** — специфическая погрешность, характерная для любых компьютерных расчётов.

Чтобы оценить ошибку округления, найдём расстояние (шаг) между двумя соседними нормализованными числами **одного порядка**. Легко видеть, что это расстояние определяется как

$$h(e)=(1+(f+1)\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}-(1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}=2^{e-2^{n-1}-m+1}.$$

Здесь $0\leq f< 2^m-1$. Видно, что при фиксированных $n$ и $m$ шаг $h$ оказывается тем больше, чем выше порядок числа. Эту закономерность нетрудно понять: так получается потому, что числа, представимые в IEEE-арифметике, **распределелены на числовой оси неравномерно** — между соседними степенями двойки заключено одинаковое количество чисел.

Очевидно, что погрешность округления будет наибольшей, если число попало строго в центр шага между узлами сетки нормализованных чисел. Поэтому для погрешности округления числа со смещённым порядком $e$ можно дать следующую оценку:

$$\Delta^\text{r}_e\leq h(e)/2=2^{e-2^{n-1}-m}.$$

Наибольшей эта оценка окажется для максимально возможного порядка:

$$\Delta^\text{r}_e\leq \Delta^\text{r}=\max_e\big[h(e)/2\big]=2^{2^n-2-2^{n-1}-m}=2^{2^{n-1}-m-2}.$$

Можно перейти к относительным величинам:

$$\delta^\text{r}_e\leq\frac{\Delta^\text{r}_e}{(1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1}}=\frac{2^{-m-1}}{1+f\cdot 2^{-m}},$$

$$\delta^\text{r}_e\leq\delta^\text{r}=\max_f\big[\delta_e^\text{r}\big]=\frac{1}{2}\cdot 2^{-m}.$$


Видно, что оценка на относительную ошибку не зависит от порядка числа: в каждом порядке она уменьшается от максимального значения $\delta^\text{r}$ для наименьшего числа с $f=0$ до значения

$$\frac{\delta^\text{r}}{2-2^{-m}}<\delta^\text{r}$$

для наибольшего числа с $f=2^m-1$.

Выясним смысл величины $\delta^\text{r}$. Пусть даны два числа $a$ и $b$ такие, что $b>a$. Их относительная разность есть

$$\frac{b-a}{a}=\frac{b}{a}-1.$$

Если эта величина окажется меньше (или равной) $\delta^\text{r}$, то в машинной арифметике числа $a$ и $b$ невозможно будет различить из-за погрешностей округления. Таким образом, **условие различимости** чисел формулируется в виде неравенства

$$\frac{b}{a}-1> \delta^\text{r}\quad \text{или}\quad b>a(1+\delta^\text{r}).$$

Поскольку при $b=a$ это неравенство не удовлетворяется ($\delta^\text{r}>0$), в машинной арифметике числа $a$ и $a(1+\delta^\text{r})$ неразличимы, т.е. $a=a(1+\delta^\text{r})$, или $1+\delta^\text{r}=1$. Отсюда следует, что $\delta^\text{r}=\varepsilon_\text{маш}$. Таким образом, по смыслу $\varepsilon_\text{маш}$ — это **максимальная относительная ошибка, которая может быть допущена при округлении числа в конечно-разрядной арифметике**.

## Числа одинарной и двойной точности

Стандарт IEEE754 (последнее обновление — 2019 г.) определяет два основных представления чисел в IEEE-арифметике: представление с **одинарной точностью** (float) и представление с **двойной точностью** (double). Они отличаются количеством бит, выделенных на хранение мантиссы и порядка числа.

При храненении чисел с одинарной точностью на одно число выделяется $32$ бита (4 байта), причём на мантиссу отведено $m=23$ бита, а на порядок — $n=8$ бит.

При храненении чисел с двойной точностью на одно число выделяется $64$ бита (8 байт), причём на мантиссу отведено $m=52$ бита, а на порядок — $n=11$ бит.

В таблице ниже приведены основные характеристики этих двух представлений. Все они получаются по общим формулам, выведенным выше. Для удобства приведём сводку этих формул:

$$a_2^\text{norm}=(-1)^s\cdot (1+f\cdot 2^{-m})\cdot 2^{e-2^{n-1}+1},\quad s\in\{0,1\},\quad f\in[0,2^m-1],\quad e\in[1,2^n-2];$$

$$\text{OFL}=[1-2^{-(m+1)}]\cdot 2^{2^{n-1}}\underset{\strut{m\,\gtrsim\, 10}}{\simeq} 2^{2^{n-1}},\quad\text{UFL}=2^{2-2^{n-1}},\quad\varepsilon_\text{маш}=\frac{1}{2}\cdot 2^{-m}.$$

Для оценок полезно помнить, что $2^{10}\simeq 10^3$.

**Таблица.** *Сравнение характеристик представлений чисел с одинарной и двойной точностью*

|  | Одинарная точность (float) |  Двойная точность (double) |
|-------|:-------:|:-------:|
| Число бит | $32=1+23+8$  | $64=1+52+11$ |
| Формула для $a_2^\text{norm}$ | $(-1)^s\cdot (1+f\cdot 2^{-23})\cdot 2^{e-127}$ | $(-1)^s\cdot (1+f\cdot 2^{-52})\cdot 2^{e-1023}$ |
| OFL   | $\simeq 3,4\cdot 10^{38}$ | $\simeq 1,8\cdot 10^{308}$ |
| UFL   | $\simeq 1,2\cdot 10^{-38}$  | $\simeq 2,2\cdot 10^{-308}$ |
| $\varepsilon_\text{маш}$   | $\simeq 6,0\cdot 10^{-8}$  | $\simeq 1,1\cdot 10^{-16}$   |

Отметим, что при работе с числами на границе диапазона машинной арифметики (или числами, выходящими за эти границы) может возникнуть потребность в использовании специальных библиотек. Например, попытка вычислить значение OFL для двойной точности вызывает ошибку переполнения:

````python
try:
    (1 - 2**(-53)) * 2**(1024) # OFL для представления с двойной точностью
except Exception as e:
    print(e)
````

Любопытно, что **по отдельности** числа $2^{-53}$ и $2^{1024}$ обрабатываются без затруднений:

````python
2**(-53)
````

````python
2**1024
````

Проблема при вычислении OFL возникает из-за необходимости конвертации целого числа $2^{1024}$ в число с плавающей точкой:



````python
try:
    float(2**1024)
except Exception as e:
    print(e)
````

Дело в том, что в Python для величин типа ```int``` реализовано **динамическое управление памятью**, благодаря которому целые числа могут быть сколь угодно длинными (верхняя граница дипазона целых чисел не фиксирована на программном уровне и определяется только доступным объёмом памяти). С другой стороны, диапазон чисел типа ```float``` ограничен программно, поэтому очень большие целые числа не могут быть приведены к этому типу напрямую ([документация по встроенным типам Python](https://docs.python.org/3/library/stdtypes.html)). Отметим, что сведения о диапазоне чисел типа ```float``` (и другие параметры типа) можно получить с помощью функции ```float_info``` библиотеки ```sys``` ([документация](https://docs.python.org/3/library/sys.html)):

Справку по обоим типам (в том числе три машинные константы) можно получить с помощью функции ```np.finfo``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.finfo.html#numpy.finfo)). Возвращаясь к примеру с подсчётом OFL, заметим, что указанную выше проблему можно решить элементарно, раскрыв скобки в вычисляемом выражении:

Настраиваемая фиксированная точность — основная отличительная черта типа ```Decimal```. Кроме того, фиксация точности во многих случаях позволяет избежать "артефактов" машинной арифметики, т.к. модуль работает в десятичной системе счисления. Пример:

Нужно, однако, следить за правильностью создания чисел типа ```Decimal```. В ячейке выше числа заданы конвертацией из строк. При попытке заменить строки десятичными дробями (```float```) результат будет другой:

````python
Decimal(0.3) - Decimal(0.2) - Decimal(0.1)
````

## Задача 4

Рассмотрим модель представления чисел в IEEE-арифметике следующего вида:

$$S=\{\pm b_0,b_1b_2\cdot 2^{\pm a}\}.$$

Здесь $a,b_1,b_2\in\{0,1\}$, а $b_0=1$ всегда, за исключением случая, когда $a=b_1=b_2=0$; в этом случае $b_0\in\{0,1\}$.

Нарисовать множество $S$ на действительной оси. Определить количество чисел в рассматриваемой арифметике. Вычислить машинные константы OFL, UFL, $\varepsilon_\text{маш}$.

*Примечание.* В рассматриваемой арифметике нет субнормальных чисел, в отличие от "настоящей" IEEE-арифметики.

**Решение.** Пусть сначала $b_0=1$. Составим таблицу перебора всех возможных мантисс:

$$\begin{gather}
  & b_0 &\quad& b_1 &\quad& b_2 &\quad& \text{В десятичном виде}\\
  M_1& 1   &\quad& 0   &\quad& 0   &\quad&   1   \\
  M_2& 1   &\quad& 0   &\quad& 1   &\quad&   5/4 \\
  M_3& 1   &\quad& 1   &\quad& 0   &\quad&   3/2 \\
  M_4& 1   &\quad& 1   &\quad& 1   &\quad&   7/4 \\
\end{gather}$$

Для каждого из этих четырёх чисел существует три возможных порядка: $a=\pm 1$ или $a=0$. Таким образом, имеем 12 чисел:

$$\begin{gather}
  M_i\cdot 2^0 &\quad& M_i\cdot 2^{-1} &\quad& M_i\cdot 2^{+1} \\
   1     &\quad& 1/2   &\quad& 2      \\
   5/4   &\quad& 5/8   &\quad& 5/2    \\
   3/2   &\quad& 3/4   &\quad& 3    \\
   7/4   &\quad& 7/8   &\quad& 7/2    \\
\end{gather}$$

Каждое из них может иметь один из двух знаков. Кроме того, при $a=b_1=b_2=0$ имеем ещё два числа: нуль ($b_0=0$) и один ($b_1=1$). Но единица уже учтена в таблице, т.е. в рассматриваемой арифметике она дублируется. Таким образом, общее количество чисел (мощность конечного множества $S$):

$$2\cdot 12 + 2 - 1 = 25.$$

Из второй таблицы видно, что $\text{OFL}=7/2$, $\text{UFL}=1/2$. **Машинное эпсилон в силу своего смысла совпадает с половиной последнего разряда**, т.е. в данном случае

$$\varepsilon_\text{маш}=\frac{1}{2}\cdot 2^{-2}=\frac{1}{8}.$$

Его можно было бы найти и по выведенной выше общей формуле, заметив, что $m=2$ (число бит, выделенных на хранение **дробной части** мантиссы).

Остаётся отметить числа из множества $S$ на числовой прямой. Это можно сделать с помощью программы. Тривиальный вариант заключается в ручном задании чисел из таблицы:

````python
import matplotlib.pyplot as plt
import numpy as np

def graph_triv():
    x = np.array([1, 5/4, 3/2, 7/4, 1/2, 5/8, 3/4, 7/8, 2, 5/2, 3, 7/2])
    x = np.concatenate((-x, x, [0])) # добавляем отрицательные и нуль
    y = [0]*25

    fig, ax = plt.subplots(figsize=(10, 2))
    ax.scatter(x, y, color='r', s=20)

    # рисуем числовую ось
    ax.plot([np.min(x)-1/2, np.max(x)+1/2], [0, 0], color='k', zorder=0)

    ax.set_title("Числа из множества $S$")

    plt.yticks([])

    plt.show()

graph_triv()
````

Видно, что распределение чисел неравномерно.
