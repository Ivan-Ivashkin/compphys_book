# Постановка задачи численного дифференцирования

Одной из простейших задач вычислительной математики является **задача численного дифференцирования**. Эта задача обычно является составляющей более сложных задач. Понимать основы численного диференцирования важно в первую очередь для численного решения дифференциальных уравнений (обыкновенных и в частных производных).

С точки зрения математического анализа задача дифференцирования функции $\varphi=\varphi(x)$, определённой на отрезке $[a,b]$, в точке $x=\xi\in[a,b]$ сводится к вычислению предела дифференциального отношения:

$$\varphi'(\xi)=\lim_{x\to \xi}\frac{\varphi(x)-\varphi(\xi)}{x-\xi}.$$

Полагая $x=\xi+\Delta x$, приходим к эквивалентной записи:

$$\varphi'(\xi)=\lim_{\Delta x\to 0}\frac{\varphi(\xi+\Delta x)-\varphi(\xi)}{\Delta x}.$$

В курсах анализа такие пределы вычисляют для некоторого набора функций (составляют **таблицу производных**). Затем доказывают ряд **правил дифференцирования**. После этого задача дифференцирования произвольной функции (в области её определения) сводится к последовательному применению правил: ими пользуются для упрощения дифференцируемого выражения до одной или нескольких табличных функций.

Как уже отмечалось на первом семинаре, в вычислительной математике любая (непрерывная) функция $\varphi(x)$ заменяется совокупностью своих значений в узлах некоторой **сетки**. В результате получается новый объект — **сеточная функция**.

**Определение 1.** Будем говорить, что на отрезке $[a,b]$ введена *сетка*, если задано *конечное* множество

$$G=\{x_i\in\mathbb{R}, 0\leq i\leq n: a\leq x_0<x_1<...<x_n\leq b\}.$$

Точки $x_i$ будем называть *узлами сетки*, а величины $h_i=x_i-x_{i-1}$, $1\leq i\leq n$ — её *шагами*. Если $h_1=h_2=...=h_n\equiv h$, то сетка называется *равномерной*.

**Определение 2.** Пусть на отрезке $[a,b]$ введена сетка $G$. *Сеточной функцией* называется *конечное* множество пар

$$F=\big\{(f,x): f\in\mathbb{R}, x\in G\big\},$$

в котором каждое значение $x$ принадлежит лишь одной паре. Величины $f$ называются *значениями сеточной функции*.

**Определение 3.** Пусть на отрезке $[a,b]$ определена некоторая гладкая функция $\varphi=\varphi(x)$. *Сеточной проекцией* функции $\varphi$ называется сеточная функция

$$F_\varphi=\big\{(\varphi(x),x): x\in G\big\},$$

где $G$ — некоторая сетка, введённая на отрезке $[a,b]$.

Операцию получения сеточной функции $F_\varphi$ из гладкой (нужное число раз дифференцируемой) функции $\varphi$ называют **проектированием функции на сетку**. Сформулировав третье определение, мы зафиксировали конкретный способ проектирования: он сводится к вычислению конечного набора значений

$$f_i=\varphi(x_i)$$

в узлах $x_i$ некоторой сетки. Вообще говоря, можно рассматривать и другие (более сложные) способы построения сеточных проекций. В нашем курсе мы ограничимся наиболее простым — тем, который следует из третьего определения.

В терминах приведённых выше определений задача численного дифференцирования формулируется следующим образом.

Пусть на отрезке $[a,b]$ определена некоторая гладкая (нужное число раз дифференцируемая) функция $\varphi=\varphi(x)$, а $F_\varphi$ — её сеточная проекция. Требуется определить по $F_\varphi$ сеточную функцию $F'_\varphi$ такую, что $F'_\varphi\simeq F_{\varphi'}$. Точный смысл знака приближённого равенства в последней формуле станет ясен из дальнейшего.

# Простейшие формулы численного дифференцирования

Простейший способ приближённого вычисления производной подсказывается её определением. Составим дифференциальное отношение

$$\frac{\varphi(\xi+\Delta x)-\varphi(\xi)}{\Delta x}.$$

Если величина $\Delta x$ достаточно мала (но конечна!), то это отношение близко к $\varphi'(\xi)$. Формализуем указанный способ вычисления производной.

Зададим на отрезке $[a,b]$ сетку $G$. Здесь и далее мы для простоты будем считать её равномерной — это предположение ничего не меняет с принципиальной точки зрения, но упрощает формулы. Построим сеточную проекцию $F_\varphi$ функции $\varphi$:

$$F_\varphi=\big\{(f_m,x_m):f_m=\varphi(x_m), x_m\in G\big\}.$$

Обозначения $x_m$ и $f_m$ введены для того, чтобы подчеркнуть конечность множества $F_\varphi$. Искомую сеточную функцию $F'_\varphi$ определим так:

$$F'_\varphi=\bigg\{(f_m',x_m):f_m'=\frac{f_{m+1}-f_m}{h},x_m\in G' \bigg\}.$$

Через $h$ здесь обозначен шаг сетки $G$. Обратим внимание на то, что значение $f'_m$ не определено для последнего узла сетки $G$, поскольку при $m=n$ невозможно вычислить $f_{m+1}$, так как достигнута правая граница сетки. В связи с этим в формуле для $F'_\varphi$ вместо $G$ написано $G'$ — имеется в виду сетка из $n$ узлов, совпадающих с первыми $n$ узлами сетки $G$.

Выясним, насколько точен предложенный метод дифференцирования, т.е. оценим **ошибку метода**. Для этого заметим, что

$$f_m'=\frac{f_{m+1}-f_m}{h}=\frac{\varphi(x_{m+1})-\varphi(x_{m})}{h},$$

$$\varphi(x_{m+1})=\varphi(x_{m}+h)=\varphi(x_{m})+\varphi'(x_{m})h+\varphi''(x_{m})\frac{h^2}{2}+O(h^3),$$

$$f_m'=\varphi'(x_{m})+\varphi''(x_{m})\frac{h}{2}+O(h^2).$$

Из последнего равенства следует, что отличие каждого значения $f'_m$ сеточной функции $F'_\varphi$ от соответствующего значения $\varphi(x_m)$ сеточной проекции $F_{\varphi'}$ оказывается тем меньше, чем меньше шаг сетки $h$, т.е. в пределе $h\to 0$ сеточные функции $F'_\varphi$ и $F_{\varphi'}$ совпадают (ошибки округления мы пока не рассматриваем). Именно эту связь между сеточными функциями мы имели в виду, когда писали равенство $F'_\varphi\simeq F_{\varphi'}$ при формулировке задачи численного дифференцирования.

В реальных вычислениях шаг сетки всегда выбирается малым, но конечным, и поэтому разность $f'_m-\varphi'(x_m)$ оказывается ненулевой. С точностью до членов первого порядка малости по $h$ можем написать:

$$|f'_m-\varphi'(x_m)|=|\varphi''(x_{m})|\frac{h}{2}.$$

Положим $M_2=\displaystyle\max_{x\in[a,b]}|\varphi''(x)|\geq\displaystyle\max_m|\varphi''(x_m)|$ (индекс "2" указывает на второй порядок производной). Тогда

$$|f'_m-\varphi'(x_m)|\leq\frac{M_2 h}{2}\equiv \Delta_\text{forw}^{(\text{m})},\quad 0\leq m\leq n-1.$$

Величина $\Delta_\text{forw}^{(\text{m})}$ и является искомой ошибкой метода.

В дальнейшем мы для краткости будем опускать обозначения сеточных функций и сеток, характеризуя метод дифференцирования только формулой для вычисления $f_m'$. Рассмотренная выше формула

$$f_m'=\frac{f_{m+1}-f_m}{h}$$

носит название "**производная вперёд**". Наряду с этой формулой можно рассмотреть также **производную назад**

$$f_m'=\frac{f_{m}-f_{m-1}}{h}$$

и формулу **центральной разности**

$$f_m'=\frac{f_{m+1}-f_{m-1}}{2h}.$$

Читателю предлагается самостоятельно убедиться в том, что для производной назад (как и для производной вперёд)

$$\Delta_\text{back}^{(\text{m})}=\frac{M_2 h}{2},$$

а для формулы центральной разности

$$\Delta_\text{cent}^{(\text{m})}=\frac{M_3 h^2}{6},$$

где $M_3=\displaystyle\max_{x\in[a,b]}|\varphi'''(x)|$.



# Порядки точности, аппроксимации и сходимости

Выше мы оценили погрешность метода для трёх формул численного дифференцирования. Полученные оценки включают константы $M_2$ и $M_3$ — максимумы второй и третьей производных соответственно. Как мы увидим ниже, подобную оценку можно дать для произвольной формулы численного дифференцирования, причём в неё войдёт константа $M_k=\displaystyle\max_{x\in[a,b]}|\varphi^{(k)}(x)|$. Очевидно, что если в качестве функции $\varphi(x)$ взят полином

$$P_n(x)=\sum\limits_{i=0}^{n}a_ix^i,$$

то $M_k=0$ при $n<k$, т.е. рассматриваемый метод численного дифференцирования точен на полиномах со степенью, меньшей $k$. Это даёт основание ввести следующее определение.

**Определение 4.** *Порядком точности формулы численного дифференцирования* называется наибольшая из степеней полиномов, для которых формула даёт точный результат дифференцирования в предположении отсутствия ошибок округления.

Легко видеть, что в приведенных выше примерах **порядок точности совпадает со степенью** $h$ в оценке для погрешности метода, т.е. производные вперёд и назад имеют первый порядок точности, а формула центральной разности — второй.

Хотя определение 4 часто встречается в литературе, мы в дальнейшем редко будем им пользоваться. Дело в том, что для дифференциальных задач вычислительной математики принято вводить более фундаментальные понятия **аппроксимации** и **порядка аппроксимации**. Мы дадим соответствующие общие определения, когда перейдём к изучению теории разностных схем для обыкновенных дифференциальных уравнений. Пока же ограничимся следующим алгоритмическим подходом.

**Алгоритм исследования порядка аппроксимации формулы численного дифференцирования**

**Шаг 1.** Подставить в рассматриваемую формулу вместо $f_m$ сеточную проекцию $\varphi(x_m)$ функции $\varphi$. Иллюстрация из приведённого выше примера:

$$f_m'=\frac{f_{m+1}-f_m}{h}=\frac{\varphi(x_{m+1})-\varphi(x_{m})}{h}=\frac{\varphi_{m+1}-\varphi_{m}}{h}.$$

Замена $f'_m$ на $\varphi(x_m)$ нужна для следующего шага алгоритма, так как **сеточную функцию нельзя разложить в ряд Тейлора**, а достаточно гладкую функцию $\varphi(x)$ — можно.

**Шаг 2.** Воспользоваться разложением в ряд Тейлора вблизи одной выбранной точки сетки (например, $x_m$ или другой удобной) для вычисления значений функции в остальных точках сетки, входящих в формулу. Иллюстрация из приведённого выше примера:

$$\varphi(x_{m+1})=\varphi(x_{m}+h)=\varphi(x_{m})+\varphi'(x_{m})h+\varphi''(x_{m})\frac{h^2}{2}+O(h^3).$$

**Шаг 3.** Подставить полученные разложения в формулу численного дифференцирования и получить оценку вида

$$|f'_m-\varphi'(x_m)|\leq Ch^p,\quad C\in\mathbb{R}_+,\quad p\in \mathbb{N}$$

с не зависящей от параметров сетки константой $C$. Число $p$ называется **порядком аппроксимации**.


Проще говоря, в случае задачи численного дифференцирования **порядок аппроксимации — это степень шага сетки** $h$ в лидирующем члене погрешности метода. Подчеркнём, что порядок аппроксимации метода может как совпадать с порядком точности, так и отличаться от него.

Ниже мы рассмотрим **метод неопределённых коэффициентов**, позволяющий легко построить формулу численного дифференцирования любого заданного порядка аппроксимации. Здесь же остановимся на следующем вопросе: **есть ли смысл строить и использовать формулы высоких порядков аппроксимации?**

Пусть перед нами стоит задача численного дифференцирования некоторой функции. Предположим, что мы вычисляем её производную по формуле первого порядка аппроксимации. Пусть выбран достаточно крупный шаг сетки $h$. При этом ошибка метода составляет $Ch$. Изменим шаг $h$ — например, уменьшим его вдвое. Тогда и ошибка метода уменьшится вдвое.

Если при тех же условиях мы будем использовать метод второго порядка аппроксимации, то в результате уменьшения шага сетки в два раза ошибка метода уменьшится уже в $2^2=4$ раза. В случае же метода произвольного порядка аппроксимации $p$ при уменьшении шага сетки вдвое ошибка уменьшается в $2^p$ раз.

Как видим, **методы высокого порядка аппроксимации представляют интерес**, так как с их помощью даже на неподробных сетках можно получить результат, сопоставимый по точности с результатом, полученным методом низкого порядка на очень подробной сетке.

С другой стороны, **существует некоторая граница, до которой разумно увеличивать порядок аппроксимации формул численного дифференцирования**. Это видно хотя бы из того, что при классическом способе построения формул порядок аппроксимации оказывается связанным с количеством используемых точек сетки. Ниже мы увидим, что чем выше требуемый порядок аппроксимации, тем больше точек приходится включать в сеточный шаблон. Но формулу с широким шаблоном невозможно использовать для вычисления производной вблизи границ расчётной области. При слишком широком шаблоне и сложной геометрии области количество "недоступных" точек может оказаться велико.

Кроме того, следует помнить, что с увеличением порядка аппроксимации формулы численного дифференцирования повышаются требования на гладкость функции. Например, в оценку погрешности метода сотого порядка аппроксимации вошел бы максимум модуля 101-й производной функции. На практике о степени гладкости функции обычно известно немногое, и часто приходится решать задачи, в которых функции вообще терпят разрывы. Наконец, вычисления не всегда проводятся на равномерных сетках, а строить формулу высокого порядка аппроксимации на неравномерной сетке сложнее.

Говоря о порядке аппроксимации, нельзя не упомянуть также о **порядке сходимости**. Эти порядки, как мы увидим в общей теории разностных схем, связываются **основной теоремой вычислительной математики**, в которой фигурируют три центральных понятия: **аппроксимация**, **устойчивость** и **сходимость**. Их строгие определения будут даны позже, но на семинарах понятиями аппроксимации и сходимости мы будем пользоваться ещё до введения определений. На этом этапе понятия аппроксимации и сходимости (а также соответствующих порядков) кажутся похожими. Важно понимать, в чём состоит принципиальная разница.

**Порядок аппроксимации** определяется по оценке ошибки метода, полученной **аналитически** при помощи разложений в ряд Тейлора в предположении нужной гладкости функции. **Порядок сходимости** определяется по ошибке метода, полученной при помощи **вычислений на компьютере** для конкретной задачи. На этапе тестирования программы обычно используют задачи с известным решением, поэтому ошибку можно легко определить по разности численного и аналитического решений. Чтобы определить порядок сходимости, обычно последовательно запускают программу, реализующую рассматриваемый численный метод, **на совокупности сгущающихся сеток** (например, на сетках с шагами $h$, $h/2$, $h/4$ и т.д.) и исследуют, во сколько раз уменьшается ошибка метода при измельчении сетки. Существует также способ оценки порядка сходимости, применимый в случае задач, не имеющих аналитического решения (именно для таких задач и создаются численные методы). Этот способ мы рассмотрим позже — на семинаре, посвящённом численному интегрированию.

Подчеркнём существенные отличия порядков аппроксимации и сходимости.

* При исследовании порядка аппроскимации оценки сугубо теоретические. При исследовании порядка сходимости оценки получают в реальном расчёте.

* Если функция, с которой приходится иметь дело, достаточно гладкая, то порядок сходимости совпадёт с порядком аппроксимации. Если степень гладкости функции ниже, чем предполагалось в исследовании аппроксимации, то порядок сходимости будет меньше порядка аппроксимации.

* Порядок сходимости в реальной задаче может и не быть целым числом.

# Погрешности округления в задаче численного дифференцирования

Итак, если некоторая формула численного дифференцирования имеет порядок аппроксимации $p$, то ошибка метода $\Delta^{(\text{m})}\sim h^p$. Отсюда как будто бы следует, что нужно выбирать сетку с наименьшим возможным $h$ — тогда погрешность в вычислении производной тоже будет наименьшей. Однако в расчётах на реальном компьютере неизбежны ошибки округления, которые при слишком малом шаге вносят значительную погрешность в результат дифференцирования. Убедимся в этом на конкретных примерах.

## Численные примеры

Проверим, прежде всего, что результат численного дифференцирования существенно зависит от величины шага. Для этого рассмотрим функцию $\varphi(x)=\exp(x)$ на отрезке $[0,l]$. Её производная, как известно, совпадает с самой функцией: $\varphi'(x)=\varphi(x)=\exp(x)$. Зададим равномерную сетку с некоторым шагом $h$ и построим на этой сетке разностный аналог функции $\varphi'(x)$ — производную вперёд. Результаты отобразим на графике (для начала возьмём $l=1$):

````python
import numpy as np

def exp_deriv(h: float, l: float):
    x = np.arange(0, l, h)
    y = np.exp(x)

    f_n1 = y[1:]
    f_n = y[:-1]

    f_prime = (f_n1 - f_n) / h

    return f_prime, x[:-1]
````

````python
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def graph_exp_deriv(h: float, l: float):
    f_prime, x = exp_deriv(h, l)

    fig, ax = plt.subplots()
    ax.scatter(x, f_prime, s=5, c='b')

    grid_exp = np.arange(0, l, h)
    ax.plot(grid_exp, np.exp(grid_exp), c='r')

    plt.title("Производная функции $\\varphi(x)=\\exp(x)$")
    plt.xlabel("$x$")
    plt.ylabel("$\\varphi'(x)$")

    plt.grid()
    plt.show()

# ползунок для интерактивного графика (виджета)
h_w = FloatSlider(min=10**(-6), max=0.1, step=10**(-3), value=0.1)

interact(lambda h: graph_exp_deriv(h, 1), h=h_w);
````

Попытка запустить отрисовку графика при $h<10^{-6}$ приводит к чрезмерному расходованию памяти. Поэтому для $h<10^{-6}$ возьмём меньший отрезок: $l=10^{-6}$.

````python
# создаём ползунок заново, чтобы не менять график из предыдущей ячейки
h_again = FloatSlider(min=10**(-6), max=0.1, step=10**(-3), value=0.1)

interact(lambda h: graph_exp_deriv(h*10**(-6), 10**(-6)), h=h_again);
````

Как видим, уменьшение шага, сопровождающееся увеличением количества узлов в сетке, **сначала уточняет результат** дифференцирования, **а потом только ухудшает** его: при малых шагах порядка $10^{-12}$ точки группируются в отрезки, совершенно не соответствующие истинному ходу производной.

Приведём другой пример. Будем вычислять дифференциальное отношение

$$\frac{\varphi(\xi+h)-\varphi(\xi)}{h}$$

для функции $\varphi(x)=\sin x$. С уменьшением шага $h$ оно должно приближаться к величине $\varphi '(\xi)=\cos\xi$. По сути, мы пользуемся формулой "производная вперёд" на сетке, состоящей всего из двух точек: $\xi$ и $\xi+h$. При этом ошибка численного дифференцирования есть

$$\Delta(\xi,h)=\Bigg|\frac{\sin(\xi+h)-\sin\xi}{h}-\cos\xi\Bigg|.$$

Условимся рассматривать шаги в диапазоне от $1$ до $10^{-16}$ ($\varepsilon_{маш}$ для двойной точности).

Функция для вычисления массива значений $\Delta=\Delta(\xi, h)$ при заданном $\xi$ и $h$, меняющемся от $1$ до $10^{-16}$:

````python
def deriv_sin(xi: float):
    h = np.logspace(-16, 0, 1000, base=10) # сетка шагов по степеням десятки (от 10^{-16} до 1)
    deriv = (np.sin(xi+h) - np.sin(xi)) / h

    return abs(deriv - np.cos(xi)), h
````

Построим график зависимости ошибки численного дифференцирования от величины шага. Кроме того, выведем на печать минимальную ошибку, значение шага, при котором она достигается, и величину $\varepsilon_\text{маш}$, восстановленную по минимальной ошибке (см. формулу ниже).

````python
# ползунок для интерактивного графика (виджета)
xi_w = FloatSlider(min=0, max=10000, step=0.05, value=np.pi/3)

# функция для построения графика
def graph_sin_deriv(xi: float):
    delta, grid_h = deriv_sin(xi)

    fig, ax = plt.subplots()
    ax.plot(grid_h, delta)

    plt.title("Погрешность производной вперёд $\\Delta(x,h)$ в зависимости от $h$ при $x=\\xi$")
    plt.xlabel("$h$")
    plt.ylabel("$\\Delta(x,h)$")

    plt.xscale('log') # логарифмические (по основанию 10) шкалы по x и y
    plt.yscale('log')

    plt.grid()
    plt.show()

    delta_min = np.min(delta)
    h_opt = grid_h[np.argmin(delta)]
    epsilon_mach = delta_min**2 / 4

    print("Минимальная ошибка: {:.3e}".format(delta_min))
    print("Шаг, при котором достигается минимальная ошибка: {:.3e}".format(h_opt))
    print("Машинное эпсилон: {:.3e}".format(epsilon_mach))

interact(graph_sin_deriv, xi=xi_w);
````

Из графика видно, что с уменьшением шага величина ошибки сначала падает, а потом возрастает, достигая минимума при некотором **оптимальном шаге** $h_\text{opt}$. Стохастический характер роста ("гребёнка" в левой части графика) указывает на то, что при $h<h_\text{opt}$ основной вклад в ошибку дают погрешности округления.

## Теоретическая оценка погрешности округления

Теоретически оценить погрешность округления можно следующим образом. Поскольку $\varepsilon_\text{маш}$ есть **максимальная относительная погрешность округления**, абсолютная погрешность округления $\Delta^{(\text{r})}(f_{m+1})$ величины $f_{m+1}$, входящей в формулу производной вперёд, не превосходит $\varepsilon_\text{маш} f_{m+1}\leq M_0\varepsilon_\text{маш}$, где $M_0=\max\limits_{x\in[a,b]}|\varphi(x)|$. Аналогичная оценка справедлива для $\Delta^{(\text{r})}(f_{m})$. По свойствам (вообще говоря, предельных) погрешностей имеем:

$$\Delta_\text{forw}^{(\text{r})}=\frac{\Delta^{(\text{r})}(f_{m+1})+\Delta^{(\text{r})}(f_m)}{h}=\frac{2M_0\varepsilon_\text{маш}}{h}.$$

Таким образом, полная погрешность численного дифференцирования по формуле производной вперёд есть

$$\Delta_\text{forw}(h)=\Delta_\text{forw}^{(\text{m})}+\Delta_\text{forw}^{(\text{r})}=\frac{M_2 h}{2}+\frac{2M_0\varepsilon_\text{маш}}{h}.$$

При больших $h$ доминирует ошибка метода, растущая линейно с увеличением шага. При малых $h$ существенной становится ошибка округления, нарастающая при $h\to 0$ как $h^{-1}$. Значит, в полном соответствии с численной демонстрацией, существует некое **оптимальное значение шага** $h_\text{opt}$, минимизирующее $\Delta_\text{forw}$. Найдём его:

$$\Delta_\text{forw}'(h)=\frac{M_2}{2}-\frac{2M_0\varepsilon_\text{маш}}{h^2}=0,\quad h_\text{opt}=2\sqrt{\frac{M_0\varepsilon_\text{маш}}{M_2}}.$$

Наименьшая величина ошибки оценивается как

$$\Delta_\text{forw}^{\text{min}}=\Delta_\text{forw}(h_\text{opt})=2\sqrt{M_0M_2\varepsilon_\text{маш}}.$$

Если $\varphi(x)=\sin x$, то $M_0=M_2=1$ и $\Delta_\text{forw}^\text{min}=2\sqrt{\varepsilon_\text{маш}}$. Это даёт возможность оценить величину $\varepsilon_\text{маш}$: $\varepsilon_\text{маш}=(\Delta_\text{forw}^\text{min})^2/4$, что мы и сделали в приведённой выше программе. Заметим, что при $\xi=1$ оценка на $\varepsilon_\text{маш}$ получается явно заниженной, а с увеличением $\xi$ она увеличивается (поскольку растёт минимальная ошибка).

**Упражнение.** Объяснить, почему полученное в примере значение $\varepsilon_\text{маш}$ отличается от правильного.

Аналогичный анализ ошибок округления с вычислением оптимального шага можно провести для производной назад и центральной разности. Приведём только ответы: для производной назад формулы те же, что и для производной вперёд, а для формулы центральной разности

$$\Delta_\text{cent}(h)=\Delta_\text{cent}^{(\text{m})}+\Delta_\text{cent}^{(\text{r})}=\frac{M_3 h^2}{6}+\frac{M_0\varepsilon_\text{маш}}{h},\quad h_\text{opt}=\bigg(\frac{3M_0\varepsilon_\text{маш}}{M_3}\bigg)^{1/3}.$$

# Метод неопределённых коэффициентов построения формул численного дифференцирования

Мы видели, что формула производной вперёд

$$f_m=\frac{f_{m+1}-f_m}{h}$$

имеет **первый порядок аппроксимации**, а формула центральной разности

$$f_m=\frac{f_{m+1}-f_{m-1}}{2h}$$

имеет **второй порядок аппроксимации**, т.е. является потенциально более точной. Возникает вопрос: чем вторая формула принципиально отличается от первой? Какое существенное отличие объясняет разный порядок точности формул?

Нетрудно заметить, что в первом случае для вычисления значения производной в узле $x_m$ сетки требуется знать значения функции в узлах $x_m$ и $x_{m+1}$, т.е. в вычислении задействованы "текущий" узел $x_m$ и **один** соседний справа узел $x_{m+1}$. С другой стороны, в формуле центральной разности задействованы **два** соседних узла: узел слева $x_{m-1}$ и узел справа $x_{m+1}$, — а общее количество узлов, рассматриваемых при вычислении производной в данной точке сетки, равно трём ($x_m$, $x_{m-1}$ и $x_{m+1}$). Говорят, что эти формулы отличаются **сеточным шаблоном**: у производной вперёд шаблон двухточечный (точки $x_m$ и $x_{m+1}$), а у центральной разности — трёхточечный (точки $x_m$, $x_{m-1}$ и $x_{m+1}$).

После сказанного естественно выдвинуть следующую гипотезу: **чем больше точек** включает сеточный шаблон, **тем более высокий порядок аппроксимации** имеет построенная на нём формула численного дифференцирования. Далее мы подтвердим эту гипотезу, доказав, что **на шаблоне из $N+1$ точки всегда можно построить единственную формулу численного дифференцирования с порядком аппроксимации, равным $N$**.

Пусть сеточный шаблон включает $i$ точек слева от "текущего" узла $x_m$ и $j$ точек справа от него (т.е. всего в шаблоне $i+j+1\equiv N+1$ точка). Общий вид построенной на этом шаблоне формулы численного дифференцирования:

$$f'_m=\frac{\alpha_{-i}f_{m-i}\,+\, ...\,+\,\alpha_{-1}f_{m-1}+\alpha_0f_m+\alpha_1f_{m+1}\,+\,...\,+\,\alpha_jf_{m+j}}{h}=\frac{1}{h}\sum\limits_{k=-i}^{j}\alpha_k f_k.$$

Множитель $1/h$ включён в формулу из размерных соображений. Среди коэффициентов $\alpha_k$ могут быть нулевые. Например, для формулы производной вперёд $i=0$, $j=1$, $\alpha_0=-1$, $\alpha_1=1$; для формулы центральной разности $i=j=1$, $\alpha_0=0$, $\alpha_1=1$, $\alpha_{-1}=-1$. Значения сеточной функции $f_k$ получены в результате проектирования гладкой функции $\varphi$ на сетку, поэтому $f_k=\varphi(x_k)$, т.е.

$$f'_m=\frac{1}{h}\sum\limits_{k=-i}^{j}\alpha_k \varphi(x_m+kh).$$

Разложим функцию $\varphi$ в ряд Тейлора в окрестности точки $x_m$:

$$\varphi(x_m+kh)=\sum\limits_{n=0}^q\frac{\varphi^{(n)}(x_m)}{n!}(kh)^n+O(h^{q+1}).$$

Подставим это разложение в формулу для $f'_m$:

$$f'_m=\sum\limits_{n=0}^q\varphi^{(n)}(x_m)h^{n-1}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

Выпишем явно первые два члена суммы по $n$:

$$f'_m=\frac{\varphi(x_m)}{h}\sum\limits_{k=-i}^j\alpha_k+\varphi'(x_m)\sum\limits_{k=-i}^j\alpha_k k +\sum\limits_{n=2}^q\varphi^{(n)}(x_m)h^{n-1}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

**Потребуем**, чтобы

$$\sum\limits_{k=-i}^j\alpha_k=0,\quad \sum\limits_{k=-i}^j\alpha_k k=1.$$

Первое условие необходимо для того, чтобы исключить из правой части слагаемое, содержащее значение функции $\varphi(x_m)$ и не убывающее при измельчении сетки. Второе условие обеспечивает аппроксимацию первой производной функции. Тогда

$$f'_m=\varphi'(x_m)+\sum\limits_{n=2}^q\varphi^{(n)}(x_m)h^{n-1}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

Пусть требуется получить формулу численного дифференцирования с порядком аппроксимации $p<q$. Это значит, что в сумме по $n$ первый ненулевой член должен включать $h^p$, т.е. все члены с $2\leq n<p+1$ должны обратиться в нуль. В силу произвольности функции $\varphi$ и шага $h$ это требование даёт $p-1$ условие на коэффициенты $\alpha_k$ **в дополнение к двум наложенным выше условиям** (множитель $1/n!$ опускаем, т.к. он отличен от нуля):

$$\sum\limits_{k=-i}^j\alpha_k k^n=0\quad\text{для }n=2,3,...,p.$$

Таким образом, всего имеем $p+1$ условие на $N+1$ коэффициентов $\alpha_k$. Каждое условие представляет собой линейное уравнение на $\alpha_k$, т.е. мы имеем **систему из $p+1$ линейного алгебраического уравнения с $N+1$ неизвестной** (СЛАУ). Как известно, существование единственного решения можно гарантировать, если $p+1=N+1$ и матрица системы не является вырожденной. Если $p+1>N+1$ (переопределённая система), то выбранный **шаблон недостаточно широк**. Если же $p+1<N+1$ (недоопределённая система), то выбранный **шаблон избыточен**. Поэтому далее мы считаем, что $p+1=N+1$, т.е. $p=N$.

Полученная СЛАУ может быть записана в векторно-матричном виде следующим образом:

$$
  \begin{pmatrix}
    1  & 1 & \dots & 1 & 1\\
    -i & -i+1 & \dots & j-1 & j\\
    (-i)^2 & (-i+1)^2 & \dots & (j-1)^2 & j^2 \\
    \vdots & \vdots & \ddots & \vdots & \vdots\\
    (-i)^p & (-i+1)^p & \dots & (j-1)^p & j^p
  \end{pmatrix}\,
  \begin{pmatrix}
    \alpha_{-i} \\
    \alpha_{-i+1} \\
    \vdots\\
    \alpha_{j-1}\\
    \alpha_{j}
  \end{pmatrix}=
  \begin{pmatrix}
    0 \\
    1 \\
    \vdots\\
    0\\
    0
  \end{pmatrix}.
$$

Детерминант матрицы системы имеет вид **определителя Вандермонда** порядка $p$, для которого в курсе линейной алгебры доказывается (по индукции) формула:

$$
  \begin{vmatrix}
    1 & x_1 & \dots & x_1^{p-1}\\
    1 & x_2 & \dots & x_2^{p-1}\\
    \vdots & \vdots & \ddots & \vdots\\
    1 & x_p & \dots & x_p^{p-1}
  \end{vmatrix}=\prod\limits_{1\leq b<a\leq p}(x_a-x_b).
$$

В нашем случае в роли $x_a$ и $x_b$ выступают целые числа от $-i$ до $j$ включительно. Поскольку среди них нет повторяющихся (каждый узел входит в шаблон единожды), соответствующий определитель Вандермонда отличен от нуля. Таким образом, рассматриваемая СЛАУ имеет единственное **нетривиальное** (поскольку правая часть отлична от нуля) решение.

Полученный результат можно легко обобщить на неравномерную сетку. Сформулируем его в виде теоремы.


**Теорема 1.** Пусть на отрезке $[a,b]$ определена гладкая функция $\varphi$ и введена (вообще говоря, неравномерная) сетка $G$. Тогда на сеточном шаблоне из $N+1$ точки (с фиксированной "текущей" точкой) можно построить *единственную* формулу численного дифференцирования, аппроксимирующую первую производную $\varphi'$ с порядком аппроксимации $N$.

Отметим, что при одинаковом общем количестве точек шаблоны могут отличаться расположением "текущей" точки. Например, и производная вперёд, и производная назад строятся по двум точкам, но в первом случае $i=0$, $j=1$, а во втором случае $i=1$, $j=0$, т.е. у этих формул **разные** двухточечные шаблоны.

Из доказательства теоремы также следует, что **минимальное число точек в шаблоне равно двум**, поскольку на коэффициенты $\alpha_k$ приходится накладывать как минимум два условия: $\sum\alpha_k=0$ и $\sum\alpha_k k=1$.

Напишем программу "**Калькулятор формул численного дифференцирования**", выдающую формулу заданного порядка аппроксимации $p$ на минимальном шаблоне. Шаблон будем характеризовать числами $i$ и $j$, как и выше.

Функция для вычисления коэффициентов:

````python
import numpy as np
from numpy.polynomial.polynomial import polyvander

def coeffs(p: int, i: int, j: int):
    template = [k for k in range(-i, j+1, 1)] # задаём шаблон

    A = polyvander(template, p) # строим матрицу Вандермонда
    A = A.transpose() # транспонируем матрицу

    f = np.zeros(p+1) # задаём правую часть
    f[1] = 1

    return np.linalg.solve(A, f)
````

Можно проверить работоспособность написанной функции на трёх известных нам примерах:

````python
coeffs(1, 0, 1) # производная вперёд
````

````python
coeffs(1, 1, 0) # производная назад
````

````python
coeffs(2, 1, 1) # центральная разность
````

Программа выдаёт правильные коэффициенты, начиная с самого левого узла ($k=-i$). Попробуем получить, например, формулу четвёртого порядка аппроксимации на шаблоне с $i=j=2$:

````python
coeffs(4, 2, 2) # формула O(h^4)
````

Теоретическое рассмотрение (см. задачу 1 ниже) показывает, что соответствующая формула имеет вид:

$$f'_m=\frac{f_{m-2}-8f_{m-1}+8f_{m+1}-f_{m+2}}{12h}.$$

Было бы удобнее и в программе получать коэффициенты формул в рациональном виде, а не в виде округлённых десятичных дробей. Кроме того, желательно добавить в начало программы проверку корректности задачи: формула может быть построена при выполнении условия $i+j=p$.

Усовершенствованная версия калькулятора:

````python
from fractions import Fraction

class TemplateError(Exception):
    pass

def FDC_calculator(p: int, i: int, j: int):
    if p != i + j:
        raise TemplateError('Заданный шаблон не соответствует требуемому порядку аппроксимации')

    # функция, преобразующая элементы массива в обыкновенные дроби
    func = np.vectorize(lambda x: str(Fraction(x).limit_denominator()))
    coeffs_frac = func(coeffs(p, i, j))

    return coeffs_frac

try:
    print(FDC_calculator(4, 2, 2)) # формула O(h^4)
except TemplateError as e:
    print(e)
````

# Задача 1

Методом неопределённых коэффициентов построить формулу для численного нахождения **первой** производной гладкой функции $\varphi$ в точке $x_m$ равномерной сетки по значениям функции в точках

$$x_{m-2},\quad x_{m-1},\quad x_{m},\quad x_{m+1},\quad x_{m+2}.$$

Оценить погрешность численного дифференцирования и оптимальный шаг сетки.

**Решение.** Вводим пять неопределённых коэффициентов (по числу точек) и составляем выражение для $f'_m$:

$$f'_m=\frac{Af_{m-2}+Bf_{m-1}+Cf_m +Df_{m+1}+Ef_{m+2}}{h}.$$

Раскладываем каждый член числителя по формуле Тейлора. Поскольку задействовано пять точек, порядок аппроксимации не может быть больше четырёх, т.е. **для определения коэффициентов** в формуле для $f'_m$ достаточно получить члены вплоть до $h^3$ (члены с $h^4$ точно не уйдут, каковы бы ни были коэффициенты). Значит, числитель достаточно разложить до $h^4$ (с учётом последующего деления на $h$). Однако по условию **нужно также оценить погрешность**, поэтому раскладывать будем до пятого порядка по $h$: тогда после нахождения коэффициентов мы сможем сразу определить константу, стоящую при $h^4$ (она необходима для оценки погрешности метода). Имеем (остаточные члены для краткости опускаем):

$$\begin{align}
  f_{m-2}&=\varphi(x_m-2h)=f_m-2hf'_m+\frac{(2h)^2}{2}f''_m-\frac{(2h)^3}{6}f'''_m+\frac{(2h)^4}{24}f^{(4)}_m-\frac{(2h)^5}{120}f^{(5)}_m;\\[0.2cm]
  f_{m-1}&=\varphi(x_m-h)=f_m-hf'_m+\frac{h^2}{2}f''_m-\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m-\frac{h^5}{120}f^{(5)}_m;\\[0.2cm]
  f_{m+1}&=\varphi(x_m+h)=f_m+hf'_m+\frac{h^2}{2}f''_m+\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m+\frac{h^5}{120}f^{(5)}_m;\\[0.2cm]
  f_{m+2}&=\varphi(x_m+2h)=f_m+2hf'_m+\frac{(2h)^2}{2}f''_m+\frac{(2h)^3}{6}f'''_m+\frac{(2h)^4}{24}f^{(4)}_m+\frac{(2h)^5}{120}f^{(5)}_m.
\end{align}$$

Подставляем эти разложения в формулу для $f'_m$. Группируем члены с одинаковыми степенями $h$:

$$\begin{align}
  f'_m&=\frac{1}{h}(A+B+C+D+E)f_m+(-2A-B+D+2E)f'_m\,+\\[0.2cm]
  &+\frac{h}{2}(4A+B+D+4E)f''_m+\frac{h^2}{6}(-8A-B+D+8E)f'''_m\,+\\[0.2cm]
  &+\frac{h^3}{24}(16A+B+D+16E)f^{(4)}_m+\frac{h^4}{120}(-32A-B+D+32E)f^{(5)}_m.
\end{align}$$

Сравнивая левую и правую части, составляем систему уравнений:

$$
  \begin{cases}
    A+B+C+D+E=0,\\
    -2A-B+D+2E=1,\\
    4A+B+D+4E=0,\\
    -8A-B+D+8E=0,\\
    16A+B+D+16E=0.
  \end{cases}
$$

Вычитая из пятого уравнения третье, находим:

$$12A+12E=0,\quad A=-E.$$

Подставляя это в третье уравнение, выясняем, что

$$B+D=0,\quad B=-D.$$

Тогда из первого уравнения видно, что $C=0$, а четвёртое и второе уравнения принимают вид:

$$-16A-2B=0,\quad -4A-2B=1.$$

Отсюда

$$A=\frac{1}{12},\quad B=-\frac{8}{12}.$$

Соответственно,

$$E=-\frac{1}{12},\quad D=\frac{8}{12}.$$

Итак, получаем следующую формулу:

$$f'_m=\frac{f_{m-2}-8f_{m-1}+8f_{m+1}-f_{m+2}}{12h}.$$

При этом первый нетривиальный член погрешности метода имеет вид:

$$\frac{h^4}{120}(-32A-B+D+32E)f^{(5)}_m=\frac{h^4}{30}f^{(5)}_m\leq \frac{M_5h^4}{30}\equiv\Delta^{(\text{m})}.$$

Оценка на погрешность округления следует из вида полученной формулы:

$$\Delta^{(\text{r})}=\frac{1+8+8+1}{12h}M_0\varepsilon_\text{маш}=\frac{3}{2h}M_0\varepsilon_\text{маш}.$$

Полная погрешность:

$$\Delta(h)=\Delta^{(\text{m})}+\Delta^{(\text{r})}=\frac{M_5h^4}{30}+\frac{3}{2h}M_0\varepsilon_\text{маш}.$$

Минимизируя $\Delta(h)$, находим оптимальный шаг:

$$\Delta'(h)=\frac{2M_5h^3}{15}-\frac{3}{2h^2}M_0\varepsilon_\text{маш}=0,\quad h_\text{opt}=\bigg(\frac{45M_0\varepsilon_\text{маш}}{4M_5}\bigg)^{1/5}.$$

Минимальная погрешность:

$$\Delta^\text{min}=\Delta(h_\text{opt})=\frac{M_0\varepsilon_\text{маш}}{h_\text{opt}}\bigg(\frac{M_5h^5_\text{opt}}{30M_0\varepsilon_\text{маш}}+\frac{3}{2}\bigg)=\frac{15M_0\varepsilon_\text{маш}}{8h_\text{opt}}.$$

# Задача 2

Оценить оптимальный шаг (равномерной) сетки и минимальную погрешность при численном дифференцировании функций $y(x)=\sin x$ и $y(x)=\sin 100x$ по формуле производной вперёд и по формуле четвёртого порядка аппроксимации. Рассмотреть случаи вычислений с одинарной и двойной точностью.

**Решение.** Погрешности обеих формул оценивались выше. Приведём полученные оценки ещё раз.

*   Производная вперёд:

$$f_m'=\frac{f_{m+1}-f_m}{h},\quad h_\text{opt}=2\sqrt{\frac{M_0\varepsilon_\text{маш}}{M_2}},\quad \Delta^{\text{min}}=2\sqrt{M_0M_2\varepsilon_\text{маш}}.$$

*   Формула четвёртого порядка аппроксимации:

$$f'_m=\frac{f_{m-2}-8f_{m-1}+8f_{m+1}-f_{m+2}}{12h},\quad h_\text{opt}=\bigg(\frac{45M_0\varepsilon_\text{маш}}{4M_5}\bigg)^{1/5},\quad \Delta^{\text{min}}=\frac{15M_0\varepsilon_\text{маш}}{8h_\text{opt}}.$$

Константы $M_0$, $M_2$ и $M_5$ зависят от рассматриваемой функции. В случае функции $y(x)=\sin x$ имеем:

$$M_0=M_2=M_5=1.$$

Если же $y(x)=\sin 100x$, то

$$M_0=1,\quad M_2=10^4,\quad M_5=10^{10}.$$

При вычислениях с одинарной точностью $\varepsilon_\text{маш}=6\cdot 10^{-8}$. Для двойной точности $\varepsilon_\text{маш}=10^{-16}$.

Подстановку констант в формулы осуществим с помощью программы.




````python
import numpy as np

def forward(M0: float, M2: float, eps: float):
    h_opt = 2 * np.sqrt(M0 * eps / M2)
    delta_min = 2 * np.sqrt(M0 * M2 * eps)

    return h_opt, delta_min

def fourth(M0: float, M5: float, eps: float):
    h_opt = (45 * M0 * eps / (4 * M5) ) ** (1/5)
    delta_min = 15 * M0 * eps / (8 * h_opt)

    return h_opt, delta_min

eps1 = 6 * 10 ** (-8) # одинарная точность
eps2 = 10 ** (-16) # двойная точность
````

````python
def calc(M0: float, M2: float, M5: float):
    h_opt, delta_min = forward(M0, M2, eps1)
    print("Производная вперёд, float: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))

    h_opt, delta_min = forward(M0, M2, eps2)
    print("Производная вперёд, double: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))

    h_opt, delta_min = fourth(M0, M5, eps1)
    print("Формула IV порядка, float: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))

    h_opt, delta_min = fourth(M0, M5, eps2)
    print("Формула IV порядка, double: h_opt={:.3e}, delta_min={:.3e}".format(h_opt, delta_min))
````

````python
print("y(x) = sin(x)")
calc(1, 1, 1)
````

````python
print("y(x) = sin(100x)")
calc(1, 10**4, 10**10)
````

Результаты, полученные в задаче, подтверждают сказанное выше: при использовании формул высокого порядка аппроксимации достаточно точный результат можно получить на относительно неподробных сетках. Оптимальный шаг численного дифференцирования по формуле четвёртого порядка аппроксимации даже в арифметике двойной точности оказался больше оптимального шага численного дифференцирования по формуле первого порядка как при использовании арифметики одинарной точности, так и при использовании арифметики двойной точности (т.е. для расчёта по формуле высокого порядка аппроксимации нужна менее подробная сетка). Минимальная ошибка численного дифференцирования тем меньше, чем больше порядок аппроксимации используемой формулы численного дифференцирования и чем больше памяти выделяется под хранение числа в арифметике (минимальная ошибка в арифметике двойной точности меньше, чем в арифметике одинарной точности).

# Вычисление производных высших порядков

Изложенный выше метод неопределённых коэффициентов очевидным образом обобщается на случай производных порядка $s$. Пусть, как и прежде, сеточный шаблон включает $i$ точек слева от $x_m$ и $j$ точек справа. Общий вид формулы численного дифференцирования:

$$f_m^{(s)}=\frac{1}{h^s}\sum\limits_{k=-i}^{j}\alpha_k \varphi(x_m+kh).$$

После подстановки разложения функции $\varphi$ по формуле Тейлора имеем:

$$\begin{align}
f_m^{(s)}=\frac{\varphi(x_m)}{h^s}\sum\limits_{k=-i}^j\alpha_k+\frac{\varphi'(x_m)}{h^{s-1}}\sum\limits_{k=-i}^j\alpha_kk\,&+\,...\,+\,\varphi^{(s)}(x_m)\sum\limits_{k=-i}^j\frac{\alpha_kk^s}{s!}+\\
&+\sum\limits_{n=s+1}^q\varphi^{(n)}(x_m)h^{n-s}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).
\end{align}$$

Налагаем $s+1$ условие:

$$\sum\limits_{k=-i}^j\alpha_k=0,\quad \sum\limits_{k=-i}^j\alpha_kk=0,\quad ... \quad \sum\limits_{k=-i}^j\alpha_kk^{s-1}=0,\quad \sum\limits_{k=-i}^j\frac{\alpha_kk^s}{s!}=1.$$

Получаем:

$$f_m^{(s)}=\varphi^{(s)}(x_m)+\sum\limits_{n=s+1}^q\varphi^{(n)}(x_m)h^{n-s}\sum\limits_{k=-i}^j\frac{\alpha_k k^n}{n!}+O(h^{q+1}).$$

Если требуется построить формулу $p$-го порядка аппроксимации, то нужно занулить ещё $p-1$ член в сумме по $n$ добавлением условий вида

$$\sum\limits_{k=-i}^j\alpha_k k^n=0\quad\text{для }n=s+1,...,s+p-1.$$

Таким образом, полное число условий на коэффициенты $\alpha_k$ оказывается равным $p+s$. Ясно, что матрица получившейся СЛАУ снова имеет вид матрицы Вандермонда (порядка $p+s$). Изменяется только вектор правой части: это вектор длины $p+s$, все компоненты которого, за исключением одной, нулевые; на $(s+1)$-м месте в этом векторе стоит $s!$.

Итак, мы приходим к результату, который, на первый взгляд, совершенно аналогичен теореме 1: если количество точек в шаблоне $N+1$ совпадает с комбинацией $p+s$, то на этом шаблоне можно построить единственную формулу численного дифференцирования $p$-ого порядка аппроксимации для $s$-й производной. Отсюда получаем связь: $p=N+1-s$. Однако нужно обратить внимание на один частный случай, в котором порядок аппроксимации построенной формулы оказывается на единицу выше, чем $N+1-s$. Речь идёт о формулах для производных **чётного порядка**, построенных на **симметричных** шаблонах.

Чтобы понять, почему в этом специальном случае порядок аппроксимации оказывается на единицу выше ожидаемого, необходимо заметить, что **при чётном порядке производной и симметричном шаблоне неопределённые коэффициенты тоже симметричны** (попарно равны), т.е. $\alpha_{k}=\alpha_{-k}$.

Действительно, пусть рассматриваемый шаблон симметричен, т.е. $i=j$ (откуда $N=2j$). Легко видеть, что тогда при **нечётных** $n$ всем условиям вида

$$\sum\limits_{k=-i}^j\alpha_k k^n=\sum\limits_{k=-j}^j\alpha_k k^n=0$$

можно удовлетворить, положив коэффициенты попарно равными: $\alpha_{j}=\alpha_{-j}$, $\alpha_{j-1}=\alpha_{-(j-1)}$ и т.д. Так можно поступить только в том случае, если порядок производной $s$ **чётен**: $s=2r$, иначе мы придём к противоречию с условием

$$\sum\limits_{k=-i}^j\frac{\alpha_kk^s}{s!}=\sum\limits_{k=-j}^j\frac{\alpha_kk^s}{s!}=1.$$

Выбрав коэффициенты попарно равными, мы переходим от СЛАУ из $N+1=2j+1$ уравнения с $2j+1$ неизвестной к СЛАУ меньшей размерности из $j+1$ уравнения (только с чётными $n$) с $j+1$ неизвестной ($\alpha_0$, ..., $\alpha_j$):

$$\sum\limits_{k=0}^j\alpha_k=0,\quad \sum\limits_{k=0}^j\alpha_kk^2=0,\quad ... \quad \sum\limits_{k=0}^j\alpha_kk^{2r-2}=0,\quad \sum\limits_{k=0}^j\frac{\alpha_kk^{2r}}{(2r)!}=\frac{1}{2},$$

$$\sum\limits_{k=0}^j\alpha_k k^n=0\quad\text{для }n=2r+2,...,2j.$$

Здесь учтена связь $p=N+1-s=2(j-r)+1$. Матрица СЛАУ меньшей размерности имеет вид

$$\begin{equation}
  \begin{pmatrix}
    1 & 1 & \dots & 1 & 1\\
    0 & 1 & \dots & (j-1)^2 & j^2 \\
    0 & 1 & \dots & (j-1)^4 & j^4 \\
    \vdots & \vdots & \ddots & \vdots & \vdots\\
    0 & 1 & \dots & (j-1)^{2j} & j^{2j}
  \end{pmatrix}
\end{equation}$$

и, очевидно, не является вырожденной. Поскольку, кроме того, правая часть СЛАУ меньшей размерности отлична от нуля, существует единственное нетривиальное решение. Таким образом, коэффициенты $\alpha_0$, ..., $\alpha_j$ определены оставшимися уравнениями однозначно, а с учётом попарных равенств однозначно определены все коэффициенты $\alpha_k$ (в количестве $2j+1$).

В силу единственности решения исходной СЛАУ (с матрицей Вандермонда) найденные указанным способом коэффициенты $\alpha_k$ совпадают с теми, которые мы получили бы, не делая исходного предположения о симметричности коэффициентов.

Итак, мы увидели, что **в случае симметричного шаблона и чётного порядка производной искомые коэффициенты симметричны (попарно равны)**. Заметим теперь, что согласно **общей** формуле для $f_m^{(s)}$ первый нетривиальный член погрешности метода имеет вид:

$$\varphi^{(s+p)}(x_m)h^{p}\sum\limits_{k=-i}^j\frac{\alpha_k k^{s+p}}{(s+p)!}.$$

Но в интересующем нас частном случае $s+p=2j+1$, т.е. рассматриваемый член включает сумму

$$\sum\limits_{k=-j}^j\alpha_k k^{2j+1}.$$

Она равна нулю в силу существующих между коэффициентами попарных равенств (выше мы уже отмечали это для всех сумм с нечётными $n$). Таким образом, хотя ожидаемый порядок аппроксимации есть $p=N+1-s$, симметричность коэффициентов автоматически обеспечивает равенство нулю члена погрешности с $h^p$, так что реальный порядок аппроксимации повышается до $N+2-s$. Иначе говоря, благодаря специальному виду шаблона, удаётся удовлетворить не $p+s$ уравнениям на коэффициенты, а $p+s+1$ уравнению.

Аналогичными рассуждениями можно показать, что при **нечётном** порядке производной формула, построенная на симметричном шаблоне, имеет **антисимметричные коэффициенты**: $\alpha_k=-\alpha_{-k}$ (откуда, между прочим, $\alpha_0=0$). Но легко убедиться, что к повышению порядка аппроксимации на единицу это не приводит (см. задачу 1, в которой приведён пример как раз такого случая).

Итак, мы доказали следующую теорему.

**Теорема 2.** Пусть на отрезке $[a,b]$ определена гладкая функция $\varphi$ и введена (вообще говоря, неравномерная) сетка $G$. Тогда на сеточном шаблоне из $N+1$ точки (с фиксированной "текущей" точкой) можно построить *единственную* формулу численного дифференцирования, аппроксимирующую производную $\varphi^{(s)}$. При этом порядок аппроксимации $p$ оказывается равным $N+1-s$. Если шаблон симметричен, а порядок производной $s$ чётен, то порядок аппроксимации повышается до $N+2-s$.

Из сказанного следует простое правило, которым полезно пользоваться при решении задач: **если из числа точек в шаблоне вычесть порядок производной, то получится порядок аппроксимации формулы**, которую можно построить на данном шаблоне для данной производной. Нужно только помнить, что при симметричном шаблоне и чётном порядке производной найденный порядок аппроксимации надо увеличить на единицу.

Из теоремы 2 также вытекает, что **минимальный шаблон, необходимый для аппроксимации $s$-й производной, должен включать $s+1$ точку**.

# Задача 3

Методом неопределённых коэффициентов построить формулу для численного нахождения **второй** производной гладкой функции $\varphi$ в точке $x_m$ равномерной сетки по значениям функции в точках

$$x_{m-1},\quad x_{m},\quad x_{m+1}.$$

Оценить погрешность численного дифференцирования и оптимальный шаг сетки.

**Решение.** Эта задача аналогична задаче 1. Заметим, что рассматриваемый шаблон симметричен (и включает три точки), а порядок производной чётен (и равен двум). Поэтому порядок аппроксимации будет $3-2+1=2$ (а не $3-2=1$). Вводим неопределённые коэффициенты:

$$f''_m=\frac{Af_{m-1}+Bf_{m}+Cf_{m+1}}{h^2}.$$

Поскольку формула имеет второй порядок аппроксимации, первый нетривиальный член погрешности будет содержать $h^2$. Для определения коэффициентов этот член удерживать не требуется, но нам надо также найти ошибку метода, т.е. необходимо знать коэффициент при $h^2$. Поэтому слагаемые числителя будем раскладывать по формуле Тейлора вплоть до $h^4$: тогда после деления на $h^2$ получим как раз члены до $h^2$ включительно. Имеем:

$$f_{m\pm 1}=\varphi(x_m\pm h)=f_m\pm hf'_m+\frac{h^2}{2}f''_m\pm\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m;$$

$$f''_m=\frac{1}{h^2}(A+B+C)f_m+\frac{1}{h}(-A+C)f'_m+\frac{1}{2}(A+C)f''_m+\frac{h}{6}(-A+C)f'''_m+\frac{h^2}{24}(A+C)f^{(4)}_m.$$

Получаем три условия на коэффициенты:

$$\begin{equation}
  \begin{cases}
    A+B+C=0,\\
    -A+C=0,\\
    A+C=2.\\
  \end{cases}
\end{equation}$$

Отсюда

$$A=C=1,\quad B=-2.$$

Видно, что коэффициент при $f'''_m$ занулился автоматически (поэтому порядок аппроксимации и повысился на единицу). Ошибка метода составляет

$$\Delta^{(\text{m})}=\frac{M_4h^2}{12}.$$

Искомая формула имеет вид:

$$f''_m=\frac{f_{m+1}-2f_m+f_{m-1}}{h^2},$$

так что ошибка округления есть

$$\Delta^{(\text{r})}=\frac{1+2+1}{h^2}M_0\varepsilon_\text{маш}=\frac{4}{h^2}M_0\varepsilon_\text{маш}.$$

Полная погрешность:

$$\Delta(h)=\Delta^{(\text{m})}+\Delta^{(\text{r})}=\frac{M_4h^2}{12}+\frac{4}{h^2}M_0\varepsilon_\text{маш}.$$

Минимизируя $\Delta(h)$, находим оптимальный шаг:

$$\Delta'(h)=\frac{M_4h}{6}-\frac{8}{h^3}M_0\varepsilon_\text{маш}=0,\quad h_\text{opt}=\bigg(\frac{48M_0\varepsilon_\text{маш}}{M_4}\bigg)^{1/4}.$$

Для функции $y=f(x)$, заданной таблично, вычислить значение **второй** производной в точке $x=3$ с максимально возможной точностью.

**Решение.** Имеем равномерную сетку с $h=0,5$. Шаблон пятиточечный, несимметричный ($x_m=3$, $i=4$, $j=0$). Значит, максимальный порядок аппроксимации есть $5-2=3$. Вводим неопределённые коэффициенты:

$$f''_m=\frac{Af_{m-4}+Bf_{m-3}+Cf_{m-2}+Df_{m-1}+Ef_{m}}{h^2}.$$

Поскольку в знаменателе стоит $h^2$, **для определения коэффициентов** в разложениях по формуле Тейлора достаточно удерживать первые четыре члена (ошибку метода оценивать не требуется, поэтому члены пятого порядка нам не нужны). Имеем:

$$\begin{align}
  f_{m-4}&=f_m-4hf'_m+\frac{(4h)^2}{2}f''_m-\frac{(4h)^3}{6}f'''_m+\frac{(4h)^4}{24}f^{(4)}_m,\\[0.2cm]
  f_{m-3}&=f_m-3hf'_m+\frac{(3h)^2}{2}f''_m-\frac{(3h)^3}{6}f'''_m+\frac{(3h)^4}{24}f^{(4)}_m,\\[0.2cm]
  f_{m-2}&=f_m-2hf'_m+\frac{(2h)^2}{2}f''_m-\frac{(2h)^3}{6}f'''_m+\frac{(2h)^4}{24}f^{(4)}_m.\\[0.2cm]
  f_{m-1}&=f_m-hf'_m+\frac{h^2}{2}f''_m-\frac{h^3}{6}f'''_m+\frac{h^4}{24}f^{(4)}_m.
\end{align}$$

Подставляем эти разложения в формулу для $f''_m$. Группируем члены с одинаковыми степенями $h$:

$$\begin{align}
  f''_m&=\frac{1}{h^2}(A+B+C+D+E)f_m+\frac{1}{h}(-4A-3B-2C-D)f'_m\,+\\[0.2cm]

Сравнивая левую и правую части, составляем систему уравнений:

## Реализация численных методов

Библиотека ```numpy``` содержит несколько функций, которые можно использовать для вычисления конечно-разностных аналогов производных.

Функция ```diff``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.diff.html)) вычисляет разности между соседними элементами поданного на вход массива, возвращая массив **на единицу меньшей длины**. Поэтому её можно использовать для быстрого вычисления производной вперёд. Пример для линейной функции $y(x)=2x+3$:

````python
import numpy as np
import matplotlib.pyplot as plt

x, h = np.linspace(0, 20, 11, retstep=True) # дополнительно возвращает шаг сетки
y = 2 * x + 3
y_deriv = np.diff(y) / h # делим на шаг сетки

print(y_deriv, len(y_deriv))
````

Чтобы получить массив той же длины, что и исходный, необходимо задать один из необязательных аргументов ```prepend``` (элемент, добавляемый в начало **перед вычислением разностей**) или ```append``` (элемент, добавляемый в конец **перед вычислением разностей**):

````python
last = 2 * 20 + 3

# для примера добавляем last+h, чтобы вставленный элемент был заметен
y_deriv_app = np.diff(y, append=last+h) / h

print(y_deriv_app, len(y_deriv_app))
````

С помощью функции ```diff``` можно вычислять **разности высших порядков**. Например, вторые разности определяются как разности соседних элементов в массиве первых разностей. Иначе говоря, если

$$\delta_i^{(1)} = f_{i+1}-f_i,$$

то

$$\delta_i^{(2)}=\delta_{i+1}^{(1)}-\delta_{i}^{(1)}=f_{i+2}-2f_{i+1}+f_i.$$

Нетрудно заметить, что это числитель формулы, аппроксимирующей вторую производную (ср. с задачей 3). Поэтому делением на $h^2$ получаем массив приближённых значений второй производной (в данном примере они совпадают с точными значениями):

````python
y_deriv2 = np.diff(y, n=2) / (h**2)

print(y_deriv2, len(y_deriv2))
````

Заметим, что при сопоставлении этого массива с сеткой нужно сдвинуть индекс аргумента, связав разность $\delta_i^{(2)}$ с ($i-1$)-м узлом, поскольку в числителе формулы из задачи 3 стоит всё-таки $f_{m+1}-2f_m+f_{m-1}$. Естественно, что чем выше порядок разности, тем сильнее укорачивается исходный массив.

Пример использования функции ```diff``` с визуализацией:

````python
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
N_w = IntSlider(min=10, max=500, step=10, value=50)

def sin_diff(N: int):
    x, h = np.linspace(0, 2*np.pi, N, retstep=True)
    y = np.sin(x)
    sin_d = np.diff(y) / h
    sin_dd = np.diff(y, n=2) / h ** 2

    fig, ax = plt.subplots()
    ax.scatter(x[:N-1], sin_d, s=10, label="$y'(x)=\\cos x$") # не удлиняем массив разностей, а укорачиваем массив аргументов
    ax.scatter(x[1:N-1], sin_dd, s=10, label="$y''(x)=-\\sin x$") # массив сдвинут и укорочен

    ax.legend()

    x_plot = np.linspace(0, 2*np.pi, 1000) # сетка для графиков точных производных
    ax.plot(x_plot, np.cos(x_plot))
    ax.plot(x_plot, -np.sin(x_plot))

    plt.title("Производные функции $y(x)=\\sin x$ (с помощью diff)")
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.grid()
    plt.show()

interact(sin_diff, N=N_w);
````

Существует также функция ```ediff1d``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.ediff1d.html)), представляющая собой, по сути, усечённую версию функции ```diff```: она тоже вычисляет разности между последовательными элементами массива, но с её помощью можно получить только разности первого порядка. Кроме того, отсутствует возможность указать ось массива, вдоль которой должны вычисляться разности, поскольку массив, подаваемый на вход функции ```ediff1d```, автоматически приводится к одномерному. В случае же функции ```diff``` можно выбрать ось, задав необязательный аргумент ```axis```. В силу перечисленных причин функция ```ediff1d```, очевидно, малополезна.

Более удобной является функция ```gradient``` ([документация](https://numpy.org/doc/stable/reference/generated/numpy.gradient.html)). Она возвращает массив конечных разностей **той же длины, что и исходный**, причём расчёт ведётся по формуле центральной разности. Точнее: по значениям заданного массива $f_i$ рассчитывается величина $(f_{i+1}-f_{i-1})/2$. На границах массива, где применение центральной разности невозможно, она автоматически заменяется на одностороннюю разность (слева — на числитель формулы производной вперёд, справа — на числитель формулы производной назад). Сказанное легко проверить на простом примере:

````python
g = [5., 7., 8., 12., 4., 3., 0., -1.]

print(g)
print(np.gradient(g).tolist())
````

Таким образом, делением элементов выходного массива функции ```gradient``` на $h$ (в случае равномерной сетки) мы получаем массив значений производной, приближенной со вторым порядком аппроксимации (во всех точках, кроме граничных). Удобно, что сигнатура функции предусматривает возможность автоматического деления. Если сетка равномерная, то достаточно просто указать величину шага. Будем считать, например, что значения функции $g$ из предыдущей ячейки отнесены к равномерной сетке с шагом $h=2$. Тогда приближение для производной можно получить так:

````python
g_d = np.gradient(g, 2)
g_d.tolist()
````

Нетрудно проверить, что вычисления выполнены в точности по формуле центральной разности

$$f'_m=\frac{f_{m+1}-f_{m-1}}{2h}$$

(для всех значений, исключая граничные).

Если же сетка неравномерная, то для автоматического деления на (переменный) шаг необходимо вторым аргументом передать в функцию ```gradient``` массив из узлов используемой сетки. Разумеется, так же можно поступить и в случае равномерной сетки. Иначе говоря, численное дифференцирование с помощью функции ```gradient``` предельно упрощается: **в функцию надо передать массив значений и массив аргументов** (узлов сетки).

Более того, если требуется получить второй порядок аппроксимации и для граничных точек, можно указать необязательный аргумент ```edge_order=2``` (по умолчанию он равен 1). В этом случае значения производных в граничных точках будут вычисляться по односторонним формулам второго порядка:

$$f'_m=\frac{-3f_m + 4f_{m+1}-f_{m+2}}{2h},\quad f'_m=\frac{f_{m-2} - 4f_{m-1}+3f_{m}}{2h}.$$

Например:

````python
x = np.array([0, 1, 2, 3, 4])
f = x**2

print(np.gradient(f, edge_order=1))
print(np.gradient(f, edge_order=2))
````

Вычислим снова производные синуса с использованием функции ```gradient```:

````python
# создаём ползунок заново, чтобы не менять предыдущий график
N_again = IntSlider(min=10, max=500, step=10, value=50)

def sin_gradient(N: int):
    x = np.linspace(0, 2*np.pi, N)
    y = np.sin(x)

    sin_d = np.gradient(y, x, edge_order=2)
    sin_dd = np.gradient(sin_d, x, edge_order=2)

    fig, ax = plt.subplots()
    ax.scatter(x, sin_d, s=10, c='b', label="$y'(x)=\\cos x$")
    ax.scatter(x, sin_dd, s=10, c='r', label="$y''(x)=-\\sin x$",)

    ax.legend()

    x_plot = np.linspace(0, 2*np.pi, 1000) # сетка для графиков точных производных
    ax.plot(x_plot, np.cos(x_plot), c='b')
    ax.plot(x_plot, -np.sin(x_plot), c='r')

    plt.title("Производные функции $y(x)=\\sin x$ (с помощью gradient)")
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.grid()
    plt.show()

interact(sin_gradient, N=N_again);
````

Хорошо видно, что при одинаковых (и небольших) $N$ этот график и график, построенный выше с использованием функции ```diff```, отличаются: ```gradient``` даёт точки, подходящие к кривым ближе.

**Упражнение.** Из графика при малых $N$ ясно видно, что значения второй производной (красные точки) в граничных узлах вычисляются с меньшей точностью, чем в остальных узлах, несмотря на то, что ```edge_order=2```. Почему так получается? Как это исправить?

В заключение это раздела сделаем ряд важных **замечаний**:

1. Если при решении некоторой задачи часть уравнений аппроксимирована с порядком $p_1$, а другая часть уравнений — с порядком $p_2$, то итоговый порядок аппроксимации будет равен $\min (p_1,p_2)$. Если производная в граничных точках вычисляется с первым порядком аппроксимации, а во внутренних точках — со вторым, то общий порядок аппроксимации будет равен одному. Если же вычислять производные в граничных точках со вторым порядком аппроксимации, то итоговый порядок аппроксимации будет равен двум; это грамотный вариант с точки зрения вычислительной математики.

2. Сказанное в первом пункте ещё раз подтверждает, что использование готовых решений в качестве "чёрных ящиков" без понимания деталей может привести как минимум к неожиданным результатам, а как максимум — к серьёзным ошибкам. В рамках этого семинара обсуждался способ проверки порядка сходимости (см. выше) при известном аналитическом решении задачи. Другие варианты будут предложены в семинаре 11.

3. Подчеркнём, что оценка правильности написанной программы "на глаз" по графикам недостаточна. Такой анализ программы может выявить только грубые ошибки. Необходимо использовать инструментарий тестов на порядок сходимости на сгущающихся сетках. Подробно этот вопрос будет изучаться позднее.

## Символьные вычисления

До сих пор мы обсуждали задачу численного дифференцирования в её классической (для вычислительной математики) постановке. Можно, однако, изначально рассматривать другой подход к вычислению производных с помощью компьютера.

Мы уже говорили, что задача **аналитического** дифференцирования сводится к последовательному применению **ряда правил**, по которым одни функции заменяются другими. Дифференцируя, например, выражение $x^2$, мы заменяем его на $2x$. Мы не выполняем при этом никаких арифметических действий (не вычисляем конечные разности) и даже не особенно вникаем в тот факт, что $x^2$ есть функция, обладающая какими-то свойствами. Вместо этого мы работаем с выражением $x^2$ как с некоторым **символом**, который, согласно правилам, должен быть заменён **другим символом** $2x$. Сталкиваясь с более сложной функцией, например, $x\sin (x^2)$, мы рассматриваем её как **комбинацию (суперпозицию) символов** и тоже применяем некоторые правила замены одних символов другими. В упомянутом примере цепочка замен может быть записана следующим образом:

$$\begin{gather}
  \big[x\sin (x^2)\big]'\quad\Rightarrow\quad(x)' \sin (x^2)+x\big[\sin (x^2)\big]'\quad\Rightarrow\quad\sin (x^2)+x\cos(x^2)\cdot (x^2)'\quad\Rightarrow\\
  \Rightarrow\quad\sin (x^2)+x\cos(x^2)\cdot 2x \quad\Rightarrow\quad \sin (x^2)+2x^2\cos(x^2).
\end{gather}$$

Очевидно, что аналитический способ вычисления производных можно алгоритмизовать и реализовать в виде некоторой программы. При исполнении такой программы компьютер не будет выполнять арифметические операции, относящиеся непосредственно к дифференцированию. Вместо этого он будет анализировать заданное выражение, составленное в рамках некоторого синтаксиса из определённых **символов**, и преобразовывать (упрощать) его в соответствии с системой правил. Другими словами, он будет дифференцировать функции так же, как их дифференцируют в математическом анализе.

Описанный подход составляет суть так называемых **символьных вычислений**. Они могут применяться не только в задачах дифференцирования, но и во многих других математических задачах. Отличительная особенность символьных вычислений заключается в том, что **компьютер работает с математическими объектами формально**, выполняя над ними не арифметические операции, а заранее запрограммированные процедуры преобразований (точнее — компьютер выполняет то одно, то другое).

Очевидным достоинством символьных  вычислений является **аналитическая точность результата** (нет ошибок метода и ошибок округления). Однако при этом **теряется основное преимущество компьютерных расчётов — возможность получить решение в тех случаях, когда его не удаётся найти аналитически**. По существу, при выполнении символьных вычислений компьютер превращается в математика, который действует теми же аналитическими методами, что и математик-человек, но имеет гораздо большую производительность, знает наизусть все первообразные, все специальные функции, не допускает ошибок, описок и т.д. Однако задачи, не имеющие аналитического решения, оказываются для такого компьютера-математика недоступными.

Из сказанного следует, что сфера применения символьных вычислений не так широка, как сфера применения численных методов. Тем не менее существуют специальные пакеты программ для символьных вычислений (их называют **системами компьютерной алгебры**), которые в ряде случаев оказываются очень полезными. Чаще всего в таких программах комбинируются **и аналитические, и численные возможности**. Известным примером служит система компьютерной алгебры Mathematica.

В Python символьные вычисления можно выполнять с помощью средств достаточно мощной библиотеки ```sympy```. Подробно изучить её читатель сможет с помощью [документации](https://docs.sympy.org/latest/index.html). Мы приведём только несколько примеров с пояснениями в контексте дифференцирования функций.

Для выполнения любых символьных вычислений и преобразований необходимо предварительно задать **символьные переменные**. Это можно сделать, например, с помощью функции ```Symbol``` (конструктора одноимённого класса). Она преобразовывает поданную на вход строку в символьную переменную — объект класса ```Symbol```:

````python
from sympy import Symbol

x = Symbol('x') # переменная x
type(x)
````

Если требуется задать сразу несколько переменных, то можно применить функцию ```symbols``` (или её аналог — функцию ```var```). При этом отдельные переменные во входных строках должны разделяться пробелами или запятыми. Можно указывать типы переменных, накладывать на них ограничения и снабжать их индексами:

````python
from sympy import symbols, var

a, b, t = symbols('a b t') # переменные a, b и t
n = symbols('n', integer=True) # целочисленная переменная n
y1, z32 = var('y_1, z^3_2', positive=True) # положительнозначные переменные y_1 и z^3_2
````

Буквы латинского и греческого алфавитов, часто использующиеся в формулах, можно импортировать из модуля ```symby.abc```. Отметим, что имена символьной переменной и переменой, использующейся в коде, не обязаны совпадать:

````python
from sympy.abc import phi, omega

f = phi # переменная фи
w = omega # переменная омега
````

Алгебраические выражения можно составлять из объявленных переменных, пользуясь обычным синтаксисом языка ```Python```. Например:

````python
expr1 = 2*a / (a - 5) - 5 / (a + 5) + 2*a**2 / (25 - a**2)
expr2 = (2*b**2 - 3*b + 4) / (b-1)**2 - (b-2) / (b-1)
````

Очень удобно организован вывод символьных выражений — в ipynb-файле они автоматически преобразуются к TeX-формуле:

````python
expr1
````

Можно вызвать и встроенную функцию ```print```. Тогда математическая формула будет напечатана в строку из "обычных" символов:

````python
print(expr2)
````

Если выражение содержит элементарные функции или константы, то их нужно предварительно импортировать:

````python
from sympy import sin, cos, pi

expr3 = (sin(w*t + f))**2 + (cos(w*t + f + 2*pi*n))**2
expr3
````

Обратим внимание на то, что выражение ```expr3``` было частично упрощено: ```sympy``` автоматически отбросил из аргумента косинуса слагаемое $2\pi n$. Если бы заданная выше переменная $n$ не являлась целочисленной, это преобразование не было бы выполнено. Можно упрощать выражения и целенаправленно — с помощью функции ```simplify```:



````python
from sympy import simplify

simplify(expr1)
````

````python
simplify(expr2)
````

````python
simplify(expr3)
````

Для символьного дифференцирования заданных выражений применяется функция ```diff```. Она принимает на вход функцию, производную которой надо вычислить, и аргумент (символ), по которому требуется продифференцировать:

````python
from sympy import diff

y = sin(x)
diff(y, x)
````

Если требуется найти производные высоких порядков, то аргумент нужно указать несколько раз — столько, сколько раз надо продифференцировать:

````python
diff(y, x, x)
````

Точно так же можно находить производные функций нескольких переменных, в том числе и смешанные:

````python
g = sin(x) * sin (y1) * z32**2 / (x * y1)
g
````

````python
diff(g, x, z32, y1, y1)
````

Если нужно создать **символ производной** некоторой функции, а не выполнять дифференцирование (для некоторых операций это бывает необходимо), то можно воспользоваться конструктором класса ```Derivative```:

````python
from sympy import Derivative

y_d = Derivative(y, x)
y_d
````

````python
g_d = Derivative(g, x, z32, y1, y1)
g_d
````

Из подобного символа можно получить сам результат дифференцирования при помощи метода ```doit```:

````python
y_d.doit()
````

К результату дифференцирования (полученному как с помощью ```diff```, так и с помощью ```doit```) можно применить метод ```evalf``` для вычисления значения производной в конкретной точке:

````python
y_d.doit().evalf(subs={x: 0})
````

Большое преимущество библиотеки ```sympy``` по сравнению с другими системами компьютерной алгебры состоит в том, что **она не существует обособленно, а реализована именно как библиотека**, и поэтому может взаимодействовать с остальным кодом. В частности, результаты, полученные в ```sympy```, после некоторого преобразования можно обрабатывать средствами библиотеки ```numpy```. Покажем, как это делается, на примере дифференцирования.

Зададим какую-нибудь сложную функцию:

$$f(x)=\Bigg(\frac{\arcsin\sqrt{x}+\cos^2(3x-1)}{\pi^{x^3}+\ln^2\big(1+e^x\big)}\Bigg)^{x^2\text{ch}\,x}.$$

Составим её символьное выражение:

````python
from sympy import asin, sqrt, E, ln, cosh

numerator = asin(sqrt(x)) + (cos(3*x-1))**2
denominator = pi**(x**3) + (ln(1+E**x))**2
dimension = x**2 * cosh(x)

f = (numerator / denominator) ** dimension
f
````

Выполним дифференцирование:

````python
f_d = diff(f, x)
f_d
````

Результат получился весьма громоздким. Применение функции ```simplify``` в данном случае помогает мало: обработка выражения занимает слишком много времени. Однако результат можно и не упрощать, а просто преобразовать в векторизованную функцию ```numpy``` (например, если в дальнейшем с этой производной требуется выполнять какие-то расчёты). Для этого сначала с помощью команды ```lambdify``` получаем лямбда-функцию, а затем стандартным образом выполняем её векторизацию:

````python
from sympy import lambdify
import numpy as np

f_deriv = np.vectorize(lambdify(x, f_d, 'numpy'))
````

Теперь с этой функцией можно обычным (для ```numpy```) способом выполнять любые расчёты. Построим, например, график найденной производной. Для сравнения запрограммируем также исходную функцию и продифференцируем её численно:

````python
def composed_func(x: float):
    numerator = np.arcsin(np.sqrt(x)) + (np.cos(3*x - 1))**2
    denominator = np.power(np.pi, x**3) + (np.log(1 + np.exp(x)))**2
    dimension = x**2 * np.cosh(x)

    return np.power(numerator/denominator, dimension)
````

````python
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# ползунок для интерактивного графика (виджета)
N_w = IntSlider(min=10, max=500, step=10, value=20)

def cf_deriv(f_deriv, N: int):
    f = np.vectorize(composed_func)

    x = np.linspace(0.0, 0.5, N)
    y = f(x)
    y_deriv = np.gradient(y, x) # численное дифференцирование

    grid = np.linspace(0.0, 0.5, 1000) # сетка для производной, найденной символьно
    y_d = f_deriv(grid) # производная, найденная символьно

    fig, ax = plt.subplots()
    ax.scatter(x, y_deriv, c='b', s=10)
    ax.plot(grid, y_d, c='r', zorder=0)

    plt.title("Производная функции $y=f(x)$")
    plt.xlabel("$x$")
    plt.ylabel("$f(x)$")

    plt.grid()
    plt.show()

interact(lambda N: cf_deriv(f_deriv, N), N=N_w);
````

Таким образом, сочетание средств библиотек ```numpy``` и ```sympy``` даёт нам в руки довольно мощный инструмент аналитического дифференцирования на компьютере. Другое дело, что применимость этого инструмента очень ограничена: символьные вычисления нельзя использовать в весьма распространённом случае таблично заданных функций. Кроме того, очевидно, что невозможно аналитически продифференцироать неизвестную функцию. Построить же конечно-разностную аппроксимацию производной в общем виде (не конкретизируя значения массива) можно всегда. Именно этим и объясняется ведущая роль численных методов дифференцирования. Как мы увидим в дальнейшем, они широко используются при построении разностных схем.
